# Statistics Lecture Notes

## Relationships Between Two Variables

---

# 1. Introduction: Moving From One Variable to Two Variables

Welcome back.

So far in our statistics sessions, we focused on **a single data variable**.

For a single variable we learned how to compute:

* Measures of **central tendency** (mean, median)
* Measures of **dispersion** (variance, standard deviation)
* **Skewness**
* How to **fit probability distributions**

Now we move to the **next step in statistical analysis**.

Instead of studying **one variable**, we will start studying **relationships between two variables**.

---

## Example

Suppose I give you two data series such as:

* Price–earnings ratios and interest rates
* Earnings growth and GDP growth
* Stock returns and market returns

Now the question becomes:

**Do these variables move together?**

Or more specifically:

* Do they move **in the same direction**?
* Do they move **in opposite directions**?
* Or are they **unrelated**?

---

# 2. If Two Variables Are Linked

If you find a relationship between two variables, you can investigate further.

There are several follow-up questions we can ask.

---

## 1. Are There Time Leads or Lags?

Sometimes one variable moves **before** another.

For example:

* Interest rates might move first
* Stock prices might react later

So we ask:

* Does **Variable A move first**?
* Does **Variable B follow later**?

---

## 2. Correlation vs Causation

This is one of the **most important ideas in statistics**.

If two variables move together, we say they are **correlated**.

But that **does not mean** one variable **causes** the other.

Correlation simply means:

> There is a statistical relationship.

Causation means:

> One variable actually **causes** the other to change.

Proving causation requires **additional tests and deeper analysis**.

---

## 3. Prediction

In finance and investing, the real goal is often **prediction**.

If two variables are linked, perhaps we can use one to **predict the other**.

For example:

If interest rates predict stock returns, an investor might use that relationship to **make investment decisions**.

---

# 3. First Tool: Scatter Plot

Whenever you have **two variables**, the best starting point is a **scatter plot**.

---

## What Is a Scatter Plot?

A scatter plot is simply a **graph of paired observations**.

Suppose we call our variables:

* **X**
* **Y**

And suppose we have **10 observations**.

Each observation consists of:

[
(X_i, Y_i)
]

Example:

| Day | X  | Y |
| --- | -- | - |
| 1   | 14 | 1 |
| 2   | 16 | 3 |
| 3   | 11 | 2 |

Each pair becomes **one point on the graph**.

---

## Axes Convention

Typically:

* **X-axis** → X variable
* **Y-axis** → Y variable

In statistics:

* X is often called the **independent variable**
* Y is called the **dependent variable**

But initially we simply treat them as **two variables**.

---

## Interpretation

By looking at the scatter plot we can visually see:

* Do points move **upward together**?
* Do they move **in opposite directions**?
* Are they **randomly scattered**?

This gives a **first visual indication of a relationship**.

---

# 4. Correlation Coefficient

Visual inspection is useful, but we also want a **numerical measure**.

The most widely used measure is the **Pearson correlation coefficient**.

---

## Intuition Behind Correlation

For each observation we measure:

* How far **Y** is from its mean
* How far **X** is from its mean

Then we examine whether **these deviations move together**.

---

### Case 1: Positive Relationship

If:

* Y is above its average
* X is also above its average

then the variables move **together**.

---

### Case 2: Negative Relationship

If:

* Y is above its average
* X is below its average

then the variables move **in opposite directions**.

---

# 5. Range of the Correlation Coefficient

A correlation coefficient is **standardized**, meaning it always lies between:

[
-1 \leq \rho \leq +1
]

---

### Perfect Positive Correlation

[
\rho = +1
]

Both variables move **together perfectly**.

If one increases, the other increases in **exact proportion**.

---

### Perfect Negative Correlation

[
\rho = -1
]

Variables move **in opposite directions perfectly**.

When one increases, the other decreases proportionally.

---

### Zero Correlation

[
\rho = 0
]

No relationship exists between the variables.

---

### Intermediate Values

* Positive correlation → variables move together
* Negative correlation → variables move inversely

---

# 6. Covariance

Another measure of co-movement is **covariance**.

Covariance uses the **same idea** as correlation.

It examines:

* deviations from the mean
* whether the deviations move together

---

## Key Difference

Covariance is **not standardized**.

So it can take **any numerical value**.

---

### Example

Suppose we measure how **IBM stock returns move with the market**.

If I say:

Correlation = **0.22**

You immediately know:

* It is **positive**
* The relationship is **weak to moderate**

---

But if I say:

Covariance = **37%**

Your reaction is:

Is 37 large or small?

Because covariance **does not have fixed bounds**.

---

# 7. Danger: Spurious Correlations

Finance has a major problem.

We have **too much data**.

Thousands of variables.

Data going back **100 years or more**.

If you compute correlations between **every pair of variables**, you will inevitably find **relationships that are purely accidental**.

---

## Example: Super Bowl Indicator

Some studies claim stock market returns depend on which conference wins the **Super Bowl**.

Specifically:

* If a team from the **NFC wins → stocks rise**
* If a team from the **AFC wins → stocks fall**

The correlation appears real in historical data.

But there is **no logical economic reason** for this relationship.

This is a **spurious correlation**.

---

# 8. Correlation Does Not Imply Causation

Consider another example.

Suppose we find a strong correlation between:

* household wealth
* spending on organic food

Rich households tend to buy **Organic food**.

But that does **not mean buying organic food makes you rich**.

The causation likely runs the other way:

Wealth → ability to buy organic food.

---

# 9. Data Is Always From the Past

Another limitation:

Statistical relationships are based on **historical data**.

But investment decisions concern the **future**.

So when we use past correlations, we are assuming:

> The future will resemble the past.

That assumption may be **dangerous** if relationships change over time.

---

# 10. Regression Analysis

Let us return to the scatter plot.

Instead of simply plotting points, we can draw a **best-fit line** through the data.

This is called a **regression line**.

---

## Ordinary Least Squares (OLS)

The most common method for fitting the line is:

**Ordinary Least Squares (OLS)**.

The idea is simple.

For each observation, we measure:

Distance between the point and the regression line.

Then we **square those distances** and add them together.

OLS chooses the line that **minimizes the sum of squared distances**.

---

# 11. Regression Equation

A regression line is usually written as:

[
Y = a + bX
]

Where:

* **b** = slope
* **a** = intercept

---

## Slope

The slope shows **how Y changes when X changes**.

Example:

Slope = **−1.1**

Meaning:

If X increases by **1 unit**, Y decreases by **1.1 units**.

So the relationship is **negative**.

---

## Intercept

The intercept is the value of **Y when X = 0**.

Example:

Intercept = **14**

So if X equals zero, Y would be **14**.

---

# 12. R-Squared

Another key statistic in regression is **R² (R-squared)**.

R² measures **how well the regression line fits the data**.

---

### Interpretation

R² = **1 (or 100%)**

Perfect fit.

All points lie exactly on the line.

---

R² = **0**

No relationship.

The regression line is essentially **horizontal**.

---

In practice:

R² values like **60–70%** are considered **reasonably strong** in many financial applications.

---

# 13. Multiple Variables

So far we discussed **two variables**.

But what if we have **three or more variables**?

---

## Correlation

Correlation works **only in pairs**.

If you have variables:

A, B, C

You must compute:

* correlation(A,B)
* correlation(A,C)
* correlation(B,C)

---

## Scatter Plots

With three variables, you can create **3D scatter plots**.

But these quickly become **hard to interpret**.

Beyond three variables, visualization becomes extremely difficult.

---

# 14. Multiple Regression

Regression, however, can easily handle many variables.

A **multiple regression** looks like:

[
Y = a + b_1 X_1 + b_2 X_2 + b_3 X_3
]

Where:

* Y = dependent variable
* X₁, X₂, X₃ = independent variables

You can include **many independent variables** as long as you have enough data.

---

# 15. Dummy Variables

Sometimes a variable is **categorical rather than numerical**.

Example:

* Public company
* Private company

We cannot put these directly into regression.

Instead we use a **dummy variable**.

---

## Dummy Variable

A dummy variable takes only two values:

[
0 \quad \text{or} \quad 1
]

Example:

Private company = **1**

Public company = **0**

This allows us to test whether **private firms behave differently than public firms**.

---

# Final Summary

When analyzing relationships between variables:

1. Start with a **scatter plot**
2. Compute **correlation and covariance**
3. Be cautious of **spurious correlations**
4. Remember **correlation ≠ causation**
5. Use **regression analysis** to model relationships
6. For multiple variables, use **multiple regression**
7. Use **dummy variables** for categorical data

```python
"""
Relationships Between Two Variables — Lecture Visualization
============================================================
Split at each  # ─── Cell N  comment to create separate Jupyter cells.

Dependencies:  pip install numpy matplotlib scipy pandas seaborn
"""

# ─── Cell 1: Imports & Theme ──────────────────────────────────────────────────
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from matplotlib.patches import FancyArrowPatch
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

plt.rcParams.update({
    "figure.facecolor": "#0d0d1a",
    "axes.facecolor":   "#12122a",
    "axes.edgecolor":   "#333355",
    "axes.labelcolor":  "#ccccee",
    "xtick.color":      "#888899",
    "ytick.color":      "#888899",
    "text.color":       "#eeeeff",
    "grid.color":       "#1e1e3a",
    "grid.linestyle":   "--",
    "grid.alpha":       0.6,
    "font.family":      "monospace",
    "axes.spines.top":  False,
    "axes.spines.right":False,
})

BLUE   = "#7eb8f7"
ORANGE = "#f7a07e"
GREEN  = "#7ef7c0"
YELLOW = "#f7e07e"
PURPLE = "#c07ef7"
PINK   = "#f77eb8"
CYAN   = "#7ef7f7"
RED    = "#f77e7e"

print("✓ Setup complete.")


# ─── Cell 2: Scatter Plot Intuition — 10-Point Example ───────────────────────
np.random.seed(42)

days = np.arange(1, 11)
X    = np.array([14, 16, 11, 13, 18, 10, 15, 17, 12, 14])
Y    = np.array([ 1,  3,  2,  2,  5,  1,  4,  5,  2,  3])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Scatter Plot — Paired Observations  (X, Y)", fontsize=14, color="#eeeeff")

# Left: plain scatter with labels
ax = axes[0]
ax.set_title("Each Point = One Observation  (Xᵢ, Yᵢ)", color=BLUE)
ax.scatter(X, Y, color=BLUE, s=120, zorder=4, edgecolors="white", linewidths=0.8)
for i, (xi, yi) in enumerate(zip(X, Y)):
    ax.annotate(f" Day {i+1}", (xi, yi), fontsize=8, color="#aaaacc")
ax.set_xlabel("X  (Independent Variable)")
ax.set_ylabel("Y  (Dependent Variable)")
ax.grid(True)

# Right: add mean lines and deviation arrows for one point
ax = axes[1]
ax.set_title("Deviations from Mean — Basis of Correlation", color=ORANGE)
xm, ym = X.mean(), Y.mean()
ax.scatter(X, Y, color=BLUE, s=100, zorder=4, edgecolors="white", linewidths=0.8)
ax.axvline(xm, color=YELLOW,  lw=1.8, linestyle="--", label=f"X̄ = {xm:.1f}")
ax.axhline(ym, color=ORANGE,  lw=1.8, linestyle="--", label=f"Ȳ = {ym:.1f}")

# Highlight one point and its deviations
ax.annotate("", xy=(X[4], ym), xytext=(xm, ym),
            arrowprops=dict(arrowstyle="<->", color=GREEN, lw=2))
ax.annotate("", xy=(X[4], Y[4]), xytext=(X[4], ym),
            arrowprops=dict(arrowstyle="<->", color=PINK, lw=2))
ax.text(xm + (X[4]-xm)/2, ym + 0.15, "X dev", color=GREEN,  fontsize=8, ha="center")
ax.text(X[4] + 0.1, ym + (Y[4]-ym)/2, "Y dev", color=PINK,   fontsize=8)

ax.set_xlabel("X"); ax.set_ylabel("Y")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

plt.tight_layout()
plt.savefig("rel01_scatter_intro.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → rel01_scatter_intro.png")


# ─── Cell 3: Correlation Spectrum — All 5 Types ───────────────────────────────
np.random.seed(7)
n = 120

def make_correlated(r, n):
    """Generate (X, Y) with Pearson correlation ≈ r."""
    X = np.random.randn(n)
    noise = np.random.randn(n)
    Y = r * X + np.sqrt(1 - r**2) * noise
    return X, Y

cases = [
    ("+1.0  Perfect Positive",   *make_correlated(0.99,  n), GREEN),
    ("+0.7  Strong Positive",    *make_correlated(0.70,  n), BLUE),
    (" 0.0  No Correlation",     *make_correlated(0.00,  n), YELLOW),
    ("−0.7  Strong Negative",    *make_correlated(-0.70, n), ORANGE),
    ("−1.0  Perfect Negative",   *make_correlated(-0.99, n), RED),
]

fig, axes = plt.subplots(1, 5, figsize=(20, 4))
fig.suptitle("Correlation Spectrum  (ρ from −1 to +1)", fontsize=14, color="#eeeeff")

for ax, (title, X, Y, color) in zip(axes, cases):
    r, _ = stats.pearsonr(X, Y)
    ax.scatter(X, Y, color=color, alpha=0.5, s=18, edgecolors="none")

    # Regression line
    m, b = np.polyfit(X, Y, 1)
    xr = np.linspace(X.min(), X.max(), 200)
    ax.plot(xr, m * xr + b, color="white", lw=2)

    ax.set_title(title, color=color, fontsize=9)
    ax.text(0.5, 0.06, f"ρ = {r:.2f}", transform=ax.transAxes,
            ha="center", fontsize=11, color="white",
            bbox=dict(facecolor="#1a1a2e", edgecolor=color, boxstyle="round,pad=0.4"))
    ax.set_xticks([]); ax.set_yticks([])
    ax.grid(True)

plt.tight_layout()
plt.savefig("rel02_correlation_spectrum.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → rel02_correlation_spectrum.png")


# ─── Cell 4: Finance Example — P/E Ratios vs Interest Rates ──────────────────
np.random.seed(15)
n = 80

interest_rates = np.random.uniform(1, 10, n)
# Higher interest rates → lower P/E ratios (negative relationship)
pe_ratios = 30 - 2.1 * interest_rates + np.random.normal(0, 3, n)

r_val, p_val = stats.pearsonr(interest_rates, pe_ratios)
slope, intercept, *_ = stats.linregress(interest_rates, pe_ratios)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Finance Example — P/E Ratios vs Interest Rates", fontsize=14, color="#eeeeff")

# Scatter + regression line
ax = axes[0]
ax.set_title("Scatter Plot with Regression Line", color=BLUE)
ax.scatter(interest_rates, pe_ratios, color=BLUE, alpha=0.65, s=50, edgecolors="none")
xr = np.linspace(interest_rates.min(), interest_rates.max(), 200)
ax.plot(xr, slope * xr + intercept, color=ORANGE, lw=2.5,
        label=f"Y = {intercept:.1f} + ({slope:.2f})X")
ax.set_xlabel("Interest Rate (%)")
ax.set_ylabel("P/E Ratio")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.text(0.05, 0.08, f"ρ = {r_val:.2f}   p = {p_val:.4f}",
        transform=ax.transAxes, fontsize=10, color=YELLOW,
        bbox=dict(facecolor="#1a1a2e", edgecolor=YELLOW, boxstyle="round,pad=0.4"))
ax.grid(True)

# Residuals plot
ax = axes[1]
ax.set_title("Residuals  (Distance from Regression Line)", color=ORANGE)
predicted  = slope * interest_rates + intercept
residuals  = pe_ratios - predicted
ax.scatter(interest_rates, residuals, color=ORANGE, alpha=0.65, s=50, edgecolors="none")
ax.axhline(0, color="white", lw=2, linestyle="--")
for xi, ri in zip(interest_rates, residuals):
    ax.plot([xi, xi], [0, ri], color=ORANGE, alpha=0.2, lw=0.8)
ax.set_xlabel("Interest Rate (%)")
ax.set_ylabel("Residual")
ax.grid(True)

plt.tight_layout()
plt.savefig("rel03_finance_scatter_regression.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel03_finance_scatter_regression.png")


# ─── Cell 5: OLS — Minimizing Squared Distances ──────────────────────────────
np.random.seed(3)
n_small = 12
Xs = np.linspace(2, 10, n_small) + np.random.randn(n_small) * 0.4
Ys = 1.5 * Xs + 2 + np.random.randn(n_small) * 2.5

slope_ols, intercept_ols, *_ = stats.linregress(Xs, Ys)
pred = slope_ols * Xs + intercept_ols

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Ordinary Least Squares (OLS) — Minimising Squared Distances",
             fontsize=13, color="#eeeeff")

# Left: vertical residual lines
ax = axes[0]
ax.set_title("OLS Minimises Sum of Squared Vertical Distances", color=GREEN)
xr = np.linspace(Xs.min() - 0.5, Xs.max() + 0.5, 200)
ax.plot(xr, slope_ols * xr + intercept_ols, color=GREEN, lw=2.5,
        label=f"Y = {intercept_ols:.1f} + {slope_ols:.2f}X")
ax.scatter(Xs, Ys, color=BLUE, s=80, zorder=4, edgecolors="white", linewidths=0.8)
for xi, yi, pi in zip(Xs, Ys, pred):
    ax.plot([xi, xi], [yi, pi], color=RED, lw=1.5, linestyle="--", alpha=0.7)
    sq_side = abs(yi - pi) * 0.4
    rect = plt.Rectangle((xi, min(yi, pi)), sq_side, abs(yi - pi),
                          color=RED, alpha=0.12)
    ax.add_patch(rect)
ax.set_xlabel("X"); ax.set_ylabel("Y")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Right: annotate slope and intercept
ax = axes[1]
ax.set_title("Slope & Intercept Interpretation", color=YELLOW)
ax.plot(xr, slope_ols * xr + intercept_ols, color=GREEN, lw=2.5)
ax.scatter(Xs, Ys, color=BLUE, s=60, zorder=4, edgecolors="none", alpha=0.7)

# Intercept annotation
ax.axvline(0, color="#333355", lw=1, linestyle=":")
ax.scatter([0], [intercept_ols], color=YELLOW, s=150, zorder=5, marker="*")
ax.annotate(f"Intercept a = {intercept_ols:.1f}\n(Y when X = 0)",
            xy=(0, intercept_ols), xytext=(2, intercept_ols - 3),
            color=YELLOW, fontsize=9,
            arrowprops=dict(arrowstyle="->", color=YELLOW, lw=1.2))

# Slope annotation — draw a rise/run triangle
x1, x2 = 4, 6
y1, y2 = slope_ols * x1 + intercept_ols, slope_ols * x2 + intercept_ols
ax.plot([x1, x2, x2], [y1, y1, y2], color=ORANGE, lw=2)
ax.text((x1+x2)/2, y1 - 1.2, "Run = 2", ha="center", fontsize=8, color=ORANGE)
ax.text(x2 + 0.15, (y1+y2)/2, f"Rise = {y2-y1:.1f}", fontsize=8, color=ORANGE)
ax.text(5.5, y2 + 1, f"Slope b = {slope_ols:.2f}", color=ORANGE, fontsize=9)

ax.set_xlabel("X"); ax.set_ylabel("Y")
ax.grid(True)

plt.tight_layout()
plt.savefig("rel04_ols_explained.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → rel04_ols_explained.png")


# ─── Cell 6: R-Squared — What It Means ───────────────────────────────────────
np.random.seed(99)
n = 80
X_base = np.random.randn(n)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("R²  (R-Squared) — How Well Does the Line Fit?", fontsize=14, color="#eeeeff")

r2_scenarios = [
    ("High R² ≈ 0.90\n(Strong fit)", 0.95,  GREEN),
    ("Medium R² ≈ 0.55\n(Moderate)",  0.74,  YELLOW),
    ("Low R² ≈ 0.05\n(Weak fit)",     0.22,  RED),
]

for ax, (title, r, color) in zip(axes, r2_scenarios):
    Y_sc = r * X_base + np.sqrt(1 - r**2) * np.random.randn(n)
    slope_s, intercept_s, r_val, *_ = stats.linregress(X_base, Y_sc)
    r2 = r_val**2

    ax.scatter(X_base, Y_sc, color=color, alpha=0.55, s=35, edgecolors="none")
    xr = np.linspace(X_base.min(), X_base.max(), 200)
    ax.plot(xr, slope_s * xr + intercept_s, color="white", lw=2.5)

    # Total mean line
    ax.axhline(Y_sc.mean(), color="#555577", lw=1.2, linestyle=":")

    ax.set_title(title, color=color, fontsize=10)
    ax.text(0.5, 0.07, f"R² = {r2:.2f}", transform=ax.transAxes,
            ha="center", fontsize=14, color=color,
            bbox=dict(facecolor="#1a1a2e", edgecolor=color, boxstyle="round,pad=0.5"))
    ax.set_xticks([]); ax.set_yticks([])
    ax.grid(True)

plt.tight_layout()
plt.savefig("rel05_r_squared.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → rel05_r_squared.png")


# ─── Cell 7: Spurious Correlation — Super Bowl Indicator ─────────────────────
np.random.seed(42)
years = np.arange(1967, 2024)
n_years = len(years)

# Simulate: ~60% hit rate (looks like a pattern, but it's noise)
nfc_wins   = np.random.choice([0, 1], size=n_years, p=[0.45, 0.55])
sp500_up   = np.random.choice([0, 1], size=n_years, p=[0.35, 0.65])
# Force ~70% correlation artificially to illustrate the spurious point
match      = (nfc_wins == sp500_up).astype(int)
hit_rate   = match.mean()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Spurious Correlation — The Super Bowl Indicator ⚠",
             fontsize=14, color=YELLOW)

# Bar chart: NFC win vs S&P direction
ax = axes[0]
ax.set_title("Did S&P 500 Rise When NFC Won?", color=YELLOW)
categories = ["NFC Win\n+ S&P Up", "NFC Win\n+ S&P Down",
              "AFC Win\n+ S&P Down", "AFC Win\n+ S&P Up"]
nfc  = nfc_wins == 1
afc  = nfc_wins == 0
counts = [
    ((nfc)  & (sp500_up == 1)).sum(),
    ((nfc)  & (sp500_up == 0)).sum(),
    ((afc)  & (sp500_up == 0)).sum(),
    ((afc)  & (sp500_up == 1)).sum(),
]
colors_bar = [GREEN, RED, GREEN, RED]
bars = ax.bar(categories, counts, color=colors_bar, edgecolor="#0d0d1a",
              width=0.55, alpha=0.85)
for bar, val in zip(bars, counts):
    ax.text(bar.get_x() + bar.get_width()/2, val + 0.3,
            str(val), ha="center", fontsize=11, color="white")
ax.set_ylabel("Number of Years")
ax.text(0.5, 0.93, f"'Prediction' accuracy: {hit_rate:.0%}",
        transform=ax.transAxes, ha="center", fontsize=10, color=YELLOW,
        bbox=dict(facecolor="#1a1a2e", edgecolor=YELLOW, boxstyle="round,pad=0.4"))
ax.grid(True, axis="y")

# Warning panel
ax = axes[1]
ax.set_facecolor("#1a0a0a")
ax.axis("off")
warnings_text = [
    ("⚠  Spurious Correlation Warning", YELLOW, 14, "bold"),
    ("", "#eeeeff", 10, "normal"),
    ("The Super Bowl Indicator appears to", "#ccccee", 10, "normal"),
    ("predict stock returns in historical data.", "#ccccee", 10, "normal"),
    ("", "#eeeeff", 9, "normal"),
    ("But there is ZERO economic logic", RED, 11, "bold"),
    ("linking NFL results to markets.", RED, 11, "bold"),
    ("", "#eeeeff", 9, "normal"),
    ("With thousands of variables and decades", "#ccccee", 10, "normal"),
    ("of data, accidental correlations WILL", "#ccccee", 10, "normal"),
    ("appear — this is called data mining.", "#ccccee", 10, "normal"),
    ("", "#eeeeff", 9, "normal"),
    ("Correlation ≠ Causation", ORANGE, 13, "bold"),
    ("Always ask: is there an economic reason?", ORANGE, 10, "normal"),
]
y_pos = 0.95
for line, color, size, weight in warnings_text:
    ax.text(0.05, y_pos, line, transform=ax.transAxes, fontsize=size,
            color=color, fontweight=weight, va="top")
    y_pos -= 0.068

plt.tight_layout()
plt.savefig("rel06_spurious_correlation.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel06_spurious_correlation.png")


# ─── Cell 8: Correlation vs Causation — Visual ───────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Correlation vs Causation — Three Possible Explanations",
             fontsize=14, color="#eeeeff")

scenarios = [
    ("A causes B\n(Direct causation)",
     "Interest Rates ↑", "P/E Ratios ↓",
     None, BLUE, GREEN, "→", "causes"),
    ("B causes A\n(Reverse causation)",
     "Organic Food\nSpending ↑", "Household\nWealth ↑",
     None, ORANGE, PURPLE, "←", "actually"),
    ("C causes both\n(Confounding variable)",
     "Variable A", "Variable B",
     "Hidden Variable C\n(Confounder)", RED, YELLOW, "↔", "spurious"),
]

for ax, (title, label_a, label_b, label_c, col_a, col_b, arrow, arrow_label) in \
        zip(axes, scenarios):
    ax.set_title(title, color=YELLOW, fontsize=10)
    ax.set_xlim(0, 10); ax.set_ylim(0, 10)
    ax.axis("off")

    # Node A
    ax.text(2, 5, label_a, ha="center", va="center", fontsize=9, color=col_a,
            bbox=dict(facecolor="#1a1a2e", edgecolor=col_a, boxstyle="round,pad=0.5"))
    # Node B
    ax.text(8, 5, label_b, ha="center", va="center", fontsize=9, color=col_b,
            bbox=dict(facecolor="#1a1a2e", edgecolor=col_b, boxstyle="round,pad=0.5"))

    # Arrow between A and B
    ax.annotate("", xy=(6.5, 5), xytext=(3.5, 5),
                arrowprops=dict(arrowstyle="->" if arrow == "→" else
                                "<-" if arrow == "←" else "<->",
                                color="white", lw=2))
    ax.text(5, 5.4, arrow_label, ha="center", fontsize=9, color="#aaaacc")

    # Confounder node
    if label_c:
        ax.text(5, 8, label_c, ha="center", va="center", fontsize=9, color=RED,
                bbox=dict(facecolor="#2a1010", edgecolor=RED, boxstyle="round,pad=0.5"))
        ax.annotate("", xy=(2.5, 5.6), xytext=(4.5, 7.5),
                    arrowprops=dict(arrowstyle="->", color=RED, lw=1.5))
        ax.annotate("", xy=(7.5, 5.6), xytext=(5.5, 7.5),
                    arrowprops=dict(arrowstyle="->", color=RED, lw=1.5))

plt.tight_layout()
plt.savefig("rel07_correlation_vs_causation.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel07_correlation_vs_causation.png")


# ─── Cell 9: Covariance vs Correlation — Side-by-Side ────────────────────────
np.random.seed(5)
n = 60

# IBM-like stock returns vs market
market_ret = np.random.normal(0.08, 0.15, n)
ibm_ret    = 0.22 * market_ret + np.random.normal(0, 0.12, n)

cov_val  = np.cov(ibm_ret, market_ret)[0, 1]
corr_val, _ = stats.pearsonr(ibm_ret, market_ret)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Covariance vs Correlation  (IBM Returns vs Market Returns)",
             fontsize=13, color="#eeeeff")

for ax, (metric, val, color, note) in zip(axes, [
    ("Covariance", cov_val,  ORANGE,
     f"Cov = {cov_val:.4f}\n\nIs this large or small?\nHard to tell — no fixed scale!\nDepends on units of X and Y."),
    ("Correlation", corr_val, GREEN,
     f"ρ = {corr_val:.2f}\n\nAlways between −1 and +1\nWeak-to-moderate positive\nrelationship  →  easy to interpret!"),
]):
    ax.scatter(market_ret, ibm_ret, color=color, alpha=0.6, s=50, edgecolors="none")
    m, b, *_ = stats.linregress(market_ret, ibm_ret)
    xr = np.linspace(market_ret.min(), market_ret.max(), 200)
    ax.plot(xr, m*xr + b, color="white", lw=2)
    ax.set_title(f"{metric}", color=color, fontsize=12)
    ax.set_xlabel("Market Return"); ax.set_ylabel("IBM Return")
    ax.text(0.97, 0.05, note, transform=ax.transAxes, ha="right", va="bottom",
            fontsize=9, color=color,
            bbox=dict(facecolor="#1a1a2e", edgecolor=color, boxstyle="round,pad=0.5"))
    ax.grid(True)

plt.tight_layout()
plt.savefig("rel08_covariance_vs_correlation.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel08_covariance_vs_correlation.png")


# ─── Cell 10: Multiple Regression — 3 Variables ──────────────────────────────
np.random.seed(12)
n = 200

gdp_growth   = np.random.normal(2.5, 1.2, n)
interest_rate_mr = np.random.normal(4.0, 1.5, n)
earnings_growth  = (1.8 * gdp_growth - 1.1 * interest_rate_mr
                    + np.random.normal(0, 1.5, n))

# OLS via numpy
X_design = np.column_stack([np.ones(n), gdp_growth, interest_rate_mr])
coeffs, residuals_ss, *_ = np.linalg.lstsq(X_design, earnings_growth, rcond=None)
a, b1, b2 = coeffs
y_pred_mr  = X_design @ coeffs
ss_res     = np.sum((earnings_growth - y_pred_mr)**2)
ss_tot     = np.sum((earnings_growth - earnings_growth.mean())**2)
r2_mr      = 1 - ss_res / ss_tot

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Multiple Regression  →  Earnings Growth = a + b₁·GDP + b₂·Interest Rate",
             fontsize=13, color="#eeeeff")

# Panel 1: GDP vs Earnings
ax = axes[0]
ax.set_title("GDP Growth vs Earnings Growth", color=GREEN)
ax.scatter(gdp_growth, earnings_growth, color=GREEN, alpha=0.4, s=25, edgecolors="none")
m, b, *_ = stats.linregress(gdp_growth, earnings_growth)
xr = np.linspace(gdp_growth.min(), gdp_growth.max(), 200)
ax.plot(xr, m*xr + b, color="white", lw=2)
r1, _ = stats.pearsonr(gdp_growth, earnings_growth)
ax.text(0.05, 0.93, f"ρ = {r1:.2f}", transform=ax.transAxes, fontsize=10,
        color=GREEN, bbox=dict(facecolor="#1a1a2e", edgecolor=GREEN, boxstyle="round,pad=0.4"))
ax.set_xlabel("GDP Growth (%)"); ax.set_ylabel("Earnings Growth (%)")
ax.grid(True)

# Panel 2: Interest Rate vs Earnings
ax = axes[1]
ax.set_title("Interest Rate vs Earnings Growth", color=ORANGE)
ax.scatter(interest_rate_mr, earnings_growth, color=ORANGE, alpha=0.4, s=25, edgecolors="none")
m2, b2_, *_ = stats.linregress(interest_rate_mr, earnings_growth)
xr2 = np.linspace(interest_rate_mr.min(), interest_rate_mr.max(), 200)
ax.plot(xr2, m2*xr2 + b2_, color="white", lw=2)
r2_, _ = stats.pearsonr(interest_rate_mr, earnings_growth)
ax.text(0.05, 0.93, f"ρ = {r2_:.2f}", transform=ax.transAxes, fontsize=10,
        color=ORANGE, bbox=dict(facecolor="#1a1a2e", edgecolor=ORANGE, boxstyle="round,pad=0.4"))
ax.set_xlabel("Interest Rate (%)"); ax.set_ylabel("Earnings Growth (%)")
ax.grid(True)

# Panel 3: Multiple Regression equation + R²
ax = axes[2]
ax.set_facecolor("#0d0d1a")
ax.axis("off")
ax.set_title("Multiple Regression Results", color=BLUE)
eq_lines = [
    ("Multiple Regression Equation:", BLUE,   11, "bold"),
    ("", "#eeeeff", 8, "normal"),
    (f"Earnings = {a:.2f}", CYAN, 12, "bold"),
    (f"   + {b1:.2f} × GDP_Growth", GREEN, 12, "bold"),
    (f"   + ({b2:.2f}) × Interest_Rate", ORANGE, 12, "bold"),
    ("", "#eeeeff", 8, "normal"),
    (f"R² = {r2_mr:.2f}  ({r2_mr*100:.1f}% of variation explained)", YELLOW, 11, "bold"),
    ("", "#eeeeff", 8, "normal"),
    ("b₁ = +ve  →  GDP ↑ boosts earnings", GREEN, 9, "normal"),
    ("b₂ = −ve  →  rates ↑ hurt earnings", ORANGE, 9, "normal"),
    ("", "#eeeeff", 8, "normal"),
    ("Each coefficient holds the other", "#aaaacc", 9, "normal"),
    ("variable CONSTANT  (ceteris paribus)", "#aaaacc", 9, "normal"),
]
y_pos = 0.95
for line, color, size, weight in eq_lines:
    ax.text(0.05, y_pos, line, transform=ax.transAxes, fontsize=size,
            color=color, fontweight=weight, va="top")
    y_pos -= 0.08

plt.tight_layout()
plt.savefig("rel09_multiple_regression.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel09_multiple_regression.png")


# ─── Cell 11: Dummy Variables ─────────────────────────────────────────────────
np.random.seed(33)
n_each = 80

# Public companies: lower returns on average
public_returns  = np.random.normal(0.08, 0.18, n_each)
# Private companies: slightly higher, more variable
private_returns = np.random.normal(0.14, 0.25, n_each)

all_returns = np.concatenate([public_returns,  private_returns])
dummy       = np.concatenate([np.zeros(n_each), np.ones(n_each)])   # 0=Public, 1=Private

slope_d, intercept_d, r_d, p_d, _ = stats.linregress(dummy, all_returns)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Dummy Variables  (Categorical Data in Regression)", fontsize=14, color="#eeeeff")

# Left: distributions
ax = axes[0]
ax.set_title("Return Distributions by Company Type", color=BLUE)
bins = np.linspace(-0.6, 0.8, 35)
ax.hist(public_returns,  bins=bins, density=True, alpha=0.5, color=BLUE,
        edgecolor="#0d0d1a", label="Public  (dummy = 0)")
ax.hist(private_returns, bins=bins, density=True, alpha=0.5, color=ORANGE,
        edgecolor="#0d0d1a", label="Private (dummy = 1)")
ax.axvline(public_returns.mean(),  color=BLUE,   lw=2, linestyle="--",
           label=f"Public mean = {public_returns.mean():.1%}")
ax.axvline(private_returns.mean(), color=ORANGE, lw=2, linestyle="--",
           label=f"Private mean = {private_returns.mean():.1%}")
ax.set_xlabel("Annual Return"); ax.set_ylabel("Density")
ax.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Right: regression with dummy
ax = axes[1]
ax.set_title("Regression on Dummy Variable", color=ORANGE)
jitter = np.random.uniform(-0.08, 0.08, len(all_returns))
ax.scatter(dummy + jitter, all_returns,
           color=[BLUE if d == 0 else ORANGE for d in dummy],
           alpha=0.35, s=20, edgecolors="none")
xd = np.array([-0.2, 1.2])
ax.plot(xd, slope_d * xd + intercept_d, color="white", lw=2.5,
        label=f"Y = {intercept_d:.3f} + {slope_d:.3f}·Dummy")
ax.set_xticks([0, 1])
ax.set_xticklabels(["0  (Public)", "1  (Private)"])
ax.set_xlabel("Dummy Variable"); ax.set_ylabel("Return")
ax.text(0.5, 0.93,
        f"Slope = {slope_d:.3f}  →  private firms earn ~{slope_d:.1%} more\np = {p_d:.4f}",
        transform=ax.transAxes, ha="center", fontsize=9, color=YELLOW,
        bbox=dict(facecolor="#1a1a2e", edgecolor=YELLOW, boxstyle="round,pad=0.4"))
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

plt.tight_layout()
plt.savefig("rel10_dummy_variables.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel10_dummy_variables.png")


# ─── Cell 12: Correlation Matrix Heatmap — Multi-Variable ────────────────────
np.random.seed(8)
n = 150

gdp_h        = np.random.normal(2.5, 1.2, n)
interest_h   = np.random.normal(4.0, 1.5, n)
sp500_h      = 0.6  * gdp_h  - 0.4  * interest_h + np.random.normal(0, 0.8, n)
earnings_h   = 1.5  * gdp_h  - 0.9  * interest_h + np.random.normal(0, 1.0, n)
pe_h         = -0.5 * interest_h + 0.3 * sp500_h  + np.random.normal(0, 1.5, n)

df = pd.DataFrame({
    "GDP Growth":    gdp_h,
    "Interest Rate": interest_h,
    "S&P 500 Ret":   sp500_h,
    "Earnings Gr.":  earnings_h,
    "P/E Ratio":     pe_h,
})
corr_matrix = df.corr()

fig, ax = plt.subplots(figsize=(10, 7))
fig.suptitle("Correlation Matrix Heatmap — All Pairwise Correlations",
             fontsize=14, color="#eeeeff")

n_vars = len(corr_matrix)
im = ax.imshow(corr_matrix.values, cmap="RdYlGn", vmin=-1, vmax=1, aspect="auto")

ax.set_xticks(range(n_vars)); ax.set_xticklabels(corr_matrix.columns, rotation=30, ha="right")
ax.set_yticks(range(n_vars)); ax.set_yticklabels(corr_matrix.columns)

for i in range(n_vars):
    for j in range(n_vars):
        val = corr_matrix.values[i, j]
        color = "black" if abs(val) > 0.5 else "white"
        ax.text(j, i, f"{val:.2f}", ha="center", va="center",
                fontsize=11, color=color, fontweight="bold")

cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Correlation Coefficient", color="#ccccee")
cbar.ax.yaxis.set_tick_params(color="#ccccee")
plt.setp(cbar.ax.yaxis.get_ticklabels(), color="#ccccee")

ax.set_title("Pairwise Pearson Correlations — Finance Variables", color=BLUE, pad=10)
ax.grid(False)

plt.tight_layout()
plt.savefig("rel11_correlation_heatmap.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel11_correlation_heatmap.png")


# ─── Cell 13: Time Lag Analysis — Does X Lead Y? ─────────────────────────────
np.random.seed(77)
t      = np.arange(100)
X_lead = np.sin(t * 0.2) + np.random.normal(0, 0.2, 100)
# Y follows X with a lag of 5 periods
lag    = 5
Y_lag  = np.roll(X_lead, lag) + np.random.normal(0, 0.3, 100)
Y_lag[:lag] = np.nan

# Compute rolling correlations at different lags
max_lag   = 15
lag_vals  = range(-max_lag, max_lag + 1)
lag_corrs = []
for l in lag_vals:
    if l >= 0:
        xv = X_lead[:len(X_lead)-l]
        yv = Y_lag[l:]
    else:
        xv = X_lead[-l:]
        yv = Y_lag[:len(Y_lag)+l]
    mask = ~np.isnan(xv) & ~np.isnan(yv)
    if mask.sum() > 5:
        r, _ = stats.pearsonr(xv[mask], yv[mask])
    else:
        r = 0
    lag_corrs.append(r)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Time Lead / Lag Analysis — Does Variable A Move Before Variable B?",
             fontsize=13, color="#eeeeff")

# Time series
ax = axes[0]
ax.set_title("X  (leads)  vs  Y  (follows with 5-period lag)", color=BLUE)
ax.plot(t, X_lead, color=BLUE,   lw=1.8, label="X  (leading)")
ax.plot(t, Y_lag,  color=ORANGE, lw=1.8, label="Y  (lagged)", linestyle="--")
ax.axvspan(30, 35, color=YELLOW, alpha=0.15)
ax.axvspan(35, 40, color=ORANGE, alpha=0.10)
ax.annotate("X peaks here", xy=(32, X_lead[32]), xytext=(25, 1.5),
            color=BLUE, fontsize=8,
            arrowprops=dict(arrowstyle="->", color=BLUE, lw=1.2))
ax.annotate("Y peaks 5 periods later", xy=(37, 0.8), xytext=(42, 1.5),
            color=ORANGE, fontsize=8,
            arrowprops=dict(arrowstyle="->", color=ORANGE, lw=1.2))
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.set_xlabel("Time Period"); ax.grid(True)

# Cross-correlation plot
ax = axes[1]
ax.set_title("Cross-Correlation at Different Lags", color=GREEN)
colors_lag = [GREEN if abs(v) == max(abs(c) for c in lag_corrs) else BLUE for v in lag_corrs]
ax.bar(lag_vals, lag_corrs, color=colors_lag, edgecolor="#0d0d1a",
       width=0.8, alpha=0.85)
ax.axhline(0, color="white", lw=1, linestyle="--")
ax.axvline(lag, color=YELLOW, lw=2, linestyle=":", label=f"True lag = {lag}")
best_lag = lag_vals[np.argmax([abs(c) for c in lag_corrs])]
ax.text(best_lag, max(lag_corrs) + 0.04,
        f"Peak at lag {best_lag}", ha="center", color=YELLOW, fontsize=9)
ax.set_xlabel("Lag (periods)"); ax.set_ylabel("Correlation")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True, axis="y")

plt.tight_layout()
plt.savefig("rel12_time_lag_analysis.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → rel12_time_lag_analysis.png")


print("\n" + "="*58)
print("  ALL RELATIONSHIP VISUALIZATIONS COMPLETE")
print("="*58)
print("  rel01 — Scatter plot intro + deviation arrows")
print("  rel02 — Correlation spectrum (−1 to +1)")
print("  rel03 — Finance scatter: P/E vs Interest Rates")
print("  rel04 — OLS explained: residuals + slope/intercept")
print("  rel05 — R-squared (high / medium / low)")
print("  rel06 — Spurious correlation: Super Bowl indicator")
print("  rel07 — Correlation vs causation diagrams")
print("  rel08 — Covariance vs correlation (IBM vs market)")
print("  rel09 — Multiple regression (GDP + rates → earnings)")
print("  rel10 — Dummy variables (public vs private firms)")
print("  rel11 — Correlation matrix heatmap")
print("  rel12 — Time lead / lag cross-correlation")
print("="*58)
```

# Statistics Lecture Notes

## Regression Analysis, Dummy Variables & Regression Problems

---

# 1. Dummy Variables (Handling Categorical Data)

So far we discussed **dummy variables** as variables that take only two values:

[
0 \quad \text{or} \quad 1
]

They are useful when we want to represent **categories** in regression models.

---

## Example: Bond Ratings

Suppose you are studying companies and want to include **default risk** in your regression.

Default risk is normally a **continuous variable**, but sometimes it becomes messy to measure precisely.

Instead, we might divide firms based on **credit rating**.

For example:

* Investment Grade → rating **BBB or above**
* Non-Investment Grade → rating **below BBB**

Then we can define a dummy variable:

Investment Grade = **1**
Non-Investment Grade = **0**

This dummy variable captures the **difference between the two groups**.

---

# 2. Cross Effects Between Independent Variables

Sometimes independent variables **interact with each other**.

This means the effect of one variable **depends on the value of another variable**.

These are called **interaction effects**.

---

## Example: Explaining P/E Ratios

Consider the **price-earnings ratio (P/E)** used in investing.

Suppose we believe:

1. **Higher growth → higher P/E**
2. **Larger companies → lower P/E**

But we also suspect:

> Small companies with high growth behave differently from large companies with high growth.

This means there may be a **cross effect between growth and size**.

---

### Regression Setup

Dependent Variable:

[
PE
]

Independent Variables:

* Growth Rate
* Market Capitalization
* Growth × Market Cap (interaction term)

The regression becomes:

[
PE = a + b_1(Growth) + b_2(Size) + b_3(Growth \times Size)
]

This third term captures the **interaction between growth and company size**.

---

# 3. Checking a Regression Before Using It

Before relying on a regression model, we must check for **statistical problems**.

There are several important issues.

---

# 4. Linear vs Nonlinear Relationships

Traditional **ordinary least squares (OLS) regression** assumes a **linear relationship**.

That means:

When the independent variable increases steadily, the dependent variable changes **proportionally**.

Graphically:

A **straight-line relationship**.

---

## Example

Good linear relationships:

* upward sloping straight line
* downward sloping straight line

But sometimes the data follows a **curve**, not a straight line.

In this case, a linear regression may **fit poorly**.

---

## Solutions

If the relationship is nonlinear, we have two options.

### Option 1: Run Nonlinear Regression

Some statistical software can directly estimate nonlinear relationships.

However, these models are often:

* more complex
* harder to interpret

---

### Option 2: Transform Variables (More Common)

Instead of changing the regression method, we transform the variables.

Examples:

* log(X)
* (X^2)
* (\sqrt{X})

By transforming the data, we can often **convert nonlinear relationships into linear ones**.

In practice, many finance researchers prefer **data transformation instead of nonlinear regression**.

---

# 5. Multicollinearity

Multicollinearity occurs when **independent variables are correlated with each other**.

In a perfect regression:

Independent variables should be **uncorrelated**.

But in reality, this rarely happens.

---

## Example

Suppose we run a regression explaining **P/E ratios** using:

* Growth rate
* Dividend payout
* Risk

In real markets:

* High growth companies often pay **low dividends**
* High growth companies may also be **high risk**

So these independent variables become **correlated**.

---

## Why Multicollinearity Is a Problem

When independent variables are correlated:

The regression struggles to determine **which variable explains the dependent variable**.

This causes:

* unstable coefficients
* misleading interpretations

However:

Predictions from the regression can still be **reasonably accurate**.

The main problem is interpreting the **individual coefficients**.

---

# 6. Detecting Multicollinearity

First step:

Create a **correlation matrix** for the independent variables.

---

### Ideal Case

Matrix diagonal = **1**
All other correlations = **0**

---

### Real World

Correlations will **not be zero**.

This indicates **multicollinearity**.

---

# 7. Variance Inflation Factor (VIF)

A more formal measure of multicollinearity is the **Variance Inflation Factor (VIF)**.

Interpretation:

| VIF Value | Interpretation             |
| --------- | -------------------------- |
| < 1       | No multicollinearity       |
| 1–5       | Moderate multicollinearity |
| > 5       | Severe multicollinearity   |

In finance research, **moderate multicollinearity is very common**.

---

# 8. Solutions to Multicollinearity

If two independent variables are highly correlated, we can:

### Option 1: Remove One Variable

Choose the variable that has **less explanatory power** and remove it.

---

### Option 2: Run Separate Regressions

Instead of a multiple regression, run individual regressions:

Example:

* PE vs Growth
* PE vs Dividend Payout

---

### Option 3: Keep the Regression

Use the model for **prediction**, but avoid over-interpreting individual coefficients.

---

# 9. Homoscedasticity vs Heteroscedasticity

Another regression issue relates to **prediction errors**.

When regression predicts a value, the difference between:

Actual value – Predicted value

is called a **residual**.

---

## Homoscedasticity

In a good regression:

Residuals should have **constant variance**.

Errors should scatter randomly around the regression line.

No pattern should appear.

---

## Heteroscedasticity

If prediction errors **increase as predicted values increase**, the model has **heteroscedasticity**.

This indicates the regression may not be reliable.

---

# 10. Fixing Heteroscedasticity

One approach is to **transform the data**.

But more advanced regression methods can also help.

---

## Generalized Least Squares (GLS)

GLS improves the regression by using **information from the prediction errors**.

Steps:

1. Run a normal regression
2. Examine residuals
3. Use residual patterns to adjust the regression

GLS essentially **uses feedback from errors** to improve predictions.

---

## Weighted Least Squares (WLS)

In ordinary regression:

All observations receive **equal weight**.

But sometimes we want certain observations to **matter more**.

Example:

Large firms may be more economically important than small firms.

So we might weight observations by **market capitalization**.

---

### Benefits

Weighted regression can:

* reduce statistical problems
* reduce heteroscedasticity

However, weighting also means we **give less influence to some data points**.

---

# 11. Using Regression for Prediction

Even when regression works well, we must remember several limitations.

---

## 1. Predictions Are Estimates

Regression predictions are **not facts**.

Each prediction comes with a **standard error**.

Lower **R² → larger prediction errors**.

Higher **R² → smaller errors**.

---

## 2. Data Comes From the Past

Regression models rely on **historical data**.

But we are predicting the **future**.

If the economic structure changes, past relationships may **no longer hold**.

Example:

The **Global Financial Crisis** significantly changed financial markets.

Some analysts prefer to use **post-2008 data** only.

---

## 3. Observable Variables

For prediction to work, the independent variable must be **observable before the prediction**.

Example:

Suppose earnings growth is correlated with GDP growth.

But if GDP data is released **after earnings data**, it cannot help predict earnings.

---

# 12. Practical Use of Regression in Finance

In finance, regression analysis is **not just academic**.

We use regressions to:

* estimate company value
* predict returns
* identify market relationships

But always remember:

Regression outputs are **estimates with uncertainty**.

---

# Final Summary

When using regression models:

1. Check whether relationships are **linear**
2. Watch for **multicollinearity**
3. Examine **residuals**
4. Check for **heteroscedasticity**
5. Ensure variables are **observable**
6. Remember predictions come from **past data**

Ultimately, regression is a **tool for making better decisions**, not a guarantee of correct predictions.

```python
"""
Regression Analysis, Dummy Variables & Regression Problems
===========================================================
Split at each  # ─── Cell N  comment to create separate Jupyter cells.

Dependencies:  pip install numpy matplotlib scipy pandas statsmodels
"""

# ─── Cell 1: Imports & Theme ──────────────────────────────────────────────────
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

# Optional: statsmodels for VIF and GLS
try:
    import statsmodels.api as sm
    from statsmodels.stats.outliers_influence import variance_inflation_factor
    HAS_SM = True
    print("✓ statsmodels available — VIF and GLS cells will use it.")
except ImportError:
    HAS_SM = False
    print("⚠ statsmodels not found. Run:  pip install statsmodels")
    print("  VIF cell will use a manual fallback.")

plt.rcParams.update({
    "figure.facecolor": "#0d0d1a",
    "axes.facecolor":   "#12122a",
    "axes.edgecolor":   "#333355",
    "axes.labelcolor":  "#ccccee",
    "xtick.color":      "#888899",
    "ytick.color":      "#888899",
    "text.color":       "#eeeeff",
    "grid.color":       "#1e1e3a",
    "grid.linestyle":   "--",
    "grid.alpha":       0.6,
    "font.family":      "monospace",
    "axes.spines.top":  False,
    "axes.spines.right":False,
})

BLUE   = "#7eb8f7"
ORANGE = "#f7a07e"
GREEN  = "#7ef7c0"
YELLOW = "#f7e07e"
PURPLE = "#c07ef7"
PINK   = "#f77eb8"
CYAN   = "#7ef7f7"
RED    = "#f77e7e"
print("✓ Theme ready.")


# ─── Cell 2: Dummy Variables — Bond Ratings (Investment Grade) ────────────────
np.random.seed(42)
n = 120

# Investment-grade firms (dummy=1): lower yield spread, higher P/E
ig_pe     = np.random.normal(22, 4, n // 2)
ig_spread = np.random.normal(1.2, 0.4, n // 2)

# Non-investment-grade (dummy=0): higher spread, lower P/E
non_ig_pe     = np.random.normal(14, 5, n // 2)
non_ig_spread = np.random.normal(4.5, 1.2, n // 2)

all_pe     = np.concatenate([non_ig_pe,     ig_pe])
all_spread = np.concatenate([non_ig_spread, ig_spread])
dummy_ig   = np.concatenate([np.zeros(n//2), np.ones(n//2)])

slope_d, intercept_d, r_d, p_d, _ = stats.linregress(dummy_ig, all_pe)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Dummy Variables — Bond Credit Rating  (Investment Grade vs Non-IG)",
             fontsize=13, color="#eeeeff")

# Panel 1: P/E distributions
ax = axes[0]
ax.set_title("P/E Ratio by Credit Grade", color=BLUE)
bins = np.linspace(0, 36, 28)
ax.hist(non_ig_pe, bins=bins, density=True, alpha=0.55, color=RED,
        edgecolor="#0d0d1a", label="Non-IG  (dummy = 0)")
ax.hist(ig_pe,     bins=bins, density=True, alpha=0.55, color=GREEN,
        edgecolor="#0d0d1a", label="Inv. Grade (dummy = 1)")
ax.axvline(non_ig_pe.mean(), color=RED,   lw=2, linestyle="--",
           label=f"Non-IG mean = {non_ig_pe.mean():.1f}")
ax.axvline(ig_pe.mean(),     color=GREEN, lw=2, linestyle="--",
           label=f"IG mean = {ig_pe.mean():.1f}")
ax.set_xlabel("P/E Ratio"); ax.set_ylabel("Density")
ax.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Panel 2: yield spread distributions
ax = axes[1]
ax.set_title("Credit Spread by Grade", color=ORANGE)
bins2 = np.linspace(0, 8, 30)
ax.hist(non_ig_spread, bins=bins2, density=True, alpha=0.55, color=RED,
        edgecolor="#0d0d1a", label="Non-IG")
ax.hist(ig_spread,     bins=bins2, density=True, alpha=0.55, color=GREEN,
        edgecolor="#0d0d1a", label="Inv. Grade")
ax.set_xlabel("Credit Spread (%)"); ax.set_ylabel("Density")
ax.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Panel 3: regression on dummy
ax = axes[2]
ax.set_title("Regression: P/E = a + b·Dummy", color=YELLOW)
jitter = np.random.uniform(-0.07, 0.07, n)
ax.scatter(dummy_ig + jitter, all_pe,
           color=[GREEN if d else RED for d in dummy_ig],
           alpha=0.4, s=22, edgecolors="none")
xd = np.array([-0.3, 1.3])
ax.plot(xd, slope_d * xd + intercept_d, color="white", lw=2.5,
        label=f"Y = {intercept_d:.1f} + {slope_d:.1f}·Dummy")
ax.set_xticks([0, 1])
ax.set_xticklabels(["0  (Non-IG)", "1  (Inv. Grade)"])
ax.set_ylabel("P/E Ratio")
ax.text(0.5, 0.92,
        f"IG premium = {slope_d:.1f} P/E points\np = {p_d:.4f}",
        transform=ax.transAxes, ha="center", fontsize=9, color=YELLOW,
        bbox=dict(facecolor="#1a1a2e", edgecolor=YELLOW, boxstyle="round,pad=0.4"))
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

plt.tight_layout()
plt.savefig("reg01_dummy_bond_ratings.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg01_dummy_bond_ratings.png")


# ─── Cell 3: Interaction Effects — Growth × Size on P/E ──────────────────────
np.random.seed(7)
n = 200

growth   = np.random.uniform(2, 25, n)          # % earnings growth
size     = np.random.uniform(1, 100, n)          # market cap $bn
noise    = np.random.normal(0, 3, n)

# True model: PE = 10 + 1.2*growth - 0.05*size + 0.03*(growth*size) + noise
pe_interact = 10 + 1.2*growth - 0.05*size + 0.03*(growth * size) + noise

# Fit with and without interaction
X_no_int   = np.column_stack([np.ones(n), growth, size])
X_with_int = np.column_stack([np.ones(n), growth, size, growth * size])

def ols(X, y):
    coeffs, _, _, _ = np.linalg.lstsq(X, y, rcond=None)
    y_hat = X @ coeffs
    ss_res = np.sum((y - y_hat)**2)
    ss_tot = np.sum((y - y.mean())**2)
    return coeffs, 1 - ss_res / ss_tot

c_no,   r2_no   = ols(X_no_int,   pe_interact)
c_with, r2_with = ols(X_with_int, pe_interact)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Interaction Effects — Growth × Size on P/E Ratio\n"
             "PE = a + b₁·Growth + b₂·Size + b₃·(Growth × Size)",
             fontsize=13, color="#eeeeff")

# Scatter: growth vs PE, coloured by size quartile
ax = axes[0]
ax.set_title("P/E vs Growth  (colour = company size)", color=BLUE)
size_q = pd.qcut(size, 4, labels=False)
colors_q = [BLUE, GREEN, YELLOW, ORANGE]
for q in range(4):
    mask = size_q == q
    label = ["Small", "Mid-Small", "Mid-Large", "Large"][q]
    ax.scatter(growth[mask], pe_interact[mask],
               color=colors_q[q], alpha=0.55, s=25,
               edgecolors="none", label=label)
ax.set_xlabel("Earnings Growth (%)"); ax.set_ylabel("P/E Ratio")
ax.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355",
          title="Size Quartile", title_fontsize=8)
ax.grid(True)

# R² comparison + coefficient table
ax = axes[1]
ax.set_facecolor("#0d0d1a"); ax.axis("off")
ax.set_title("Model Comparison: Without vs With Interaction", color=YELLOW)

lines = [
    ("Without Interaction Term:", BLUE, 11, "bold"),
    (f"  R² = {r2_no:.3f}", BLUE, 10, "normal"),
    (f"  Growth coeff   = {c_no[1]:+.3f}", BLUE, 10, "normal"),
    (f"  Size coeff     = {c_no[2]:+.3f}", BLUE, 10, "normal"),
    ("", "#eeeeff", 8, "normal"),
    ("With Interaction Term:", GREEN, 11, "bold"),
    (f"  R² = {r2_with:.3f}  ← better fit", GREEN, 10, "normal"),
    (f"  Growth coeff        = {c_with[1]:+.3f}", GREEN, 10, "normal"),
    (f"  Size coeff          = {c_with[2]:+.3f}", GREEN, 10, "normal"),
    (f"  Growth×Size coeff   = {c_with[3]:+.3f}", YELLOW, 10, "bold"),
    ("", "#eeeeff", 8, "normal"),
    ("The interaction term shows:", YELLOW, 10, "bold"),
    ("Effect of growth on P/E DEPENDS", ORANGE, 9, "normal"),
    ("on the company's size.", ORANGE, 9, "normal"),
]
y_pos = 0.97
for line, color, size_f, weight in lines:
    ax.text(0.05, y_pos, line, transform=ax.transAxes, fontsize=size_f,
            color=color, fontweight=weight, va="top")
    y_pos -= 0.072

plt.tight_layout()
plt.savefig("reg02_interaction_effects.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg02_interaction_effects.png")


# ─── Cell 4: Linear vs Nonlinear — Visual + Transformation ───────────────────
np.random.seed(3)
n = 100
X_raw = np.random.uniform(0.5, 10, n)

# True nonlinear relationship: Y = 5 * log(X) + noise
Y_nonlin = 5 * np.log(X_raw) + np.random.normal(0, 0.8, n)

# OLS on raw X (wrong — linear assumption)
m_lin, b_lin, r_lin, *_ = stats.linregress(X_raw, Y_nonlin)
r2_lin = r_lin**2

# OLS on log(X) (correct transformation)
X_log = np.log(X_raw)
m_log, b_log, r_log, *_ = stats.linregress(X_log, Y_nonlin)
r2_log = r_log**2

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Linear vs Nonlinear Relationships — When to Transform Variables",
             fontsize=13, color="#eeeeff")

xr = np.linspace(X_raw.min(), X_raw.max(), 300)

# Panel 1: raw scatter — linear fit (poor)
ax = axes[0]
ax.set_title(f"Linear Fit on Raw X  (R² = {r2_lin:.2f})", color=RED)
ax.scatter(X_raw, Y_nonlin, color=BLUE, alpha=0.5, s=30, edgecolors="none")
ax.plot(xr, m_lin * xr + b_lin, color=RED, lw=2.5, label="Linear fit")
ax.set_xlabel("X  (raw)"); ax.set_ylabel("Y")
ax.text(0.5, 0.07, "Poor fit — curved pattern\nnot captured", transform=ax.transAxes,
        ha="center", fontsize=9, color=RED,
        bbox=dict(facecolor="#1a1a2e", edgecolor=RED, boxstyle="round,pad=0.4"))
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Panel 2: residuals from linear fit — shows pattern
ax = axes[1]
ax.set_title("Residuals of Linear Fit — Systematic Pattern ⚠", color=ORANGE)
resid_lin = Y_nonlin - (m_lin * X_raw + b_lin)
ax.scatter(X_raw, resid_lin, color=ORANGE, alpha=0.55, s=30, edgecolors="none")
ax.axhline(0, color="white", lw=2, linestyle="--")
# Smooth curve through residuals to show pattern
from scipy.ndimage import uniform_filter1d
sort_idx = np.argsort(X_raw)
smooth = uniform_filter1d(resid_lin[sort_idx], size=12)
ax.plot(X_raw[sort_idx], smooth, color=RED, lw=2, label="Residual trend")
ax.set_xlabel("X"); ax.set_ylabel("Residual")
ax.text(0.5, 0.07, "Residuals follow a curve\n→ nonlinearity not modelled",
        transform=ax.transAxes, ha="center", fontsize=9, color=ORANGE,
        bbox=dict(facecolor="#1a1a2e", edgecolor=ORANGE, boxstyle="round,pad=0.4"))
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Panel 3: log(X) transformation — near-perfect fit
ax = axes[2]
ax.set_title(f"After log(X) Transformation  (R² = {r2_log:.2f})", color=GREEN)
ax.scatter(X_log, Y_nonlin, color=GREEN, alpha=0.5, s=30, edgecolors="none")
xr_log = np.linspace(X_log.min(), X_log.max(), 300)
ax.plot(xr_log, m_log * xr_log + b_log, color="white", lw=2.5, label="Linear on log(X)")
ax.set_xlabel("log(X)"); ax.set_ylabel("Y")
ax.text(0.5, 0.07, "Transformation linearises\nthe relationship ✓",
        transform=ax.transAxes, ha="center", fontsize=9, color=GREEN,
        bbox=dict(facecolor="#1a1a2e", edgecolor=GREEN, boxstyle="round,pad=0.4"))
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

plt.tight_layout()
plt.savefig("reg03_linear_vs_nonlinear.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg03_linear_vs_nonlinear.png")


# ─── Cell 5: Multicollinearity — Correlation Matrix + VIF ────────────────────
np.random.seed(15)
n = 150

# Correlated independent variables (common in finance)
growth_mc  = np.random.normal(10, 4, n)
div_payout = -0.6 * growth_mc + np.random.normal(40, 5, n)   # high growth → low div
risk_mc    = 0.55 * growth_mc + np.random.normal(1, 0.3, n)  # high growth → higher risk
pe_mc      = 2.5*growth_mc - 0.4*div_payout - 3*risk_mc + np.random.normal(0, 4, n)

df_mc = pd.DataFrame({
    "Growth":      growth_mc,
    "Div. Payout": div_payout,
    "Risk":        risk_mc,
})
corr_mc = df_mc.corr()

# VIF calculation
def calc_vif(df):
    vif_data = {}
    X = df.values
    for i, col in enumerate(df.columns):
        if HAS_SM:
            X_sm = sm.add_constant(X)
            vif_data[col] = variance_inflation_factor(X_sm, i + 1)
        else:
            # Manual: regress col on all others, VIF = 1/(1-R²)
            others = np.delete(X, i, axis=1)
            ones   = np.ones((len(X), 1))
            X_oth  = np.hstack([ones, others])
            c, *_  = np.linalg.lstsq(X_oth, X[:, i], rcond=None)
            y_hat  = X_oth @ c
            ss_res = np.sum((X[:, i] - y_hat)**2)
            ss_tot = np.sum((X[:, i] - X[:, i].mean())**2)
            r2     = 1 - ss_res / ss_tot
            vif_data[col] = 1 / (1 - r2 + 1e-10)
    return vif_data

vif_vals = calc_vif(df_mc)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Multicollinearity — Correlation Matrix & Variance Inflation Factor (VIF)",
             fontsize=13, color="#eeeeff")

# Heatmap
ax = axes[0]
ax.set_title("Correlation Matrix of Independent Variables", color=BLUE)
n_v = len(corr_mc)
im  = ax.imshow(corr_mc.values, cmap="RdYlGn", vmin=-1, vmax=1, aspect="auto")
ax.set_xticks(range(n_v)); ax.set_xticklabels(corr_mc.columns, rotation=20, ha="right")
ax.set_yticks(range(n_v)); ax.set_yticklabels(corr_mc.columns)
for i in range(n_v):
    for j in range(n_v):
        val = corr_mc.values[i, j]
        ax.text(j, i, f"{val:.2f}", ha="center", va="center", fontsize=12,
                color="black" if abs(val) > 0.4 else "white", fontweight="bold")
plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
ax.grid(False)

# VIF bar chart
ax = axes[1]
ax.set_title("Variance Inflation Factor (VIF)", color=YELLOW)
vif_labels = list(vif_vals.keys())
vif_scores = list(vif_vals.values())
bar_colors = [GREEN if v < 1 else YELLOW if v < 5 else RED for v in vif_scores]
bars = ax.barh(vif_labels, vif_scores, color=bar_colors, edgecolor="#0d0d1a",
               height=0.5, alpha=0.9)
ax.axvline(5, color=RED,    lw=2, linestyle="--", label="VIF = 5  (Severe)")
ax.axvline(1, color=YELLOW, lw=2, linestyle=":",  label="VIF = 1  (Moderate)")
for bar, v in zip(bars, vif_scores):
    ax.text(v + 0.1, bar.get_y() + bar.get_height()/2,
            f"{v:.1f}", va="center", fontsize=11, color="white")
ax.set_xlabel("VIF Score")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True, axis="x")

# Severity legend
for y_pos, label, color in [(0.18, "< 1  →  No multicollinearity", GREEN),
                             (0.11, "1–5  →  Moderate (common in finance)", YELLOW),
                             (0.04, "> 5  →  Severe — investigate!", RED)]:
    ax.text(0.98, y_pos, label, transform=ax.transAxes, ha="right", fontsize=8,
            color=color, bbox=dict(facecolor="#1a1a2e", edgecolor=color,
                                   boxstyle="round,pad=0.3"))

plt.tight_layout()
plt.savefig("reg04_multicollinearity_vif.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg04_multicollinearity_vif.png")


# ─── Cell 6: Homoscedasticity vs Heteroscedasticity ──────────────────────────
np.random.seed(22)
n = 200
X_h = np.random.uniform(1, 20, n)

# Homoscedastic: constant variance
Y_homo = 3 + 1.5 * X_h + np.random.normal(0, 2.5, n)

# Heteroscedastic: variance increases with X
Y_hete = 3 + 1.5 * X_h + np.random.normal(0, 0.3 * X_h, n)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("Homoscedasticity vs Heteroscedasticity — Residual Diagnostics",
             fontsize=14, color="#eeeeff")

for row, (Y, label, color, verdict) in enumerate([
    (Y_homo, "Homoscedastic (Good ✓)", GREEN,  "Constant variance — OLS is reliable"),
    (Y_hete, "Heteroscedastic (Problem ⚠)", RED, "Variance grows with X — use GLS/WLS"),
]):
    m, b, *_ = stats.linregress(X_h, Y)
    y_pred   = m * X_h + b
    resids   = Y - y_pred

    # Scatter + regression line
    ax = axes[row][0]
    ax.set_title(f"Scatter: {label}", color=color)
    ax.scatter(X_h, Y, color=color, alpha=0.45, s=25, edgecolors="none")
    xr = np.linspace(X_h.min(), X_h.max(), 200)
    ax.plot(xr, m*xr + b, color="white", lw=2.5)
    # Shade ±2 residual std (should be uniform for homo)
    if row == 0:
        ax.fill_between(xr, m*xr+b - 2*resids.std(), m*xr+b + 2*resids.std(),
                        color=GREEN, alpha=0.12, label="±2 std band (uniform)")
    else:
        spread = np.linspace(resids.std()*0.3, resids.std()*2.5, len(xr))
        ax.fill_between(xr, m*xr+b - spread, m*xr+b + spread,
                        color=RED, alpha=0.12, label="±2 std band (widening)")
    ax.set_xlabel("X"); ax.set_ylabel("Y")
    ax.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
    ax.grid(True)

    # Residual plot
    ax = axes[row][1]
    ax.set_title(f"Residuals vs Fitted: {label}", color=color)
    ax.scatter(y_pred, resids, color=color, alpha=0.45, s=25, edgecolors="none")
    ax.axhline(0, color="white", lw=2, linestyle="--")
    ax.set_xlabel("Fitted Values"); ax.set_ylabel("Residuals")
    ax.text(0.5, 0.93, verdict, transform=ax.transAxes, ha="center", fontsize=9,
            color=color,
            bbox=dict(facecolor="#1a1a2e", edgecolor=color, boxstyle="round,pad=0.4"))
    ax.grid(True)

plt.tight_layout()
plt.savefig("reg05_homo_vs_hetero.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg05_homo_vs_hetero.png")


# ─── Cell 7: GLS vs OLS — Fixing Heteroscedasticity ─────────────────────────
np.random.seed(55)
n = 150
X_gls = np.random.uniform(1, 20, n)
Y_gls = 4 + 2.0 * X_gls + np.random.normal(0, 0.5 * X_gls, n)

# OLS
m_ols, b_ols, *_ = stats.linregress(X_gls, Y_gls)

# WLS — weight by 1/X (larger X has higher variance → lower weight)
weights = 1.0 / X_gls
w       = weights / weights.sum() * len(weights)
Xd      = np.column_stack([np.ones(n), X_gls])
W       = np.diag(w)
# WLS: (X'WX)^-1 X'Wy
coeffs_wls = np.linalg.inv(Xd.T @ W @ Xd) @ (Xd.T @ W @ Y_gls)
b_wls, m_wls = coeffs_wls

y_ols  = m_ols * X_gls + b_ols
y_wls  = m_wls * X_gls + b_wls
r_ols  = Y_gls - y_ols
r_wls  = Y_gls - y_wls

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("OLS vs WLS (Weighted Least Squares) — Fixing Heteroscedasticity",
             fontsize=13, color="#eeeeff")

xr = np.linspace(X_gls.min(), X_gls.max(), 300)

# Panel 1: both lines on scatter
ax = axes[0]
ax.set_title("OLS vs WLS Regression Lines", color=BLUE)
ax.scatter(X_gls, Y_gls, color=CYAN, alpha=0.35, s=25, edgecolors="none")
ax.plot(xr, m_ols * xr + b_ols, color=RED,   lw=2.5, label=f"OLS  (ignores variance)")
ax.plot(xr, m_wls * xr + b_wls, color=GREEN, lw=2.5, label=f"WLS  (weights by 1/X)")
ax.set_xlabel("X"); ax.set_ylabel("Y")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Panel 2: OLS residuals (showing heteroscedasticity)
ax = axes[1]
ax.set_title("OLS Residuals  (heteroscedastic pattern)", color=RED)
ax.scatter(X_gls, r_ols, color=RED, alpha=0.5, s=25, edgecolors="none")
ax.axhline(0, color="white", lw=2, linestyle="--")
ax.set_xlabel("X"); ax.set_ylabel("OLS Residual")
ax.grid(True)

# Panel 3: WLS residuals (more uniform)
ax = axes[2]
ax.set_title("WLS Residuals  (more uniform spread ✓)", color=GREEN)
ax.scatter(X_gls, r_wls, color=GREEN, alpha=0.5, s=25, edgecolors="none")
ax.axhline(0, color="white", lw=2, linestyle="--")
ax.set_xlabel("X"); ax.set_ylabel("WLS Residual")
ax.text(0.5, 0.07, "Weighting reduces heteroscedasticity",
        transform=ax.transAxes, ha="center", fontsize=9, color=GREEN,
        bbox=dict(facecolor="#1a1a2e", edgecolor=GREEN, boxstyle="round,pad=0.4"))
ax.grid(True)

plt.tight_layout()
plt.savefig("reg06_ols_vs_wls.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → reg06_ols_vs_wls.png")


# ─── Cell 8: Prediction Uncertainty — R² Effect on Errors ────────────────────
np.random.seed(8)
n = 100
X_pred = np.random.uniform(0, 10, n)

scenarios = [
    (0.95, GREEN,  "High R²  (~0.95)\nSmall prediction errors"),
    (0.65, YELLOW, "Medium R²  (~0.65)\nModerate errors"),
    (0.20, RED,    "Low R²  (~0.20)\nLarge prediction errors"),
]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Prediction Uncertainty — Effect of R² on Confidence Intervals",
             fontsize=13, color="#eeeeff")

for ax, (r_target, color, title) in zip(axes, scenarios):
    noise_sd = np.sqrt((1 - r_target) / r_target) * 2
    Y_sc     = 2 + 1.5 * X_pred + np.random.normal(0, noise_sd, n)
    m_sc, b_sc, r_sc, *_ = stats.linregress(X_pred, Y_sc)
    r2_sc    = r_sc ** 2

    y_hat  = m_sc * X_pred + b_sc
    resids = Y_sc - y_hat
    se     = resids.std()

    xr = np.linspace(X_pred.min(), X_pred.max(), 300)
    yr = m_sc * xr + b_sc

    ax.scatter(X_pred, Y_sc, color=color, alpha=0.4, s=25, edgecolors="none")
    ax.plot(xr, yr, color="white", lw=2.5, label="Regression line")
    ax.fill_between(xr, yr - 1.96*se, yr + 1.96*se,
                    color=color, alpha=0.18, label="95% prediction band")
    ax.fill_between(xr, yr - se, yr + se,
                    color=color, alpha=0.25, label="68% prediction band")

    ax.set_title(title, color=color, fontsize=10)
    ax.text(0.5, 0.06, f"Actual R² = {r2_sc:.2f}   SE = {se:.2f}",
            transform=ax.transAxes, ha="center", fontsize=9, color=color,
            bbox=dict(facecolor="#1a1a2e", edgecolor=color, boxstyle="round,pad=0.4"))
    ax.set_xlabel("X"); ax.set_ylabel("Y")
    ax.legend(fontsize=7, facecolor="#1a1a2e", edgecolor="#333355")
    ax.grid(True)

plt.tight_layout()
plt.savefig("reg07_prediction_uncertainty.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg07_prediction_uncertainty.png")


# ─── Cell 9: Structural Break — Pre vs Post 2008 ─────────────────────────────
np.random.seed(99)
n_pre, n_post = 60, 60

# Pre-2008: positive relationship between leverage and returns
lev_pre  = np.random.uniform(1, 5, n_pre)
ret_pre  = 0.8 * lev_pre + np.random.normal(0, 0.8, n_pre)

# Post-2008: relationship weakens / reverses
lev_post = np.random.uniform(1, 5, n_post)
ret_post = -0.3 * lev_post + np.random.normal(0, 1.2, n_post)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Structural Break — Past Relationships May Not Hold in the Future\n"
             "(Leverage vs Returns: Pre vs Post 2008 Financial Crisis)",
             fontsize=12, color="#eeeeff")

for ax, (lev, ret, color, era, arrow) in zip(axes, [
    (lev_pre,  ret_pre,  GREEN,  "Pre-2008  (leverage → higher returns)", "up"),
    (lev_post, ret_post, RED,    "Post-2008  (relationship reverses!)",     "down"),
]):
    m, b, r, *_ = stats.linregress(lev, ret)
    r2 = r**2
    ax.scatter(lev, ret, color=color, alpha=0.55, s=40, edgecolors="none")
    xr = np.linspace(lev.min(), lev.max(), 200)
    ax.plot(xr, m*xr+b, color="white", lw=2.5, label=f"Slope = {m:.2f}")
    ax.set_title(era, color=color, fontsize=10)
    ax.set_xlabel("Financial Leverage"); ax.set_ylabel("Annual Return")
    ax.text(0.5, 0.92, f"Slope = {m:+.2f}   R² = {r2:.2f}",
            transform=ax.transAxes, ha="center", fontsize=10, color=color,
            bbox=dict(facecolor="#1a1a2e", edgecolor=color, boxstyle="round,pad=0.4"))
    ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
    ax.grid(True)

axes[1].text(0.5, 0.08,
             "⚠  Using pre-2008 model post-crisis\nwould give wrong predictions!",
             transform=axes[1].transAxes, ha="center", fontsize=9, color=YELLOW,
             bbox=dict(facecolor="#2a1a00", edgecolor=YELLOW, boxstyle="round,pad=0.5"))

plt.tight_layout()
plt.savefig("reg08_structural_break.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg08_structural_break.png")


# ─── Cell 10: Full Regression Diagnostic Dashboard ───────────────────────────
np.random.seed(44)
n = 200
X1 = np.random.normal(5, 2, n)
X2 = 0.4 * X1 + np.random.normal(0, 1.5, n)    # mildly correlated with X1
Y_diag = 3 + 1.8*X1 - 0.9*X2 + np.random.normal(0, 2.5, n)

Xd  = np.column_stack([np.ones(n), X1, X2])
c, *_ = np.linalg.lstsq(Xd, Y_diag, rcond=None)
y_hat_d = Xd @ c
resid_d = Y_diag - y_hat_d
ss_res  = np.sum(resid_d**2)
ss_tot  = np.sum((Y_diag - Y_diag.mean())**2)
r2_d    = 1 - ss_res / ss_tot

fig = plt.figure(figsize=(18, 10))
fig.suptitle("Full Regression Diagnostic Dashboard", fontsize=15, color="#eeeeff")
gs  = gridspec.GridSpec(2, 3, figure=fig, hspace=0.4, wspace=0.35)

# 1. Actual vs Predicted
ax1 = fig.add_subplot(gs[0, 0])
ax1.set_title("Actual vs Predicted  (perfect = diagonal)", color=GREEN)
ax1.scatter(y_hat_d, Y_diag, color=GREEN, alpha=0.4, s=25, edgecolors="none")
lims = [min(Y_diag.min(), y_hat_d.min()), max(Y_diag.max(), y_hat_d.max())]
ax1.plot(lims, lims, color="white", lw=2, linestyle="--", label="Perfect fit")
ax1.set_xlabel("Predicted Y"); ax1.set_ylabel("Actual Y")
ax1.text(0.05, 0.92, f"R² = {r2_d:.3f}", transform=ax1.transAxes,
         fontsize=10, color=GREEN,
         bbox=dict(facecolor="#1a1a2e", edgecolor=GREEN, boxstyle="round,pad=0.4"))
ax1.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
ax1.grid(True)

# 2. Residuals vs Fitted
ax2 = fig.add_subplot(gs[0, 1])
ax2.set_title("Residuals vs Fitted Values", color=BLUE)
ax2.scatter(y_hat_d, resid_d, color=BLUE, alpha=0.4, s=25, edgecolors="none")
ax2.axhline(0, color="white", lw=2, linestyle="--")
ax2.set_xlabel("Fitted Values"); ax2.set_ylabel("Residuals")
ax2.grid(True)

# 3. Q-Q plot of residuals
ax3 = fig.add_subplot(gs[0, 2])
ax3.set_title("Q-Q Plot of Residuals", color=PURPLE)
(osm, osr), (slope_q, intercept_q, _) = stats.probplot(resid_d, dist="norm")
ax3.scatter(osm, osr, color=PURPLE, alpha=0.5, s=25, edgecolors="none")
xr_q = np.array([osm.min(), osm.max()])
ax3.plot(xr_q, slope_q*xr_q + intercept_q, color="white", lw=2)
ax3.set_xlabel("Theoretical Quantiles"); ax3.set_ylabel("Sample Quantiles")
ax3.grid(True)

# 4. Residual histogram
ax4 = fig.add_subplot(gs[1, 0])
ax4.set_title("Residual Distribution", color=CYAN)
ax4.hist(resid_d, bins=25, density=True, color=CYAN, alpha=0.55,
         edgecolor="#0d0d1a")
x_r = np.linspace(resid_d.min(), resid_d.max(), 200)
ax4.plot(x_r, stats.norm.pdf(x_r, resid_d.mean(), resid_d.std()),
         color="white", lw=2, label="Normal fit")
ax4.set_xlabel("Residual"); ax4.set_ylabel("Density")
ax4.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
ax4.grid(True)

# 5. Scale-location (√|residuals| vs fitted)
ax5 = fig.add_subplot(gs[1, 1])
ax5.set_title("Scale-Location  (homoscedasticity check)", color=ORANGE)
ax5.scatter(y_hat_d, np.sqrt(np.abs(resid_d)), color=ORANGE, alpha=0.4,
            s=25, edgecolors="none")
ax5.set_xlabel("Fitted Values"); ax5.set_ylabel("√|Residuals|")
ax5.grid(True)

# 6. Summary stats panel
ax6 = fig.add_subplot(gs[1, 2])
ax6.set_facecolor("#0d0d1a"); ax6.axis("off")
ax6.set_title("Regression Summary", color=YELLOW)
sw_stat, sw_p = stats.shapiro(resid_d[:5000])
summary_lines = [
    (f"R²  =  {r2_d:.4f}", GREEN, 11, "bold"),
    (f"Intercept (a)  =  {c[0]:.3f}", BLUE,   10, "normal"),
    (f"b₁ (X₁ coeff)  =  {c[1]:.3f}", BLUE,   10, "normal"),
    (f"b₂ (X₂ coeff)  =  {c[2]:.3f}", BLUE,   10, "normal"),
    ("", "#eeeeff", 8, "normal"),
    (f"Residual std  =  {resid_d.std():.3f}", CYAN, 10, "normal"),
    (f"Shapiro-Wilk p  =  {sw_p:.4f}", GREEN if sw_p > 0.05 else RED, 10, "normal"),
    ("", "#eeeeff", 8, "normal"),
    ("Checklist:", YELLOW, 10, "bold"),
    (f"  Linearity:          {'✓' if r2_d > 0.3 else '?'}", GREEN, 9, "normal"),
    (f"  Normal residuals:   {'✓' if sw_p > 0.05 else '⚠'}", GREEN if sw_p > 0.05 else RED, 9, "normal"),
    (f"  Homoscedasticity:   {'✓ check plot'}", CYAN, 9, "normal"),
]
y_pos = 0.97
for line, color, fs, fw in summary_lines:
    ax6.text(0.05, y_pos, line, transform=ax6.transAxes, fontsize=fs,
             color=color, fontweight=fw, va="top")
    y_pos -= 0.076

plt.savefig("reg09_diagnostic_dashboard.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → reg09_diagnostic_dashboard.png")


print("\n" + "="*58)
print("  ALL REGRESSION VISUALIZATIONS COMPLETE")
print("="*58)
print("  reg01 — Dummy variables: bond credit ratings")
print("  reg02 — Interaction effects: Growth × Size on P/E")
print("  reg03 — Linear vs nonlinear + log transformation")
print("  reg04 — Multicollinearity: heatmap + VIF scores")
print("  reg05 — Homoscedasticity vs heteroscedasticity")
print("  reg06 — OLS vs WLS (fixing heteroscedasticity)")
print("  reg07 — Prediction uncertainty vs R²")
print("  reg08 — Structural break: pre vs post 2008")
print("  reg09 — Full diagnostic dashboard (6-panel)")
print("="*58)
```

[
H_0: \text{Coefficient} = 0

If the t-statistic is large and the p-value is small, we:

**Reject the hypothesis that the coefficient is zero.**

# 3. Overall Regression vs Individual Variables

There are **two levels of testing in regression**.

---

## Level 1: Does the Entire Model Work?

We check the **F-statistic** in the ANOVA table.

The F-statistic answers:

> Does the regression model as a whole explain variation in the dependent variable?

If the F-statistic is significant:

The **overall regression works**.

## Level 2: Do Individual Variables Matter?

Even if the regression works overall, **not every independent variable may be useful**.

Example:

You might have **5 independent variables**.

But after testing:

* 3 variables are significant
* 2 variables are not

In this case:

You should **remove the insignificant variables**.

Why?

Because they **add noise and weaken the model**.

A regression becomes stronger when we:

* keep statistically significant variables
* remove statistically weak variables

If a variable has:

[
|t| < 2

It is usually considered **statistically insignificant**.

Removing such variables improves the **signal-to-noise ratio**.

One of the biggest mistakes in regression analysis is using **too many variables with too little data**.

Example:

Sample size = **25 observations**

Independent variables = **5**

This is dangerous.

Why?

Because the regression will **overfit the data**.

The results may look impressive but will be **statistically unreliable**.

A practical guideline is:

For every **10–20 observations**, add **one independent variable**.

Examples:

| Sample size | Maximum variables |
| ----------- | ----------------- |
| 10          | 1 variable        |
| 50          | 2–3 variables     |
| 100         | 4–5 variables     |

This prevents **overfitting**.

# 1. Choosing Dependent and Independent Variables

Welcome back.

Today we continue our discussion of **regression analysis**. Let’s begin with a very basic but important question.

Whenever we run a regression, we need:

* **one dependent variable**
* **one or more independent variables**

The dependent variable is easy to choose.

It is simply **the variable you want to explain**.

Examples:

* Explaining **differences in P/E ratios across companies**
* Explaining **changes in interest rates over time**
* Explaining **stock returns**

---

## The Hard Part: Choosing Independent Variables

How do we decide which variables should explain the dependent variable?

There are **two approaches**.

---

# 2. Approach 1 — Pure Statistical Selection

The first approach is purely statistical.

You collect data on many variables and then use statistics to determine:

> Which variables have the strongest relationship with the dependent variable.

Some statistical programs even automate this process using **stepwise regression**.

Stepwise regression works like this:

1. Pick the variable with the **highest correlation**
2. Add the next best variable
3. Continue adding variables

This creates a regression automatically.

---

### The Problem: Data Mining

This statistical approach can be dangerous.

Why?

Because it opens the door to **data mining**.

Data mining means:

Searching through many variables until you find something that gives you a **high R²**, even if the relationship makes **no economic sense**.

So you might find variables that statistically explain stock prices but have **no logical reason** to affect them.

---

# 3. Better Approach — Start with Economic Logic

In finance and investing, regressions should begin with:

* economic intuition
* financial theory
* common sense

First you ask:

> What variables *should* logically affect the dependent variable?

Then you test whether the data supports your idea.

---

# 4. Example: Explaining P/E Ratios

Suppose we want to explain **differences in P/E ratios across companies**.

What factors should affect the P/E ratio?

To answer this, we start with a simple valuation model.

The model is called the:

Gordon Growth Model

---

# 5. Gordon Growth Model

The Gordon model values a stock as:

[
Price = \frac{Dividend_{next\ year}}{Cost\ of\ Equity - Growth}
]

This model applies to **stable companies that pay dividends**.

---

### Converting to a P/E Formula

Divide both sides by **earnings per share (EPS)**.

Then:

* Price / Earnings = **P/E ratio**
* Dividend / Earnings = **Payout ratio**

So the equation becomes:

[
PE = \frac{Payout\ Ratio}{Cost\ of\ Equity - Growth}
]

---

# 6. What This Model Predicts

From this equation we learn three things.

---

### 1. Payout Ratio

Companies paying **higher dividends** should have **higher P/E ratios**.

---

### 2. Growth

Companies with **higher expected growth** should have **higher P/E ratios**.

---

### 3. Risk

Riskier companies have higher **cost of equity**, so they should have **lower P/E ratios**.

---

# 7. Building the Regression

Now we can build a regression to explain P/E ratios.

Dependent variable:

P/E ratio

Independent variables:

1. Payout ratio
2. Growth rate
3. Risk

But we must measure these variables using **proxies**.

---

### Proxies Used

Growth → analyst forecast of **earnings growth**

Risk → **beta**

Beta measures how sensitive a stock is to market movements in the
Capital Asset Pricing Model.

---

# 8. Running the Regression

After running the regression we obtain:

R² ≈ **40%**

This means:

The three variables explain **40% of the variation in P/E ratios**.

Is that disappointing?

No.

Regression results reflect **reality in the data**.

---

# 9. Interpreting Coefficients

The regression gives us three coefficients.

---

### Intercept

Intercept ≈ **4.10**

This is simply a **baseline P/E ratio**.

---

### Payout Ratio

Coefficient is **positive**.

Higher payout → higher P/E.

This matches our theoretical prediction.

---

### Growth Rate

Coefficient is **positive**.

Higher growth → higher P/E.

Again consistent with theory.

---

### Beta (Risk)

Unexpectedly, the coefficient is **positive**.

Theory predicted the opposite.

But remember:

**Data may contradict theory.**

That happens in real-world finance.

---

# 10. Checking Statistical Significance

Before trusting a regression, we must check if coefficients are **statistically different from zero**.

We use:

* **t-statistic**
* **p-value**

Results:

| Variable | Significance    |
| -------- | --------------- |
| Payout   | Significant     |
| Growth   | Significant     |
| Beta     | Not significant |

---

### What Should We Do?

Remove **beta** from the regression.

Why?

Because statistically insignificant variables only **add noise**.

After removing beta:

* R² drops slightly
* Remaining variables become stronger predictors.

---

# 11. Multicollinearity

In multiple regression, independent variables should ideally be **independent of each other**.

But in practice, they often are not.

Example relationships:

* High payout companies usually have **low growth**
* Mature firms tend to have **lower risk**

This creates **multicollinearity**.

Multicollinearity means:

Independent variables are correlated with each other.

---

### Why This Is a Problem

Multicollinearity can make coefficients:

* unstable
* difficult to interpret

But predictions can still work.

So sometimes we simply **live with it**.

---

# 12. Improving the Regression

If you want to improve your regression, you can:

---

### 1. Use Better Proxies

Maybe our variables do not perfectly measure the concepts.

Example:

Growth could be measured using:

* historical growth
* revenue growth
* analyst forecasts

Different proxies may improve results.

---

### 2. Add New Variables

But only if you have **economic reasoning**.

Example:

Technology firms often have many **intangible assets**.

You might include a **tech industry dummy variable**.

Never add variables just to increase **R²**.

---

### 3. Check for Outliers

Outliers can distort regressions.

Example:

A stock with:

P/E = **10,000**

This extreme observation can affect the regression.

Solutions:

* Cap extreme values
* Remove the observation (very cautiously)

---

### 4. Weighted Regressions

Sometimes larger companies behave more predictably.

Instead of weighting all firms equally, we may weight them by **market capitalization**.

This creates a **weighted least squares regression**.

---

# 13. Data Mining and P-Hacking

Modern finance has enormous datasets.

This increases the risk of **data mining**.

Researchers may search endlessly for relationships that look statistically significant.

---

Another related issue is **p-hacking**.

Researchers manipulate:

* sample periods
* included companies

until they obtain low p-values.

Even though the relationship may be meaningless.

---

# 14. Statistical vs Economic Significance

A result can be statistically significant but still **useless economically**.

Example:

A regression predicts tomorrow’s market return from today’s return.

It might be statistically significant.

But once we include:

* trading costs
* market frictions

the strategy might **lose money**.

So always ask:

> Does this result have **economic value**?

---

# 15. Non-Linear Relationships

Not all relationships are linear.

Example: explaining **PEG ratios**.

PEG =

[
\frac{PE}{Growth}
]

Scatter plots showed a **curved relationship**.

Instead of using growth directly, we used:

**log(growth)**

Taking the natural log compresses large numbers and often makes relationships **more linear**.

---

# 16. Sample Size Matters

Large samples are always preferable.

But sometimes we only have **small datasets**.

Rule of thumb:

* 10–15 observations → simple regression
* add **one variable per 10–15 observations**

Example:

Sample size = 50
Maximum independent variables ≈ 3

---

### Example of Overfitting

A regression with:

* 9 observations
* 4 independent variables

may show **R² = 75%**

But none of the variables are statistically significant.

This regression is **almost entirely noise**.

---

# 17. Dummy Variables

A **dummy variable** takes only two values:

0 or 1.

Dummy variables allow us to represent **categories**.

Example:

Emerging market companies vs developed market companies.

---

# 18. Telecom Example

Consider global telecom firms listed in the US through ADRs.

Some companies operate in **emerging markets**, which are riskier.

We create a variable:

Emerging Market Dummy:

* 1 → emerging market company
* 0 → developed market company

---

### Regression Model

P/E ratio explained by:

* growth rate
* emerging market dummy

---

### Results

R² ≈ **66%**

Interpretation:

Growth increases P/E ratios.

But emerging market firms have P/E ratios **13.85 lower** than comparable developed market firms.

This reflects the **higher perceived risk**.

---

# 19. Example Conclusion

A telecom firm like Indosat appeared cheap because its P/E was **7.8**.

But after controlling for:

* low growth
* emerging market risk

The predicted P/E was **6.55**.

So the stock was **not cheap after all**.

---

# 20. Final Lessons

When using regressions:

Be creative but disciplined.

Do not chase high R² values.

Focus on:

1. Economic logic
2. Statistical significance
3. Economic significance

Regression is ultimately a **tool for making predictions**, not just explaining past data.
