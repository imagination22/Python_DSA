# Statistics Lecture Notes

## Session — Descriptive (Summary) Statistics

---

# 1. Why Do We Use Summary Statistics?

Let’s begin with a simple reality.

If someone gives you **100 numbers**, it is very difficult to immediately understand what those numbers represent.

Even if you are good with numbers, looking at a long list of values rarely gives you a clear picture.

So what do statisticians try to do?

We try to **summarize large datasets using a few numbers** that capture the key characteristics of the data.

These summary numbers are called:

> **Descriptive Statistics** or **Summary Statistics**

Their purpose is to provide a **parsimonious description of data**.

---

# 2. Four Major Categories of Descriptive Statistics

In statistics, descriptive statistics fall into **four main groups**.

### 1. Measures of Centrality

These describe the **typical value** of the data.

Example question:

> What value represents the center of the dataset?

---

### 2. Measures of Dispersion

These describe **how spread out the data is** around the central value.

Example question:

> Do observations vary widely or stay close to the average?

---

### 3. Measures of Symmetry

These describe whether the data is **balanced or skewed** around the center.

Example question:

> Are extreme values more common on the upside or downside?

---

### 4. Measures of Extremes

These describe how often **very large or very small observations occur**.

Example question:

> Does the dataset produce many outliers?

---

So the four categories are:

1. **Centrality**
2. **Dispersion**
3. **Symmetry**
4. **Extremes**

---

# 3. Measures of Centrality

Measures of centrality try to answer:

> What is the **typical value** in this dataset?

There are three main measures.

---

# 4. Mean (Average)

The most commonly used measure of centrality is the **mean**, also known as the **average**.

### Formula

[
Mean = \frac{\sum X_i}{n}
]

Where:

* (X_i) = each observation
* (n) = number of observations

---

### Example

Suppose we have **100 years of stock returns**.

To compute the average return:

1. Add all 100 returns
2. Divide by 100

This gives the **average annual return**.

---

# 5. Weighted Mean

Sometimes we do not want every observation to have equal importance.

In that case we use a **weighted average**.

### Example

You may want to give:

* **More weight to recent years**
* **Less weight to older data**

The formula becomes:

[
Weighted\ Mean = \sum w_i X_i
]

where (w_i) are weights.

---

# 6. Median

The **median** is the **50th percentile** of the dataset.

It is the **middle value** when the data is ranked.

---

### Example

Suppose we have **101 observations**.

Steps:

1. Sort observations from smallest to largest.
2. Find the **51st observation**.

That value is the **median**.

Why?

Because:

* 50 observations are below it
* 50 observations are above it

---

### Why Median Matters

The median is especially useful when the data is **asymmetric**.

Outliers can distort the mean, but the **median is more robust**.

---

# 7. Mode

The **mode** is the value that occurs **most frequently**.

---

### Example

Suppose we measure the heights of **101 people**.

If the most common height is:

5 ft 9 inches

Then that is the **mode**.

---

### When Mode Is Useful

Mode is more useful when:

* Data is **categorical**
* Data is **discrete**

Example:

* Most common exam grade
* Most common height range

In finance, the **mode is rarely used**, because financial data is usually **continuous**.

---

# 8. Comparing Mean, Median, and Mode

Each measure has strengths and weaknesses.

| Measure | Strength               | Weakness                 |
| ------- | ---------------------- | ------------------------ |
| Mean    | Uses all data          | Sensitive to outliers    |
| Median  | Resistant to outliers  | Ignores magnitude        |
| Mode    | Good for discrete data | Rarely useful in finance |

In finance, the most common measures are:

> **Mean and Median**

---

# 9. Measures of Dispersion

Once we know the central value, the next question is:

> How far do observations deviate from that center?

This is called **dispersion**.

---

# 10. Range

The **simplest measure of dispersion** is the range.

### Formula

[
Range = Max - Min
]

---

### Example

If stock returns vary from:

* Minimum = −20%
* Maximum = +30%

Then:

Range = 50%

---

### Problem with Range

Range uses **only two numbers**:

* Maximum
* Minimum

It ignores **all other observations**.

---

# 11. Interquartile Range (IQR)

A better variation is the **interquartile range**.

It measures the spread between:

* 25th percentile (Q1)
* 75th percentile (Q3)

### Formula

[
IQR = Q3 - Q1
]

This captures the **middle 50% of the data**.

---

# 12. Standard Deviation

The most important dispersion measure is **standard deviation**.

It measures the **average distance of observations from the mean**.

---

### Steps to Calculate

1. Compute the mean.
2. Subtract the mean from each observation.
3. Square the differences.
4. Average the squared differences.

This gives the **variance**.

Then take the square root.

[
Standard\ Deviation = \sqrt{Variance}
]

---

# 13. Sample vs Population Variance

If we measure the entire population:

[
Variance = \frac{\sum (X_i - \mu)^2}{n}
]

If we measure a **sample**:

[
Variance = \frac{\sum (X_i - \bar{X})^2}{n-1}
]

Why divide by **n − 1**?

Because estimating the mean **uses one degree of freedom**.

---

# 14. Units of Standard Deviation

Standard deviation always has the **same units as the data**.

Example:

| Data Type   | Unit of SD   |
| ----------- | ------------ |
| Stock price | Dollars      |
| Returns     | Percent      |
| Weight      | Pounds or kg |

---

### Important Insight

This means we must be careful when comparing datasets measured in **different units**.

---

# 15. Coefficient of Variation (CV)

To compare dispersion across datasets with different units, we use:

> **Coefficient of Variation**

### Formula

[
CV = \frac{Standard\ Deviation}{Mean}
]

This removes the units.

Now we can compare:

* pounds vs kilograms
* dollars vs percentages

---

# 16. Standard Error

A related concept is the **standard error**.

Important distinction:

| Statistic          | Measures                |
| ------------------ | ----------------------- |
| Standard Deviation | variability of data     |
| Standard Error     | accuracy of sample mean |

---

### Formula

[
Standard\ Error = \frac{SD}{\sqrt{n}}
]

Where:

* SD = standard deviation
* n = sample size

---

### Why Standard Error Matters

We use samples to estimate **population values**.

Standard error tells us:

> How close the sample mean is to the population mean.

---

# 17. Confidence Intervals

Using the **Central Limit Theorem**, we can estimate ranges.

Example:

Sample mean = 10%
Standard error = 2%

Then:

| Confidence | Range    |
| ---------- | -------- |
| 68%        | 8% – 12% |
| 95%        | 6% – 14% |
| 99%        | 4% – 16% |

---

# 18. Measures of Symmetry

Now we examine whether the data is **balanced around the mean**.

A perfectly balanced dataset is **symmetric**.

Example:

The **normal distribution**.

---

# 19. Skewness

When distributions are asymmetric, we measure **skewness**.

---

### Positive Skew

Right tail is longer.

Characteristics:

* Large positive outliers
* Mean > Median

Example:

Stock prices.

---

### Negative Skew

Left tail is longer.

Characteristics:

* Large negative outliers
* Mean < Median

Example:

Insurance losses.

---

# 20. Why Skewness Matters

Skewed distributions distort averages.

Example:

Positive skew → mean becomes **too high**.

Negative skew → mean becomes **too low**.

---

# 21. Bounded Distributions

Many financial variables have **limits**.

Example:

Stock price:

* Minimum = 0
* Maximum = unlimited

This naturally creates **positive skewness**.

---

Example:

Profit margin:

* Maximum = 100%
* Minimum = −∞

This can create **negative skewness**.

---

# 22. Measures of Extremes

Finally we examine **extreme observations**.

These are called **outliers**.

To measure extreme values we use:

> **Kurtosis**

---

# 23. Kurtosis

Kurtosis measures how likely **extreme values** are.

Compared to a normal distribution:

---

### High Kurtosis

Called **Leptokurtic**.

Features:

* Fat tails
* More extreme outcomes

Common in finance.

---

### Low Kurtosis

Called **Platykurtic**.

Features:

* Thin tails
* Fewer extreme values

---

# 24. Important Insight

Variance and kurtosis measure **different things**.

You can have:

* High variance, low kurtosis
* Low variance, high kurtosis
* High variance, high kurtosis

---

# 25. Final Summary

Descriptive statistics fall into four groups:

### Measures of Centrality

* Mean
* Median
* Mode

---

### Measures of Dispersion

* Range
* Interquartile Range
* Standard Deviation
* Coefficient of Variation
* Standard Error

---

### Measures of Symmetry

* Skewness

---

### Measures of Extremes

* Kurtosis

---

# Final Intuition

Descriptive statistics allow us to transform **large datasets into a few meaningful numbers** that describe:

* the center of the data
* the variability
* the shape
* the probability of extreme outcomes

These statistics form the **foundation of statistical analysis in finance, economics, and data science**.

# Statistics & Finance Lecture Notes

## Session: Applications of Descriptive Statistics in Investing

---

# 1. Introduction

In this session we move from **theory of descriptive statistics** to **real-world financial data**.

Instead of discussing statistics abstractly, we will apply them to **financial markets**.

We will examine returns from **three major asset classes**:

1. **U.S. Stocks**
2. **10-Year U.S. Treasury Bonds**
3. **Treasury Bills (T-Bills)**

These represent different levels of **risk and return**.

---

# 2. Computing Returns for Asset Classes

To analyze performance, we compute **annual percentage returns** from **1928 to 2020**.

This gives us **93 years of data**.

### A. Treasury Bills (T-Bills)

T-bills are the simplest.

* They have **no price change**
* Investors simply receive the **interest rate (yield)**.

Therefore:

Return on T-Bill = T-Bill Interest Rate

---

### B. Treasury Bonds (T-Bonds)

For bonds, returns have **two components**:

1. **Coupon payment** (interest received)
2. **Price change of the bond**

Bond prices change when **interest rates change**.

Therefore:

Total Bond Return = Coupon + Price Change

---

### C. Stocks

Stock returns also have **two components**:

1. **Dividends received**
2. **Change in stock price**

Therefore:

Total Stock Return = Dividends + Price Change

---

# 3. Volatility Across Asset Classes

When we plot the annual returns from **1928–2020**, we observe:

* **Stocks → Most volatile**
* **Bonds → Moderate volatility**
* **T-Bills → Very stable**

But graphs alone are not enough.

To extract deeper insights, we use **descriptive statistics**.

---

# 4. Average Returns (Mean)

First, we compute **average annual returns**.

### Formula

[
Mean = \frac{\sum Returns}{Number\ of\ Years}
]

Since we have **93 years**, we divide by **93**.

### Results

Average Returns (1928–2020)

| Asset   | Average Return |
| ------- | -------------- |
| Stocks  | **11.64%**     |
| T-Bonds | **5.21%**      |
| T-Bills | **3.36%**      |

---

### Initial Conclusion

If we only look at averages:

Stocks > Bonds > T-Bills

This suggests stocks are the **best investment**.

But this ignores **risk**.

---

# 5. Measuring Risk: Standard Deviation

To measure **risk**, we compute **standard deviation**.

### Steps

1. Subtract each year's return from the average.
2. Square the difference.
3. Sum all squared differences.
4. Divide by **(n – 1)** since this is a **sample**.
5. Take the square root.

This gives **standard deviation**.

---

### Why divide by (n – 1)?

Because the data is a **sample**, not the entire population.

Using the sample mean costs **one degree of freedom**.

---

### Results

| Asset   | Standard Deviation |
| ------- | ------------------ |
| Stocks  | **19.49%**         |
| Bonds   | Lower              |
| T-Bills | Very low           |

---

### Interpretation

Stocks have:

* **Highest returns**
* **Highest risk**

T-Bills have:

* **Lowest returns**
* **Lowest risk**

Investment decisions therefore involve a **risk-return tradeoff**.

---

# 6. Standard Error

We can convert standard deviation into **standard error**.

### Formula

[
Standard\ Error = \frac{Standard\ Deviation}{\sqrt{n}}
]

For stocks:

[
SE = \frac{19.49}{\sqrt{93}} = 2.02%
]

---

# 7. Using Standard Error to Predict Returns

Assuming the **future resembles the past**, we can estimate ranges.

### 67% Confidence Interval

Mean ± 1 Standard Error

[
11.64% ± 2.02%
]

Range:

9.62% to 13.66%

---

### 95% Confidence Interval

Mean ± 2 Standard Errors

[
11.64% ± 4.04%
]

Range:

7.60% to 15.68%

---

### Insight

Predictions for stocks are **less precise** because of higher volatility.

Predictions for T-bills and bonds are **more precise**.

---

# 8. Median Return

The **median** is the **middle observation**.

With **93 observations**:

Median = **47th observation when sorted**

---

### Results

Stock median return = **14.22%**

Observation:

Median > Mean

---

### Why does this happen?

Because the distribution is **skewed**.

In this case:

Stocks show **slight negative skewness**.

This means:

* Extreme negative years occur more often than extreme positive years.

---

# 9. Skewness

Skewness measures **asymmetry** in the distribution.

### Negative Skew

* Long tail on the **left**
* Extreme losses occur occasionally

Stocks historically show **slight negative skewness**.

---

### Positive Skew

* Long tail on the **right**
* Rare large positive outcomes

T-Bills and T-Bonds show **positive skewness**.

---

# 10. Kurtosis

Kurtosis measures the **likelihood of extreme outcomes**.

For a **normal distribution**, kurtosis = **3**.

Stock returns historically have kurtosis **close to 3**.

This means:

Stock returns are **approximately normally distributed**.

---

# 11. Minimum, Maximum, and Range

We can also look at **extreme values**.

### Stocks (1928–2020)

Worst year: **−44%**

Best year: **+53%**

Range:

[
53 - (-44) = 96%
]

This shows how volatile stocks can be.

---

# 12. Interquartile Range (IQR)

Instead of extreme values, we can look at the **middle 50% of data**.

### Formula

[
IQR = Q3 - Q1
]

Where:

* Q1 = 25th percentile
* Q3 = 75th percentile

For stocks:

* Q1 ≈ −1.19%
* Q3 ≈ 25.06%

IQR = **26.25%**

Again confirming stocks are **highly volatile**.

---

# 13. Key Takeaways from 93 Years of Data

From **1928–2021**:

1. **Stocks produced the highest returns**
2. **Stocks also carried the highest risk**
3. Stock returns are **approximately normal**
4. Stocks show **slight negative skewness**
5. Bonds and T-bills show **positive skewness**
6. Dispersion measures confirm stocks are most volatile

---

# 14. Example 2: Price-to-Earnings (P/E) Ratios

Now we examine **P/E ratios** across companies.

### Formula

[
PE = \frac{Stock\ Price}{Earnings\ per\ Share}
]

This is one of the most widely used metrics in investing.

---

# 15. Three Types of P/E Ratios

### 1. Current P/E

Price ÷ earnings from the **most recent fiscal year**

---

### 2. Trailing P/E

Price ÷ earnings from the **last four quarters**

---

### 3. Forward P/E

Price ÷ **expected earnings** in the next four quarters

---

# 16. Distribution of P/E Ratios

When we plot P/E ratios across companies, we get a **histogram**.

Observation:

* Peak on the **left**
* Long **right tail**

This means P/E ratios are **positively skewed**.

---

# 17. Mean vs Median for P/E Ratios

Average current P/E ≈ **110**

Median P/E ≈ **18**

Huge difference!

---

### Why?

Because of **extreme outliers**.

Some companies had P/E ratios of:

* 25,000
* 36,000
* 42,000

These pull the **average upward**.

Therefore:

For skewed distributions, **median is more reliable than mean**.

---

# 18. Sampling Bias in P/E Ratios

There were **7,584 public companies** in the U.S.

But P/E ratios could only be computed for about **2,780 companies**.

Why?

Because companies with **negative earnings** have **no meaningful P/E ratio**.

Thus:

Almost **two-thirds of companies were excluded**.

This creates **sampling bias**.

The remaining companies are typically:

* Older
* Mature
* Profitable

---

# 19. Example 3: Cost of Capital

Cost of capital represents the **minimum return required for investments**.

Companies use it as a **hurdle rate** for projects.

A histogram of cost of capital across companies shows:

* Slight positive skew
* Less extreme dispersion than P/E ratios

---

# 20. Global Cost of Capital

Example:

African/Middle Eastern companies:

Average cost of capital = **6.69%**

Median = **6.27%**

Some regions show **positive skew**, others **negative skew**.

---

# 21. Distribution Insight

For U.S. companies:

Median cost of capital ≈ **7.4%**

50% of companies fall between:

**6% – 8.25%**

80% fall between:

**5% – 9%**

---

### Important Lesson

Since the range is narrow, analysts should **not spend excessive time refining cost of capital estimates**.

Small changes rarely affect valuation significantly.

---

# 22. Example 4: Dividend Yield

Dividend Yield measures the **income portion of stock returns**.

### Formula

[
Dividend\ Yield = \frac{Dividend}{Stock\ Price}
]

Example:

Dividend = $2
Stock price = $50

Dividend yield = **4%**

---

# 23. Distribution of Dividend Yields

Among **all companies**, the median dividend yield is **0%**.

Why?

Because **74% of companies pay no dividends**.

---

# 24. Dividend-Paying Companies

Among only **dividend-paying firms**:

Average dividend yield ≈ **4%**

Median ≈ **2.43%**

---

### Important Insight

Comparisons must use the **correct sample**.

Comparing a dividend-paying stock to **all companies** can be misleading.

---

# 25. Final Lesson: Why Descriptive Statistics Matter

Descriptive statistics help investors:

1. Understand **risk and return**
2. Identify **skewed distributions**
3. Detect **outliers**
4. Avoid **misleading averages**
5. Recognize **sampling bias**
6. Focus attention on variables that matter

---

✔ In investing, statistics are not just theoretical tools.

They help us **interpret financial data and make better decisions**.
