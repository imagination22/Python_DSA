# Lecture Notes – Probability in Finance & Decision Making

---

# 1. What is Probability?

Welcome back.

In this session we will talk about **probabilities**—how we measure them, how we use them, and why they matter for forecasting the future.

Let’s start with a simple idea.

We live in a world with **very few certainties**. Most events are uncertain.

A **probability** measures the **likelihood that a particular outcome will occur**.

---

## Range of Probabilities

Probabilities always lie between **0 and 1**.

* **0** → the event is impossible
* **1** → the event is certain

Most interesting events lie **between 0 and 1**.

For example:

* Probability > 0.5 → event is **more likely to occur**
* Probability < 0.5 → event is **less likely to occur**
* Probability = 0.5 → **even odds**

---

# 2. Why Probability Matters in Finance

In finance and investing, we constantly estimate probabilities.

Some events are **discrete**.

Discrete events have **only two outcomes**.

Examples:

* Will a company **go bankrupt** in the next five years?
* Will a company become an **acquisition target**?

These events either **happen or do not happen**.

---

## Continuous Events

Other events are **continuous**.

Example:

A company currently earns **$500 million**.

We might ask:

* What is the probability earnings exceed **$1 billion** in five years?
* What is the probability earnings become **negative**?

These questions involve **ranges of outcomes**, not just yes/no.

---

# 3. Two Philosophies of Probability

There are two major ways to interpret probability.

---

# 3.1 Frequentist View

The first is the **frequency view** of probability.

In this view, probability measures **how often an event occurs if we repeat it many times**.

Example: coin toss.

If you flip a fair coin thousands of times:

* heads ≈ 50%
* tails ≈ 50%

Eventually, the **true probability reveals itself**.

This approach works well when events are **repeatable**.

---

# 3.2 Subjective (Bayesian) View

The second view is the **subjective view**, also known as the **Bayesian approach**.

This view says:

Even if an event cannot be repeated many times, we can still assign a **belief-based probability**.

This is very common in finance.

Example:

What is the probability the **stock market rises next year**?

We cannot repeat the next year thousands of times.

Therefore, investors use **judgment and information** to estimate probabilities.

---

## Why Investors Disagree

Because probabilities are subjective:

Different investors can estimate **different probabilities for the same event**.

Example:

10,000 investors analyze the same company.

They may all reach **different conclusions** about the probability the stock will rise.

This is normal under the Bayesian view.

---

# 4. Types of Probabilities

Two special types appear frequently.

---

# 4.1 Cumulative Probability

Cumulative probability measures the likelihood of **multiple events occurring in sequence**.

Example:

Suppose an investor randomly selects stocks.

What is the probability they beat the market **10 years in a row**?

Even if the chance each year is **50%**, it is still possible by luck.

This explains why:

Some investors outperform for long periods **purely by chance**.

---

# 4.2 Conditional Probability

Conditional probability measures the probability of an event **given that another event has already occurred**.

Example:

Suppose the stock market rose sharply today.

We might ask:

What is the probability the market rises **tomorrow given today's rise**?

This is conditional probability.

---

# 5. Basic Rules of Probability

There are several fundamental rules.

---

# Rule 1: Probability Bounds

Probabilities must always satisfy:

[
0 \leq P(x) \leq 1
]

Values such as:

* 120%
* −10%

are impossible.

---

# Rule 2: Total Probability

If we list **all possible outcomes**, their probabilities must sum to **1**.

Example:

* Market up tomorrow → 60%
* Market down tomorrow → 40%

Total = 100%.

---

# Rule 3: Complement Rule

If the probability of event X is known:

[
P(\text{not X}) = 1 - P(X)
]

Example:

Probability market rises = 60%.

Then:

Probability market does **not rise** = 40%.

---

# Rule 4: Addition Rule

Probability of **X or Y happening**:

[
P(X \cup Y) = P(X) + P(Y) - P(X \cap Y)
]

Example:

Suppose:

* Probability Google stock rises = 60%
* Probability Apple stock rises = 65%
* Probability both rise together = 50%

Then:

Probability at least one rises:

[
0.60 + 0.65 - 0.50 = 0.75
]

So there is a **75% probability** that either stock rises.

---

# Rule 5: Multiplication Rule

To find the probability of **both events happening**.

If events are independent:

[
P(X \cap Y) = P(X) \times P(Y)
]

Example:

* Probability Yankees win = 50%
* Probability you go out for dinner = 25%

Then:

Probability both occur:

[
0.5 \times 0.25 = 0.125
]

So there is a **12.5% probability**.

---

## Dependent Events

If events depend on each other:

[
P(X \cap Y) = P(X) \times P(Y|X)
]

Where (P(Y|X)) is the **conditional probability**.

---

# 6. Bayesian Thinking

The Bayesian approach comes from the work of:

Thomas Bayes

His main idea:

Probability estimates depend on **prior knowledge**.

---

## Priors

Before making a probability estimate, we already possess **prior beliefs**.

Example:

You believe a company has:

* great management
* strong products

Therefore you estimate a **high probability** its stock will perform well.

---

## Posterior Probabilities

When new information arrives, we update our probability.

Example:

Suppose the CEO is suddenly accused of fraud.

Your belief changes.

The new probability after updating information is called the **posterior probability**.

This continuous updating process is called **Bayesian inference**.

---

# 7. Probabilistic Models

Probability theory leads to several useful tools.

---

# 7.1 Probit and Logit Models

These models look similar to regression models.

But there is one key difference.

The dependent variable can only be:

* **0**
* **1**

Meaning the event either **occurs or does not occur**.

---

## Example: Predicting Acquisitions

We might predict whether a company becomes an **acquisition target**.

Dependent variable:

* 1 → company acquired
* 0 → company not acquired

Independent variables might include:

* company size
* stock performance
* profitability

The model outputs a **probability between 0 and 1**.

---

### Difference Between Probit and Logit

Both models estimate probabilities.

Difference:

* Probit uses a **normal distribution**
* Logit uses a **logistic distribution**

In practice, results are usually similar.

---

# 7.2 Bankruptcy Prediction Models

A classic use of logit models is predicting bankruptcy.

Steps:

1. Start with companies from **10 years ago**.
2. Identify which firms **went bankrupt**.
3. Label variables:

* Bankruptcy = 1
* Survival = 0

4. Add explanatory variables:

* debt levels
* earnings
* earnings volatility
* liquidity

The model produces a **probability of bankruptcy**.

---

# 8. Decision Trees

Another probabilistic tool is the **decision tree**.

Decision trees are used when **risk unfolds sequentially over time**.

---

## Example: Pharmaceutical Drug Approval

A new drug must pass **three approval stages**.

At each stage:

* success
* failure

If the drug fails at any stage, the project stops.

This creates **sequential risk**.

---

## Types of Nodes

Decision trees contain three node types.

1. **Event nodes**

   * outcomes determined by probability

2. **Decision nodes**

   * choices made by managers

3. **End nodes**

   * final outcomes

---

## Rolling Back the Tree

After building the tree, we evaluate it by **rolling it backward**.

Steps:

1. Start from final outcomes.
2. Calculate expected values.
3. Move step by step toward the present.

The final result is a **single expected value today**.

---

# 9. Scenario Analysis

Another tool is **scenario analysis**.

This is useful when outcomes depend on **macro events**.

Example:

A company’s future may depend on government policy or economic conditions.

We define several possible scenarios:

* good scenario
* neutral scenario
* bad scenario

Each scenario has:

* a probability
* a financial outcome

---

## Best Practice

Scenario analysis should include:

1. all plausible scenarios
2. probabilities for each scenario
3. expected value calculation

Otherwise scenario analysis becomes meaningless.

---

# 10. Expected Value in Scenario Analysis

For each scenario:

[
Expected\ Value = Probability \times Outcome
]

Then sum across all scenarios.

This gives an **overall expected value**.

---

# 11. Problems with Best-Case/Worst-Case Analysis

Simple best-case vs worst-case analysis is often unhelpful.

Why?

Because:

* best case → everything goes perfectly
* worst case → everything fails

Reality usually lies **somewhere in between**.

Therefore realistic scenarios are more useful.

---

Probability is central to finance because the future is uncertain.

Key ideas:

* probability rules help calculate complex outcomes
* probabilistic models help forecast events

Important tools include:

* **logit models**
* **probit models**
* **decision trees**
* **scenario analysis**

These tools allow investors to **quantify uncertainty and make better decisions**.

# 1. Random Walk in Markets

Let us begin with one of the **most fundamental ideas in finance** — the **Random Walk Hypothesis**.

When we say that **markets follow a random walk**, we are essentially saying:

* On any given day

  * Probability market goes **up = 50%**
  * Probability market goes **down = 50%**

In other words, **today’s movement gives no information about tomorrow’s movement**.

---

## Example: 100 Trading Days

If markets truly follow a random walk:

* Out of **100 days**
* We expect approximately:

| Outcome   | Expected Days |
| --------- | ------------- |
| Up days   | 50            |
| Down days | 50            |

But in reality, you might observe:

* 56 up days
* 44 down days

Does that mean the market is **not random**?

**Not necessarily.**

Random processes naturally produce variation.

---

# 2. Standard Error for Probabilities

To judge whether a deviation from 50-50 is meaningful, we calculate the **standard error**.

Formula:

[
SE = \sqrt{\frac{p(1-p)}{n}}
]

Where:

* **p** = probability of success
* **n** = sample size

---

## Example: 100 Trading Days

Assume:

* ( p = 0.5 )
* ( n = 100 )

[
SE = \sqrt{\frac{0.5 \times 0.5}{100}}
]

[
SE = \sqrt{0.0025}
]

[
SE = 0.05 = 5%
]

But the professor simplified it to **2.5% because he was referring to half-range around the mean**.

---

### 95% Confidence Range

Remember:

**95% confidence interval ≈ ±2 standard errors**

So:

[
50% \pm (2 \times 2.5%)
]

Range:

**45% to 55%**

Meaning:

Even in a perfectly random market, **up days can range from 45% to 55%** just due to randomness.

---

# 3. Real Example — S&P 500 (2016–2020)

Now let’s test this with real data.

Number of trading days:

**1257**

Observed:

| Outcome   | Days | Percentage |
| --------- | ---- | ---------- |
| Up days   | 700  | 55.69%     |
| Down days | 557  | 44.31%     |

---

### Calculate Standard Error

Assume random walk:

[
p = 0.5
]

[
SE = \sqrt{\frac{0.5 \times 0.5}{1257}}
]

Result:

**≈ 1.41%**

---

### 95% Range

[
50% \pm (2 \times 1.41)
]

Range:

**47.18% – 52.82%**

But actual result:

**55.69%**

This is **outside the range**.

Therefore:

👉 We **reject the random walk hypothesis for this period**.

---

### Important Warning

This result applies only to **2016–2020**, which was a **strong bull market**.

So we **cannot conclude markets are predictable**.

---

# 4. Conditional Probability

Now we ask a more interesting question:

> If the market was **up yesterday**, what does that tell us about **tomorrow**?

---

## After an Up Day

From the data:

* Total up days: **700**

Next day results:

| Outcome | Days | Probability |
| ------- | ---- | ----------- |
| Up      | 367  | 52.43%      |
| Down    | 333  | 47.57%      |

This **52.43%** is **not statistically different from 50%**.

Conclusion:

**Knowing yesterday was an up day tells us almost nothing about tomorrow.**

---

## After a Down Day

Down days: **557**

Next day:

| Outcome | Days | Probability |
| ------- | ---- | ----------- |
| Up      | 332  | 59.6%       |
| Down    | 225  | 40.4%       |

Here:

**59.6% is significantly above 50%.**

Conclusion:

👉 After a **down day**, markets are **more likely to go up the next day**.

This indicates **short-term mean reversion**.

# 5. Probability of Consecutive Days

Another common question investors ask:

> What is the chance of several up days or down days in a row?

---

## Example: 3 Up Days in a Row

Probability of up day:

[
p = 0.5569
]

Three consecutive up days:

[
0.5569^3
]

Result:

**17.27%**

---

## Five Up Days in a Row

[
0.5569^5
]

Result:

**5.36%**

---

### Important Insight

Even in a **random market**, streaks are normal.

So when people say:

> “Markets went up 6 days in a row — it cannot be random!”

That statement ignores **probability mathematics**.

# 1. Purpose of Simulations

In finance and investing, we often try to **explain an output variable using several input variables**.

Example:
When valuing a company, we estimate:

* **Revenue growth**
* **Operating margins**

These inputs determine the final output:

**Company Value**

---

## The Problem: Point Estimates

Traditionally, we use **point estimates**.

Example:

| Variable         | Estimate |
| ---------------- | -------- |
| Revenue growth   | 12%      |
| Operating margin | 8%       |

But these numbers are **uncertain**.

Reality:

* Revenue growth might be **5% or 18%**
* Margins might be **6% or 10%**

Yet traditional valuation forces us to choose **one number**.

---

### Why We Used Point Estimates

Historically:

* Data was hard to obtain
* Computers were expensive
* Running thousands of calculations was impossible

So analysts had **no choice** but to use single estimates.

Today:

Computers allow us to run **thousands or millions of simulations easily**.

---

# 2. Monte Carlo Simulation Concept

Monte Carlo simulation replaces **point estimates** with **probability distributions**.

Instead of:

Revenue growth = **12%**

We say:

Revenue growth follows a **distribution**

Example:

| Possible Growth | Probability |
| --------------- | ----------- |
| 4%              | possible    |
| 12%             | most likely |
| 20%             | possible    |

---

## Simulation Process

Step-by-step:

1. Assign a **distribution** to each input variable.

Example:

* Revenue growth → distribution
* Margin → distribution

2. The computer randomly selects values.

Example:

Simulation #1:

* Growth = **18%**
* Margin = **5%**

3. Using these values, we calculate **company value**.

4. Repeat the process thousands of times.

Example:

| Simulation | Growth | Margin | Value |
| ---------- | ------ | ------ | ----- |
| 1          | 18%    | 5%     | $110  |
| 2          | 6%     | 20%    | $150  |
| 3          | 12%    | 9%     | $130  |

After many simulations we get a **distribution of values**.

---

## Final Output

Instead of one number:

**Value = $126**

We get:

**Distribution of values**

Example:

| Value Range | Probability |
| ----------- | ----------- |
| $90–$110    | 25%         |
| $110–$140   | 50%         |
| $140–$170   | 25%         |

This shows **uncertainty explicitly**.

---

# 3. Importance of the Model

Simulations only work if the **underlying model is correct**.

Important rule:

> A bad model with a good simulation still produces bad results.

Therefore:

### Guidelines

1. **Keep models simple**
2. **Avoid black-box models**
3. **Make relationships transparent**

# 5. Types of Uncertainty

Before choosing distributions, we must identify the type of uncertainty.

---

## 1. Continuous Uncertainty

Variable can take **any value within a range**.

Example:

Margins:

4% → 20%

Revenue growth:

−5% → 25%

These require **continuous distributions**.

---

## 2. Discrete Uncertainty

Only a few possible outcomes.

Example:

Regulation change:

| Outcome   |
| --------- |
| Change    |
| No change |

Values = **0 or 1**

---

## 3. Symmetric Risk

Upside and downside uncertainty are similar.

Example:

Margin expected = 25%

Possible:

* 20%
* 30%

Distribution is **symmetric**.

---

## 4. Asymmetric Risk

Risk is larger on one side.

Example:

Expected margin = 15%

Possible outcomes:

* 10%
* −5%

Downside risk is larger.

Distribution is **skewed**.

---

## 5. Extreme Outcomes

Sometimes rare events are possible.

Example:

* Market crash
* Disruptive innovation

Distributions must allow **fat tails**.

---

Choice depends on **nature of the variable**.

# 16. When to Use Simulations

Use Monte Carlo simulations when:

* There is **significant uncertainty**
* Input variables vary widely
* Decision depends on **risk assessment**

---

## When Not Necessary

For simple analyses where inputs are stable, simulations may be **unnecessary complexity**.

---

# Final Insight

Monte Carlo simulation helps analysts:

* Accept uncertainty
* Quantify uncertainty
* Make better decisions

Instead of hiding uncertainty behind a **single number**, simulations allow uncertainty to **play out mathematically**.

---

✔ Key idea:

> Simulations do not eliminate uncertainty — they help us **understand and measure it**.
