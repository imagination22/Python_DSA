# 1. From Descriptive Statistics to Distributions

In the previous session, we focused on **descriptive statistics**.

These included:

* Measures of **centrality** (mean, median)
* Measures of **dispersion** (standard deviation)
* Measures of **skewness**
* Measures of **kurtosis**

These statistics summarize the **key characteristics of data**.

However, numbers alone are sometimes difficult to interpret.

Very often, **visual representations of data communicate patterns much more clearly**.

So instead of just describing data with numbers, we can **visualize its distribution**.

This is where **histograms and bar charts** become useful.

---

# 2. Visualizing Data: Bar Charts vs Histograms

When we want to visualize data, we usually present it in one of two forms:

### 1. Bar Charts

Used for **discrete data**.

Discrete data consists of **distinct categories or groups**.

Example:

Suppose we compute the **average P/E ratio for different regions of the world**:

* North America
* Europe
* Asia
* Africa

Each region becomes a **separate bar**.

Thus, a **bar chart compares categories**.

---

### 2. Histograms

Used for **continuous data**.

Continuous data can take **any value within a range**.

Example:

If we look at **P/E ratios for individual companies**, we do not plot each company separately.

Instead, we group the data into **ranges**.

Example ranges:

* 0 – 4
* 4 – 8
* 8 – 12
* 12 – 16

Then we **count how many companies fall within each range**.

The result is a **histogram**, which shows the **distribution of values**.

---

# 3. Why Fit a Statistical Distribution?

Once we visualize the data with a histogram, we can attempt to **fit a known statistical distribution** to it.

Why do we do this?

Because statisticians have spent decades studying **standard distributions**.

If our data follows one of these distributions, we gain access to **many powerful statistical tools**.

For example:

If data follows a **normal distribution**, we can immediately use:

* probability rules
* confidence intervals
* statistical tests

Thus, identifying the right distribution **adds analytical power to the data**.

---

# 4. How to Choose the Right Distribution

When choosing a distribution, we focus on three key properties of the data:

### 1. Symmetry

Is the data **balanced around the center**, or skewed?

---

### 2. Skewness

If the data is asymmetric:

* Is it **positively skewed**?
* Or **negatively skewed**?

---

### 3. Tails of the Distribution

We also examine the **probability of extreme outcomes**.

Some distributions have:

* **Thin tails** (few extreme values)
* **Fat tails** (more extreme values)

---

# 5. Symmetric Distributions

Let us begin with **symmetric distributions**.

A symmetric distribution means:

The left side and right side of the distribution are **mirror images**.

---

# 6. The Normal Distribution

The **normal distribution** is the most famous statistical distribution.

It is often called the **bell curve**.

Key characteristics:

* Perfectly **symmetric**
* Mean = Median = Mode
* Skewness = **0**

---

### Standard Deviation Rule

In a normal distribution:

| Range        | Percentage of Observations |
| ------------ | -------------------------- |
| Within ±1 SD | ~68%                       |
| Within ±2 SD | ~95%                       |
| Within ±3 SD | ~99%                       |

Thus, almost all observations fall within **three standard deviations of the mean**.

---

### Testing Normality

A simple test for normality:

Check how many observations lie **outside three standard deviations**.

If **more than a few percent** fall outside this range, the distribution may **not be normal**.

---

# 7. Kurtosis in Normal Distribution

The **kurtosis of a normal distribution = 3**.

This value is often used as a **benchmark**.

If your data has:

* Kurtosis > 3 → **Fat tails**
* Kurtosis < 3 → **Thin tails**

---

# 8. The t-Distribution

When sample sizes are **small**, the normal distribution may not be appropriate.

Instead, we often use the **t-distribution**.

Key features:

* Similar shape to the normal distribution
* **Lower peak**
* **Wider tails**

This means extreme values are **more likely**.

---

### Effect of Sample Size

As sample size increases:

The **t-distribution gradually becomes a normal distribution**.

Thus:

Large samples → Normal distribution works well.

Small samples → t-distribution is safer.

---

# 9. The Triangular Distribution

Another symmetric distribution is the **triangular distribution**.

It has a **distinct triangular shape**.

Key properties:

* Symmetric
* **Bounded on both sides**
* Sharp peak

This distribution is useful when:

* Data has **clear minimum and maximum values**
* Values cannot extend to infinity

---

### Why Not Use Normal Here?

The normal distribution allows values from:

[
-\infty \quad to \quad +\infty
]

But many financial variables have **natural limits**.

Example:

Profit margins cannot exceed **100%**.

Thus, triangular distributions sometimes fit better.

---

# 10. Uniform Distribution

Another symmetric distribution is the **uniform distribution**.

Here, **every value within the range is equally likely**.

Example:

If profit margins could be anywhere between **10% and 20%**, and we have **no information about which value is more likely**, we assume a **uniform distribution**.

---

# 11. Asymmetric (Skewed) Distributions

Not all data is symmetric.

Many financial variables show **skewness**.

We must therefore use **skewed distributions**.

---

# 12. Negative Skew Distributions

Negative skew means:

The **tail extends toward negative values**.

This means extreme outcomes occur **on the downside**.

One distribution that captures this is the:

**Minimum Extreme Value Distribution**

This might apply to variables like:

* Profit margins
* Certain downside risk measures

Because:

* There may be **large negative values**
* But an **upper bound exists**

---

# 13. Positive Skew Distributions

Positive skew means:

The **tail extends toward large positive values**.

This is common in finance.

Example variables:

* Stock prices
* P/E ratios
* Wealth distribution

---

### Lognormal Distribution

One of the most common positively skewed distributions is the **lognormal distribution**.

Characteristics:

* Peak on the **left**
* Long tail on the **right**
* Values cannot be negative

---

### Important Property

In a lognormal distribution:

Greater **standard deviation** → greater **positive skewness**.

---

# 14. Fat Tail Distributions

Some datasets produce **extreme values more frequently** than the normal distribution predicts.

These are called **fat-tailed distributions**.

One example is the **Pareto distribution**.

Some researchers argue that financial returns follow **fat-tailed distributions rather than normal distributions**.

---

# 15. Pearson Kurtosis

To analyze tail behavior, we compute **Pearson Kurtosis**.

### Formula

[
Excess\ Kurtosis = Kurtosis - 3
]

---

### Interpretation

| Excess Kurtosis | Type                           |
| --------------- | ------------------------------ |
| 0               | Mesokurtic (normal-like tails) |
| < 0             | Platykurtic (thin tails)       |

> 0 | Leptokurtic (fat tails)

---

# 16. Summary: Choosing the Right Distribution

When fitting a distribution to data, follow these steps.

### Step 1: Determine Data Type

Is the data:

* **Continuous**
* **Discrete**

---

### Step 2: Check Symmetry

Is the distribution:

* **Symmetric**
* **Asymmetric**

---

### Step 3: Identify Skewness

If asymmetric:

* Positive skew
* Negative skew

---

### Step 4: Examine Tails

Do extreme values occur:

* More often than normal → **Fat tails**
* Less often than normal → **Thin tails**

---

# 17. Examples of Distribution Choices

| Data Type                  | Distribution           |
| -------------------------- | ---------------------- |
| Symmetric continuous       | Normal, t-distribution |
| Bounded symmetric          | Triangular             |
| Completely uncertain range | Uniform                |
| Positive skew              | Lognormal              |
| Negative skew              | Minimum extreme value  |
| Fat tails                  | Pareto                 |

---

# 18. Final Insight

When analyzing data:

1. Start with **descriptive statistics**.
2. Create **histograms or bar charts**.
3. Examine **symmetry, skewness, and tails**.
4. Choose the **distribution that best fits the data**.

Once a distribution is identified, you can apply **statistical theory to analyze and predict outcomes**.

---

✔ The ultimate goal is not just to **describe data**, but to **understand the probability structure behind it**.

```python
"""
Statistical Distributions — Lecture Visualization
===================================================
Run each cell in Jupyter Notebook to explore all distributions
covered in the lecture: Normal, t, Triangular, Uniform,
Lognormal, Pareto, and more.
"""

# ─── Cell 1: Imports ──────────────────────────────────────────────────────────
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

plt.rcParams.update({
    "figure.facecolor": "#0f0f1a",
    "axes.facecolor":   "#1a1a2e",
    "axes.edgecolor":   "#444466",
    "axes.labelcolor":  "#ccccee",
    "xtick.color":      "#888899",
    "ytick.color":      "#888899",
    "text.color":       "#eeeeff",
    "grid.color":       "#2a2a3e",
    "grid.linestyle":   "--",
    "grid.alpha":       0.5,
    "font.family":      "monospace",
})

PALETTE = ["#7eb8f7", "#f7a07e", "#7ef7c0", "#f7e07e",
           "#c07ef7", "#f77eb8", "#7ef7f7", "#f7f77e"]
print("✓ Setup complete.")


# ─── Cell 2: Overview — All Distributions at a Glance ─────────────────────────
x = np.linspace(-5, 5, 500)
x_pos = np.linspace(0.01, 8, 500)

distributions = [
    ("Normal",          stats.norm.pdf(x),                   x,     PALETTE[0]),
    ("t (df=3)",        stats.t.pdf(x, df=3),                x,     PALETTE[1]),
    ("Triangular",      stats.triang.pdf(x, c=0.5, loc=-3, scale=6), x, PALETTE[2]),
    ("Uniform",         stats.uniform.pdf(x, loc=-2, scale=4), x,   PALETTE[3]),
    ("Lognormal",       stats.lognorm.pdf(x_pos, s=0.8),     x_pos, PALETTE[4]),
    ("Pareto (fat tail)",stats.pareto.pdf(x_pos, b=2),       x_pos, PALETTE[5]),
    ("Min Extreme Val", stats.gumbel_l.pdf(x),               x,     PALETTE[6]),
    ("Max Extreme Val", stats.gumbel_r.pdf(x),               x,     PALETTE[7]),
]

fig, axes = plt.subplots(2, 4, figsize=(18, 8))
fig.suptitle("Statistical Distributions — Lecture Overview", fontsize=16, y=1.01, color="#eeeeff")

for ax, (name, pdf, xv, color) in zip(axes.flat, distributions):
    ax.plot(xv, pdf, color=color, lw=2.5)
    ax.fill_between(xv, pdf, alpha=0.25, color=color)
    ax.set_title(name, fontsize=11, color=color, pad=6)
    ax.set_yticks([])
    ax.grid(True)

plt.tight_layout()
plt.savefig("01_all_distributions_overview.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 01_all_distributions_overview.png")


# ─── Cell 3: Normal Distribution — Bell Curve + SD Rule ───────────────────────
x = np.linspace(-4, 4, 1000)
pdf = stats.norm.pdf(x)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Normal Distribution", fontsize=14, color="#eeeeff")

# Left: shaded SD bands
ax = axes[0]
ax.set_title("68 – 95 – 99 Rule", color=PALETTE[0])
bands = [(1, "#7eb8f7", "±1 SD  →  68%"),
         (2, "#f7a07e", "±2 SD  →  95%"),
         (3, "#7ef7c0", "±3 SD  →  99%")]
for sd, col, label in reversed(bands):
    mask = (x >= -sd) & (x <= sd)
    ax.fill_between(x[mask], pdf[mask], alpha=0.35, color=col, label=label)
ax.plot(x, pdf, color="white", lw=2)
ax.axvline(0, color="white", lw=1, linestyle=":")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_xlabel("Standard Deviations from Mean")
ax.set_yticks([])
ax.grid(True)

# Right: annotation of properties
ax = axes[1]
ax.set_title("Key Properties", color=PALETTE[0])
ax.plot(x, pdf, color=PALETTE[0], lw=2.5)
ax.fill_between(x, pdf, alpha=0.2, color=PALETTE[0])
props = [(-3.5, 0.38, "Mean = Median = Mode"),
         (-3.5, 0.34, "Skewness = 0"),
         (-3.5, 0.30, "Kurtosis = 3"),
         (-3.5, 0.26, "Symmetric (bell curve)")]
for xp, yp, txt in props:
    ax.text(xp, yp, txt, fontsize=9, color="#ccccee")
ax.set_yticks([])
ax.grid(True)

plt.tight_layout()
plt.savefig("02_normal_distribution.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 02_normal_distribution.png")


# ─── Cell 4: Normal vs t-Distribution — Effect of Sample Size ─────────────────
x = np.linspace(-5, 5, 1000)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Normal vs t-Distribution", fontsize=14, color="#eeeeff")

# Left: overlay
ax = axes[0]
ax.set_title("Fatter Tails in t-Distribution", color=PALETTE[1])
ax.plot(x, stats.norm.pdf(x), color=PALETTE[0], lw=2.5, label="Normal")
for df, alpha in [(3, 0.9), (10, 0.7), (30, 0.5)]:
    ax.plot(x, stats.t.pdf(x, df), lw=2, linestyle="--",
            color=PALETTE[1], alpha=alpha, label=f"t  (df={df})")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_yticks([])
ax.grid(True)

# Right: convergence to normal
ax = axes[1]
ax.set_title("Convergence to Normal as df → ∞", color=PALETTE[2])
dfs = [2, 5, 15, 50]
colors = plt.cm.cool(np.linspace(0.2, 0.9, len(dfs)))
for df, col in zip(dfs, colors):
    ax.plot(x, stats.t.pdf(x, df), lw=2, color=col, label=f"df = {df}")
ax.plot(x, stats.norm.pdf(x), color="white", lw=2.5, linestyle=":", label="Normal")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_yticks([])
ax.grid(True)

plt.tight_layout()
plt.savefig("03_normal_vs_t.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 03_normal_vs_t.png")


# ─── Cell 5: Symmetric Distributions — Triangular & Uniform ───────────────────
x = np.linspace(-0.5, 1.5, 1000)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Symmetric Bounded Distributions", fontsize=14, color="#eeeeff")

# Triangular
ax = axes[0]
ax.set_title("Triangular Distribution  (bounded, peaked)", color=PALETTE[2])
for c, lbl in [(0.5, "Symmetric (c=0.5)"),
               (0.2, "Left skew  (c=0.2)"),
               (0.8, "Right skew (c=0.8)")]:
    pdf = stats.triang.pdf(x, c=c)
    ax.plot(x, pdf, lw=2, label=lbl)
ax.axvline(0, color="#666688", lw=1); ax.axvline(1, color="#666688", lw=1)
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_xlabel("Value"); ax.set_yticks([])
ax.grid(True)

# Uniform
ax = axes[1]
ax.set_title("Uniform Distribution  (all values equally likely)", color=PALETTE[3])
for lo, hi, col in [(-2, 2, PALETTE[3]), (0, 1, PALETTE[0]), (-1, 3, PALETTE[1])]:
    xr = np.linspace(lo - 0.5, hi + 0.5, 1000)
    ax.plot(xr, stats.uniform.pdf(xr, loc=lo, scale=hi - lo),
            lw=2.5, color=col, label=f"[{lo}, {hi}]")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_xlabel("Value"); ax.set_yticks([])
ax.grid(True)

plt.tight_layout()
plt.savefig("04_triangular_uniform.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 04_triangular_uniform.png")


# ─── Cell 6: Skewed Distributions — Lognormal & Gumbel ────────────────────────
x_pos = np.linspace(0.01, 8, 1000)
x_sym = np.linspace(-5, 5, 1000)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Asymmetric (Skewed) Distributions", fontsize=14, color="#eeeeff")

# Lognormal — positive skew
ax = axes[0]
ax.set_title("Lognormal  (positive skew)  →  stock prices, P/E", color=PALETTE[4])
for sigma, col in [(0.4, PALETTE[0]), (0.8, PALETTE[4]), (1.2, PALETTE[1])]:
    ax.plot(x_pos, stats.lognorm.pdf(x_pos, s=sigma),
            lw=2, label=f"σ = {sigma}")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_xlabel("Value"); ax.set_yticks([])
ax.text(4, 0.55, "Larger σ\n→ more\nright skew", color="#ccccee", fontsize=9)
ax.grid(True)

# Gumbel-L — negative skew (minimum extreme value)
ax = axes[1]
ax.set_title("Min Extreme Value  (negative skew)  →  downside risk", color=PALETTE[5])
ax.plot(x_sym, stats.gumbel_l.pdf(x_sym), color=PALETTE[5], lw=2.5,
        label="Gumbel-L  (neg skew)")
ax.plot(x_sym, stats.gumbel_r.pdf(x_sym), color=PALETTE[7], lw=2.5,
        linestyle="--", label="Gumbel-R  (pos skew)")
ax.fill_between(x_sym, stats.gumbel_l.pdf(x_sym), alpha=0.2, color=PALETTE[5])
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_xlabel("Value"); ax.set_yticks([])
ax.grid(True)

plt.tight_layout()
plt.savefig("05_skewed_distributions.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 05_skewed_distributions.png")


# ─── Cell 7: Fat Tails — Normal vs Pareto + Kurtosis ─────────────────────────
x = np.linspace(-5, 5, 1000)
x_pos = np.linspace(1.0, 8, 1000)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Fat Tails & Kurtosis", fontsize=14, color="#eeeeff")

# Normal vs Laplace (fat-tail demo)
ax = axes[0]
ax.set_title("Thin vs Fat Tails", color=PALETTE[5])
ax.plot(x, stats.norm.pdf(x),    color=PALETTE[0], lw=2.5, label="Normal  (mesokurtic, K=3)")
ax.plot(x, stats.laplace.pdf(x), color=PALETTE[5], lw=2.5, label="Laplace  (leptokurtic, K>3)")
ax.plot(x, stats.uniform.pdf(x, loc=-2, scale=4) * 0.9,
        color=PALETTE[3], lw=2.5, label="Uniform  (platykurtic, K<3)")
# Zoom in on right tail
ax.set_xlim(-5, 5); ax.set_yticks([])
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.grid(True)

# Pareto distribution
ax = axes[1]
ax.set_title("Pareto Distribution  (fat tail  →  wealth, returns)", color=PALETTE[5])
for b, col in [(1.5, PALETTE[0]), (2.0, PALETTE[4]), (3.0, PALETTE[1])]:
    ax.plot(x_pos, stats.pareto.pdf(x_pos, b=b),
            lw=2, color=col, label=f"b = {b}  (shape)")
ax.fill_between(x_pos, stats.pareto.pdf(x_pos, b=1.5), alpha=0.15, color=PALETTE[0])
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#444466")
ax.set_xlabel("Value"); ax.set_yticks([])
ax.grid(True)

plt.tight_layout()
plt.savefig("06_fat_tails_kurtosis.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 06_fat_tails_kurtosis.png")


# ─── Cell 8: Kurtosis Comparison Bar Chart ────────────────────────────────────
kurtosis_data = {
    "Normal\n(Mesokurtic)":    0,
    "Uniform\n(Platykurtic)": -1.2,
    "Lognormal\n(σ=0.8)":      3.9,
    "Laplace\n(Leptokurtic)":  3.0,
    "Pareto\n(b=2)":           6.0,
    "t-dist\n(df=5)":          6.0,
}

fig, ax = plt.subplots(figsize=(12, 5))
fig.suptitle("Excess Kurtosis Comparison  (Kurtosis − 3)", fontsize=14, color="#eeeeff")

names = list(kurtosis_data.keys())
values = list(kurtosis_data.values())
colors = [PALETTE[1] if v < 0 else PALETTE[0] if v == 0 else PALETTE[5] for v in values]

bars = ax.bar(names, values, color=colors, edgecolor="#0f0f1a", linewidth=1.5, width=0.6)
ax.axhline(0, color="white", lw=1.5, linestyle="--")
ax.text(5.4, 0.3, "← Mesokurtic baseline", color="white", fontsize=9)

for bar, val in zip(bars, values):
    ax.text(bar.get_x() + bar.get_width() / 2,
            bar.get_height() + (0.15 if val >= 0 else -0.4),
            f"{val:+.1f}", ha="center", fontsize=10, color="#eeeeff")

ax.set_ylabel("Excess Kurtosis")
ax.set_ylim(-2.5, 8)
ax.grid(True, axis="y")

plt.tight_layout()
plt.savefig("07_kurtosis_comparison.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 07_kurtosis_comparison.png")


# ─── Cell 9: Bar Chart vs Histogram (Lecture Example) ─────────────────────────
np.random.seed(42)
regions = ["N. America", "Europe", "Asia", "Africa"]
pe_avg   = [22.4, 17.8, 19.1, 12.3]
pe_raw   = np.concatenate([
    np.random.lognormal(np.log(22), 0.4, 120),
    np.random.lognormal(np.log(18), 0.4, 100),
    np.random.lognormal(np.log(19), 0.4, 110),
    np.random.lognormal(np.log(12), 0.4,  70),
])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Bar Chart  vs  Histogram  (P/E Ratio Example)", fontsize=14, color="#eeeeff")

# Bar chart
ax = axes[0]
ax.set_title("Bar Chart — Discrete (avg P/E by region)", color=PALETTE[0])
bars = ax.bar(regions, pe_avg, color=PALETTE[:4], edgecolor="#0f0f1a", width=0.6)
for bar, val in zip(bars, pe_avg):
    ax.text(bar.get_x() + bar.get_width() / 2, val + 0.3,
            f"{val}", ha="center", fontsize=10)
ax.set_ylabel("Average P/E Ratio"); ax.grid(True, axis="y")

# Histogram
ax = axes[1]
ax.set_title("Histogram — Continuous (individual company P/E)", color=PALETTE[4])
n, bins, patches = ax.hist(pe_raw, bins=20, edgecolor="#0f0f1a", linewidth=0.8)
for patch, col in zip(patches, plt.cm.plasma(np.linspace(0.2, 0.9, len(patches)))):
    patch.set_facecolor(col)
ax.set_xlabel("P/E Ratio"); ax.set_ylabel("Count"); ax.grid(True, axis="y")

plt.tight_layout()
plt.savefig("08_barchart_vs_histogram.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 08_barchart_vs_histogram.png")


# ─── Cell 10: Decision Flowchart — Choosing a Distribution ────────────────────
fig, ax = plt.subplots(figsize=(14, 8))
fig.patch.set_facecolor("#0f0f1a")
ax.set_facecolor("#0f0f1a")
ax.set_xlim(0, 10); ax.set_ylim(0, 10)
ax.axis("off")
ax.set_title("How to Choose the Right Distribution", fontsize=15, color="#eeeeff", pad=12)

BOX  = dict(boxstyle="round,pad=0.5", facecolor="#1a1a2e", edgecolor="#555577", lw=1.5)
LEAF = dict(boxstyle="round,pad=0.5", facecolor="#22223a", edgecolor="#7eb8f7",  lw=2)

nodes = [
    # (x, y, text, style, color)
    (5.0, 9.2, "Start: examine your data",          BOX,  "#eeeeff"),
    (5.0, 7.8, "Continuous or Discrete?",            BOX,  "#f7e07e"),
    (2.0, 6.4, "Bar Chart\n(Discrete)",              LEAF, PALETTE[3]),
    (7.0, 6.4, "Symmetric?",                         BOX,  "#f7e07e"),
    (5.0, 5.0, "Bounded?",                           BOX,  "#f7e07e"),
    (3.5, 3.6, "Triangular\nor Uniform",             LEAF, PALETTE[2]),
    (6.5, 3.6, "Normal\nor t-dist",                  LEAF, PALETTE[0]),
    (9.0, 5.0, "Skew Direction?",                    BOX,  "#f7e07e"),
    (8.0, 3.6, "Lognormal\n(+ skew)",               LEAF, PALETTE[4]),
    (9.8, 3.6, "Min Extreme\n(- skew)",              LEAF, PALETTE[5]),
    (5.0, 1.6, "Check Tails  →  Fat? Use Pareto",   LEAF, PALETTE[1]),
]

positions = {}
for x, y, text, style, color in nodes:
    ax.text(x, y, text, ha="center", va="center", fontsize=9,
            color=color, bbox=style, zorder=3)
    positions[text.split("\n")[0]] = (x, y)

arrows = [
    ((5.0, 9.0), (5.0, 8.1),   ""),
    ((5.0, 7.6), (2.0, 6.7),   "Discrete"),
    ((5.0, 7.6), (7.0, 6.7),   "Continuous"),
    ((7.0, 6.2), (5.0, 5.3),   "Yes"),
    ((7.0, 6.2), (9.0, 5.3),   "No"),
    ((5.0, 4.8), (3.5, 3.9),   "Yes"),
    ((5.0, 4.8), (6.5, 3.9),   "No"),
    ((9.0, 4.8), (8.0, 3.9),   "+"),
    ((9.0, 4.8), (9.8, 3.9),   "−"),
    ((6.5, 3.3), (5.5, 1.9),   ""),
    ((8.0, 3.3), (5.8, 1.9),   ""),
]

for (x1, y1), (x2, y2), label in arrows:
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle="->", color="#888899", lw=1.5))
    if label:
        mx, my = (x1 + x2) / 2, (y1 + y2) / 2
        ax.text(mx + 0.15, my, label, fontsize=8, color="#aaaacc")

plt.tight_layout()
plt.savefig("09_distribution_flowchart.png", dpi=150, bbox_inches="tight",
            facecolor="#0f0f1a")
plt.show()
print("Saved → 09_distribution_flowchart.png")


print("\n" + "="*55)
print("  ALL VISUALIZATIONS COMPLETE")
print("="*55)
print("  01 — All distributions overview")
print("  02 — Normal distribution + 68-95-99 rule")
print("  03 — Normal vs t-distribution")
print("  04 — Triangular & Uniform")
print("  05 — Lognormal & Gumbel (skewed)")
print("  06 — Fat tails & Pareto")
print("  07 — Kurtosis comparison")
print("  08 — Bar chart vs Histogram")
print("  09 — Distribution selection flowchart")
print("="*55)
```

# 1. Introduction: Finance and the Normal Distribution

In this session, we move from **theoretical distributions** to **how they actually appear in financial data**.

Let me start with a confession.

One of the major criticisms of finance as a field is that **it relies too heavily on the normal distribution**.

We use it in:

* Financial theory
* Risk management
* Portfolio models
* Option pricing

But why is finance so dependent on the normal distribution?

There are two major reasons.

---

## Reason 1: Simplicity

The **normal distribution makes life easier** for researchers and analysts.

If data is normally distributed, you only need **two parameters** to describe the entire distribution:

1. **Expected value (Mean)**
2. **Standard deviation**

Many other distributions require **three or four parameters**, which makes modeling more complex.

Thus, the normal distribution allows us to **summarize large datasets very efficiently**.

---

## Reason 2: Large Data Sets

Finance has always dealt with **large volumes of data**.

Long before the term **“big data”** became popular, financial markets were already producing massive datasets.

With large datasets, the **normal distribution provides a convenient framework** for analysis.

---

## The Problem

However, assuming everything follows a normal distribution **can be dangerous**.

Many financial variables **do not follow normal distributions**, especially:

* stock prices
* short-term returns
* financial risk events

Therefore, we need ways to **test whether data actually follows a normal distribution**.

---

# 2. Example 1: Stock Returns – Disney

Let us start with **an individual company**.

Consider annual returns for **The Walt Disney Company** from **1962–2021**.

Disney has been publicly traded since the late 1950s.

For each year, we compute total return:

Return = Price Change + Dividends

---

## Step 1: Create a Histogram

We plot a **histogram of annual returns**.

Each bar represents a range of returns, such as:

* −20% to 0%
* 0% to 20%
* 20% to 40%

This gives us a **visual representation of the distribution**.

---

## Step 2: Superimpose a Normal Distribution

Next, we overlay a **normal distribution curve** on the histogram.

If the histogram and the normal curve match closely, the data may be normally distributed.

For Disney returns, the fit looks **reasonably close**, though not perfect.

However, visual inspection alone is not sufficient.

---

# 3. Q-Q Plot (Quantile–Quantile Plot)

A more powerful tool for checking normality is the **Q-Q plot**.

The idea is simple.

If the data were perfectly normal:

All observed points should lie **on a straight line**.

The line represents the **expected values under a normal distribution**.

---

### Interpretation

If the points:

* lie close to the line → distribution may be normal
* deviate significantly → distribution likely not normal

For Disney:

Most points lie **close to the line**, with a few outliers.

---

# 4. Statistical Tests for Normality

Statistical software can run **formal normality tests**.

You do not need to calculate them manually.

Common tests check:

* whether skewness is significantly different from zero
* whether kurtosis differs from the normal value

Remember:

Normal distribution characteristics:

Skewness = **0**
Kurtosis = **3**

---

### Results for Disney

Out of **nine normality tests**:

* **Seven tests do NOT reject normality**

Thus, Disney annual returns are **approximately normally distributed**.

This means we can reasonably use **normal distribution assumptions**.

For example:

We can estimate return ranges using **mean and standard deviation**.

---

# 5. Example 2: Apple Returns

Now consider another company:

**Apple Inc.**

Annual returns from **1981–2020**.

(Apple went public in 1980, so earlier data does not exist.)

---

## Observations

The histogram shows **much more extreme values**:

* Some years above **200%**
* Some years down **60–70%**

Thus, the distribution appears **much wider** than Disney.

---

## Q-Q Plot

When we plot the Q-Q graph:

Points deviate **more strongly from the normal line**.

---

### Normality Tests

Out of nine tests:

* **Six reject normality**

Thus, Apple returns **do not follow a normal distribution well**.

---

### Interpretation

We can still analyze Apple returns.

But we must be **careful about applying normal distribution assumptions**.

---

# 6. Example 3: Market Index

Next, we examine a market index:

**S&P 500**

Unlike a single stock, an index combines **hundreds of companies**.

The index represents about **500 large U.S. firms**.

---

## Effect of Diversification

Because multiple companies move in different directions:

Extreme movements are **less dramatic**.

For example:

* Individual stocks may move **+200%**
* Index returns rarely move that much

---

## Results

Histogram appears **close to normal**.

Q-Q plot points lie **very close to the line**.

Normality tests:

**None reject normality.**

Thus:

Annual returns of the S&P 500 are **approximately normally distributed**.

---

# 7. Effect of Time Horizon

However, time horizon matters.

Compare:

### Annual Returns

More likely to resemble **normal distribution**.

---

### Daily Returns

Less likely to be normal.

Daily returns often show:

* sharper peaks
* heavier tails
* more extreme observations

Thus:

Short-term returns often **violate normality assumptions**.

---

# 8. Key Observations

Two important patterns emerge.

### Rule 1

Indexes are **more likely to be normally distributed than individual stocks**.

---

### Rule 2

Long-term returns (annual) are **more normal than short-term returns (daily)**.

---

# 9. Danger of Misusing the Normal Distribution

Using the normal distribution incorrectly can create **serious risk problems**.

In a normal distribution:

Extreme events (4–6 standard deviations) are **extremely rare**.

But real financial markets sometimes show **fat tails**.

This means extreme events occur **more often than the normal distribution predicts**.

---

# 10. Risk Management Failure

Suppose a risk management system assumes **normal distribution**.

It might prepare for:

* 2 standard deviation events
* maybe 3 standard deviation events

But it might **ignore more extreme scenarios**.

If the real distribution has **fat tails**, extreme events may occur **much more often**.

This exposes firms to **unexpected losses**.

---

## Criticism by Nassim Taleb

This issue is widely discussed by **Nassim Nicholas Taleb**.

He argues that financial systems often underestimate **rare but catastrophic events**.

These are sometimes called **Black Swan events**.

---

# 11. Stock Prices vs Returns

Another important point:

Stock **prices themselves cannot be normally distributed**.

Why?

Because prices cannot go below **zero**.

But the normal distribution extends from:

−∞ to +∞.

Thus, price distributions tend to show **positive skewness**.

---

# 12. Example: Apple Stock Prices

When we plot daily prices of Apple:

We see a **left clustering and right tail**.

This distribution clearly **does not match a normal distribution**.

Normality tests confirm this.

Most tests reject the hypothesis.

---

# 13. Log Transformation

Finance theory often suggests using **log prices**.

Why?

Because:

[
\ln(0) = -\infty
]

This allows the distribution to extend from **−∞ to +∞**, making normality theoretically possible.

---

### Result

When we take **logarithms of stock prices**:

The distribution becomes **closer to normal**.

But it is **still not perfectly normal**.

For Apple, all normality tests still **reject the hypothesis**.

---

# 14. Important Lesson

Taking the **log of prices does not guarantee normality**.

It may improve the fit, but it does not always solve the problem.

Therefore:

Always **test the distribution instead of assuming it**.

---

# 15. Practical Workflow for Analyzing Financial Data

When analyzing any dataset:

### Step 1

Create a **histogram**.

---

### Step 2

Check whether the data **resembles a normal distribution**.

---

### Step 3

Use tools such as:

* Q-Q plots
* skewness
* kurtosis
* normality tests

---

### Step 4

If the data is not normal, search for **another distribution that fits better**.

Possible alternatives include:

* lognormal distributions
* fat-tail distributions
* extreme value distributions

---

# 16. Final Insight

Sometimes data will **not match any standard statistical distribution**.

In those cases:

* you can still analyze the histogram
* you can still use descriptive statistics

But your ability to make **probability-based predictions becomes more limited**.

---

✔ The key lesson is:

Never assume a distribution.

Always **look at the data first and test the assumptions**.

```python
"""
Distributions in Practice — Finance Applications
=================================================
Lecture visualization notebook.
Split at each  # ─── Cell N  comment to create separate Jupyter cells.

Dependencies:  pip install numpy matplotlib scipy yfinance pandas
"""

# ─── Cell 1: Imports & Theme ──────────────────────────────────────────────────
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
from scipy import stats
import warnings
warnings.filterwarnings("ignore")

plt.rcParams.update({
    "figure.facecolor": "#0d0d1a",
    "axes.facecolor":   "#12122a",
    "axes.edgecolor":   "#333355",
    "axes.labelcolor":  "#ccccee",
    "xtick.color":      "#888899",
    "ytick.color":      "#888899",
    "text.color":       "#eeeeff",
    "grid.color":       "#1e1e3a",
    "grid.linestyle":   "--",
    "grid.alpha":       0.6,
    "font.family":      "monospace",
    "axes.spines.top":  False,
    "axes.spines.right":False,
})

BLUE   = "#7eb8f7"
ORANGE = "#f7a07e"
GREEN  = "#7ef7c0"
YELLOW = "#f7e07e"
PURPLE = "#c07ef7"
PINK   = "#f77eb8"
CYAN   = "#7ef7f7"
RED    = "#f77e7e"

print("✓ Imports & theme ready.")
print("NOTE: Cells 2–4 use SIMULATED data matching the lecture's described patterns.")
print("      Cell 5 downloads REAL data via yfinance (requires internet).")


# ─── Cell 2: Simulate Returns — Disney, Apple, S&P 500 ───────────────────────
np.random.seed(42)
n_years = 60

# Disney: approximately normal, mean ~12%, sd ~25%
disney_returns = np.random.normal(loc=0.12, scale=0.25, size=n_years)
# Add a couple of mild outliers to make it realistic
disney_returns = np.append(disney_returns, [0.95, -0.45])

# Apple: non-normal, very wide, some extreme years (200%+, -60%)
apple_core   = np.random.normal(loc=0.25, scale=0.45, size=40)
apple_extreme = np.array([2.1, 1.8, -0.62, -0.57, 1.5, -0.40])
apple_returns = np.concatenate([apple_core, apple_extreme])

# S&P 500: most normally distributed, mean ~10%, sd ~15%
sp500_returns = np.random.normal(loc=0.10, scale=0.155, size=n_years)
sp500_returns = np.append(sp500_returns, [0.32, -0.38, 0.28])   # mild outliers only

print(f"Disney  → n={len(disney_returns)}, mean={disney_returns.mean():.1%}, sd={disney_returns.std():.1%}")
print(f"Apple   → n={len(apple_returns)},  mean={apple_returns.mean():.1%},  sd={apple_returns.std():.1%}")
print(f"S&P 500 → n={len(sp500_returns)}, mean={sp500_returns.mean():.1%}, sd={sp500_returns.std():.1%}")


# ─── Cell 3: Histogram + Normal Overlay (All Three) ──────────────────────────
datasets = [
    ("Disney  (1962–2021)",  disney_returns, BLUE),
    ("Apple   (1981–2020)",  apple_returns,  ORANGE),
    ("S&P 500 (annual)",     sp500_returns,  GREEN),
]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Annual Returns: Histogram + Normal Overlay", fontsize=14, color="#eeeeff", y=1.02)

for ax, (label, data, color) in zip(axes, datasets):
    mu, sd = data.mean(), data.std()
    ax.set_title(label, color=color, fontsize=11)

    # Histogram
    n, bins, patches = ax.hist(data, bins=18, density=True,
                                color=color, alpha=0.45, edgecolor="#0d0d1a", linewidth=0.8)

    # Normal overlay
    x = np.linspace(data.min() - 0.1, data.max() + 0.1, 300)
    ax.plot(x, stats.norm.pdf(x, mu, sd), color="white", lw=2.5, label="Normal fit")

    # Mean line
    ax.axvline(mu, color=YELLOW, lw=1.5, linestyle="--", label=f"Mean = {mu:.1%}")

    # Annotations
    ax.text(0.04, 0.93, f"μ = {mu:.1%}\nσ = {sd:.1%}", transform=ax.transAxes,
            fontsize=9, color="#ccccee", va="top",
            bbox=dict(facecolor="#1a1a2e", edgecolor="#333355", boxstyle="round,pad=0.4"))

    ax.set_xlabel("Annual Return")
    ax.set_ylabel("Density")
    ax.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
    ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.0%}"))
    ax.grid(True)

plt.tight_layout()
plt.savefig("fin01_histograms_normal_overlay.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → fin01_histograms_normal_overlay.png")


# ─── Cell 4: Q-Q Plots — Disney, Apple, S&P 500 ──────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Q-Q Plots  (Quantile–Quantile)  — Closer to Line = More Normal",
             fontsize=13, color="#eeeeff", y=1.02)

verdicts = [
    ("Disney  (1962–2021)",  disney_returns, BLUE,   "7/9 tests pass ✓ ~Normal"),
    ("Apple   (1981–2020)",  apple_returns,  ORANGE, "6/9 tests fail ✗ Not Normal"),
    ("S&P 500 (annual)",     sp500_returns,  GREEN,  "All tests pass ✓ Normal"),
]

for ax, (label, data, color, verdict) in zip(axes, verdicts):
    ax.set_title(label, color=color, fontsize=11)

    (osm, osr), (slope, intercept, _) = stats.probplot(data, dist="norm")
    ax.scatter(osm, osr, color=color, alpha=0.75, s=30, zorder=3, label="Observed")

    # Reference line
    x_line = np.array([osm.min(), osm.max()])
    ax.plot(x_line, slope * x_line + intercept, color="white", lw=2, label="Normal line")

    ax.set_xlabel("Theoretical Quantiles")
    ax.set_ylabel("Sample Quantiles")
    ax.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")

    # Verdict box
    verdict_color = GREEN if "pass" in verdict else RED
    ax.text(0.04, 0.96, verdict, transform=ax.transAxes, fontsize=9,
            color=verdict_color, va="top",
            bbox=dict(facecolor="#12122a", edgecolor=verdict_color, boxstyle="round,pad=0.4"))
    ax.grid(True)

plt.tight_layout()
plt.savefig("fin02_qq_plots.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → fin02_qq_plots.png")


# ─── Cell 5: Real Data via yfinance — S&P 500 Daily Returns ──────────────────
try:
    import yfinance as yf
    print("Downloading S&P 500 data (^GSPC) via yfinance …")
    sp = yf.download("^GSPC", start="2000-01-01", end="2024-01-01", progress=False)
    prices_real = sp["Close"].dropna()

    # Daily log returns
    daily_returns = np.log(prices_real / prices_real.shift(1)).dropna().values

    # Annual returns (resample)
    annual_px     = sp["Close"].resample("YE").last().dropna()
    annual_returns = annual_px.pct_change().dropna().values

    USE_REAL = True
    print(f"✓ Downloaded {len(daily_returns):,} daily return observations.")
    print(f"  Daily  → mean={daily_returns.mean():.4%}, sd={daily_returns.std():.4%}")
    print(f"  Annual → mean={annual_returns.mean():.2%}, sd={annual_returns.std():.2%}")

except Exception as e:
    print(f"⚠ yfinance download failed ({e}). Using simulated data instead.")
    np.random.seed(7)
    daily_returns  = np.random.standard_t(df=4, size=6000) * 0.01
    annual_returns = np.random.normal(0.10, 0.155, 24)
    USE_REAL = False


# ─── Cell 6: Daily vs Annual Returns — Normality Comparison ──────────────────
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle("S&P 500 — Daily vs Annual Returns  (Time Horizon Effect)",
             fontsize=14, color="#eeeeff", y=1.01)

pairs = [
    (axes[0, 0], daily_returns,  "Daily Returns",  CYAN,   "hist"),
    (axes[0, 1], daily_returns,  "Daily Returns",  CYAN,   "qq"),
    (axes[1, 0], annual_returns, "Annual Returns", GREEN,  "hist"),
    (axes[1, 1], annual_returns, "Annual Returns", GREEN,  "qq"),
]

for ax, data, label, color, kind in pairs:
    mu, sd = data.mean(), data.std()

    if kind == "hist":
        ax.set_title(f"{label} — Histogram", color=color)
        ax.hist(data, bins=60, density=True, color=color, alpha=0.45,
                edgecolor="#0d0d1a", linewidth=0.5)
        x = np.linspace(data.min(), data.max(), 400)
        ax.plot(x, stats.norm.pdf(x, mu, sd), color="white", lw=2, label="Normal fit")
        ax.set_xlabel("Return")
        ax.set_ylabel("Density")
        ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
        ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda v, _: f"{v:.1%}"))

        # Kurtosis annotation
        kurt = stats.kurtosis(data, fisher=False)
        sk   = stats.skew(data)
        ax.text(0.98, 0.95, f"Kurtosis = {kurt:.2f}\nSkewness = {sk:.2f}",
                transform=ax.transAxes, fontsize=9, ha="right", va="top", color="#ccccee",
                bbox=dict(facecolor="#1a1a2e", edgecolor="#333355", boxstyle="round,pad=0.4"))

    else:  # Q-Q
        ax.set_title(f"{label} — Q-Q Plot", color=color)
        (osm, osr), (slope, intercept, _) = stats.probplot(data, dist="norm")
        ax.scatter(osm, osr, color=color, alpha=0.5, s=10 if len(data) > 500 else 40)
        x_line = np.array([osm.min(), osm.max()])
        ax.plot(x_line, slope * x_line + intercept, color="white", lw=2)
        ax.set_xlabel("Theoretical Quantiles")
        ax.set_ylabel("Sample Quantiles")

        # Shapiro or KS test
        if len(data) <= 5000:
            stat, p = stats.shapiro(data[:5000])
            test_name = "Shapiro-Wilk"
        else:
            stat, p = stats.kstest(data, "norm", args=(mu, sd))
            test_name = "KS Test"
        verdict = "~Normal ✓" if p > 0.05 else "Not Normal ✗"
        v_col   = GREEN if p > 0.05 else RED
        ax.text(0.04, 0.96, f"{test_name}\np = {p:.4f}  →  {verdict}",
                transform=ax.transAxes, fontsize=9, va="top", color=v_col,
                bbox=dict(facecolor="#12122a", edgecolor=v_col, boxstyle="round,pad=0.4"))

    ax.grid(True)

plt.tight_layout()
plt.savefig("fin03_daily_vs_annual.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → fin03_daily_vs_annual.png")


# ─── Cell 7: Stock Price vs Log Price — Apple Simulation ─────────────────────
np.random.seed(21)
n_days   = 2520   # ~10 years
log_ret  = np.random.normal(0.00085, 0.018, n_days)
price    = 30 * np.exp(np.cumsum(log_ret))   # simulated Apple-like price path
log_price = np.log(price)

fig = plt.figure(figsize=(18, 10))
fig.suptitle("Stock Price vs Log Price — Distribution Analysis  (Apple-like)",
             fontsize=14, color="#eeeeff")

gs = gridspec.GridSpec(2, 3, figure=fig, hspace=0.45, wspace=0.35)

# ── Price time series
ax0 = fig.add_subplot(gs[0, :2])
ax0.plot(price, color=ORANGE, lw=1.2, alpha=0.9)
ax0.fill_between(range(len(price)), price, alpha=0.1, color=ORANGE)
ax0.set_title("Simulated Stock Price Path", color=ORANGE)
ax0.set_xlabel("Trading Day"); ax0.set_ylabel("Price ($)")
ax0.grid(True)

# ── Price histogram
ax1 = fig.add_subplot(gs[0, 2])
ax1.set_title("Price Distribution\n(Right-skewed — NOT normal)", color=ORANGE)
ax1.hist(price, bins=40, density=True, color=ORANGE, alpha=0.5, edgecolor="#0d0d1a")
x = np.linspace(price.min(), price.max(), 300)
ax1.plot(x, stats.norm.pdf(x, price.mean(), price.std()), color="white", lw=2, label="Normal fit")
ax1.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
ax1.set_xlabel("Price ($)"); ax1.set_ylabel("Density"); ax1.grid(True)
sk = stats.skew(price)
ax1.text(0.97, 0.95, f"Skewness = {sk:.2f}", transform=ax1.transAxes,
         ha="right", va="top", fontsize=9, color=YELLOW,
         bbox=dict(facecolor="#1a1a2e", edgecolor="#333355", boxstyle="round,pad=0.4"))

# ── Log price time series
ax2 = fig.add_subplot(gs[1, :2])
ax2.plot(log_price, color=PURPLE, lw=1.2, alpha=0.9)
ax2.fill_between(range(len(log_price)), log_price, alpha=0.1, color=PURPLE)
ax2.set_title("Log(Price) Path", color=PURPLE)
ax2.set_xlabel("Trading Day"); ax2.set_ylabel("ln(Price)")
ax2.grid(True)

# ── Log price histogram
ax3 = fig.add_subplot(gs[1, 2])
ax3.set_title("Log Price Distribution\n(Closer to normal, but not perfect)", color=PURPLE)
ax3.hist(log_price, bins=40, density=True, color=PURPLE, alpha=0.5, edgecolor="#0d0d1a")
x2 = np.linspace(log_price.min(), log_price.max(), 300)
ax3.plot(x2, stats.norm.pdf(x2, log_price.mean(), log_price.std()),
         color="white", lw=2, label="Normal fit")
ax3.legend(fontsize=8, facecolor="#1a1a2e", edgecolor="#333355")
ax3.set_xlabel("ln(Price)"); ax3.set_ylabel("Density"); ax3.grid(True)
sk2 = stats.skew(log_price)
stat, p = stats.shapiro(log_price[:5000])
note = f"Skewness = {sk2:.2f}\nShapiro p = {p:.4f}"
col  = GREEN if p > 0.05 else RED
ax3.text(0.97, 0.95, note, transform=ax3.transAxes,
         ha="right", va="top", fontsize=9, color=col,
         bbox=dict(facecolor="#12122a", edgecolor=col, boxstyle="round,pad=0.4"))

plt.savefig("fin04_price_vs_logprice.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → fin04_price_vs_logprice.png")


# ─── Cell 8: Fat Tails — Normal vs Real Finance Returns ──────────────────────
np.random.seed(99)
fat_tail = np.random.standard_t(df=3, size=5000) * 0.01
normal_s = np.random.normal(0, fat_tail.std(), 5000)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
fig.suptitle("Fat Tails — Why the Normal Distribution Underestimates Extreme Events",
             fontsize=13, color="#eeeeff")

# Left: overlay histogram
ax = axes[0]
ax.set_title("Normal vs Fat-Tail Return Distribution", color=BLUE)
bins = np.linspace(-0.07, 0.07, 80)
ax.hist(normal_s, bins=bins, density=True, alpha=0.45, color=BLUE,
        edgecolor="#0d0d1a", label="Normal (simulated)")
ax.hist(fat_tail,  bins=bins, density=True, alpha=0.45, color=RED,
        edgecolor="#0d0d1a", label="Fat-tail / t(df=3)")
x = np.linspace(-0.07, 0.07, 400)
ax.plot(x, stats.norm.pdf(x, 0, fat_tail.std()), color=BLUE,  lw=2)
ax.plot(x, stats.t.pdf(x, df=3, scale=fat_tail.std()), color=RED, lw=2)
ax.set_xlabel("Daily Return"); ax.set_ylabel("Density")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.grid(True)

# Right: tail zoom — show how much more often extremes occur
ax = axes[1]
ax.set_title("Right-Tail Zoom  (Extreme Events)", color=RED)
x_tail = np.linspace(0.02, 0.07, 400)
ax.plot(x_tail, stats.norm.pdf(x_tail, 0, fat_tail.std()),
        color=BLUE, lw=2.5, label="Normal")
ax.plot(x_tail, stats.t.pdf(x_tail, df=3, scale=fat_tail.std()),
        color=RED,  lw=2.5, label="Fat-tail t(df=3)")
ax.fill_between(x_tail,
                stats.norm.pdf(x_tail, 0, fat_tail.std()),
                stats.t.pdf(x_tail, df=3, scale=fat_tail.std()),
                color=RED, alpha=0.25, label="Extra probability (missed by Normal)")
ax.set_xlabel("Daily Return"); ax.set_ylabel("Density")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.text(0.035, ax.get_ylim()[1] * 0.5,
        "Black Swan\nzone", color=RED, fontsize=10,
        bbox=dict(facecolor="#1a1a2e", edgecolor=RED, boxstyle="round,pad=0.4"))
ax.grid(True)

plt.tight_layout()
plt.savefig("fin05_fat_tails.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → fin05_fat_tails.png")


# ─── Cell 9: Normality Test Scorecard ────────────────────────────────────────
def normality_scorecard(data, label):
    """Run multiple normality tests and return pass/fail results."""
    results = {}
    mu, sd = data.mean(), data.std()

    # Shapiro-Wilk (best for n < 5000)
    samp = data[:5000] if len(data) > 5000 else data
    s, p = stats.shapiro(samp);   results["Shapiro-Wilk"]      = p
    s, p = stats.kstest(data, "norm", args=(mu, sd)); results["Kolmogorov-Smirnov"] = p
    s, p = stats.normaltest(data);                    results["D'Agostino-Pearson"] = p
    ad_result = stats.anderson(data, dist="norm")
    # ad_result.statistic vs ad_result.critical_values[2] (5% significance level)
    results["Anderson-Darling"] = 0.10 if ad_result.statistic < ad_result.critical_values[2] else 0.01

    # Skewness and kurtosis Z-tests
    sk_z  = abs(stats.skew(data))  / np.sqrt(6 / len(data))
    ku_z  = abs(stats.kurtosis(data, fisher=False) - 3) / np.sqrt(24 / len(data))
    results["Skewness Z-test"]  = 0.10 if sk_z < 1.96 else 0.01
    results["Kurtosis Z-test"]  = 0.10 if ku_z < 1.96 else 0.01

    return results

fig, axes = plt.subplots(1, 3, figsize=(18, 5))
fig.suptitle("Normality Test Scorecard — Pass / Fail Summary",
             fontsize=14, color="#eeeeff", y=1.02)

subjects = [
    ("Disney  (annual)",  disney_returns,  BLUE),
    ("Apple   (annual)",  apple_returns,   ORANGE),
    ("S&P 500 (annual)",  sp500_returns,   GREEN),
]

for ax, (label, data, color) in zip(axes, subjects):
    results = normality_scorecard(data, label)
    tests   = list(results.keys())
    pvals   = list(results.values())
    passes  = [p > 0.05 for p in pvals]

    bar_colors = [GREEN if p else RED for p in passes]
    bar_vals   = [1] * len(tests)

    bars = ax.barh(tests, bar_vals, color=bar_colors, edgecolor="#0d0d1a",
                   height=0.55, alpha=0.85)

    for i, (p_val, passed) in enumerate(zip(pvals, passes)):
        ax.text(0.5, i, "PASS ✓" if passed else "FAIL ✗",
                ha="center", va="center", fontsize=10, color="white", fontweight="bold")

    n_pass  = sum(passes)
    n_total = len(passes)
    verdict_col = GREEN if n_pass >= n_total * 0.7 else RED
    ax.set_title(f"{label}\n{n_pass}/{n_total} tests pass", color=color, fontsize=10)
    ax.set_xlim(0, 1); ax.set_xticks([])
    ax.set_title(f"{label}\n{n_pass}/{n_total} pass — "
                 + ("~Normal ✓" if n_pass >= n_total * 0.7 else "Not Normal ✗"),
                 color=verdict_col, fontsize=10)
    ax.grid(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["left"].set_visible(True)

plt.tight_layout()
plt.savefig("fin06_normality_scorecard.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → fin06_normality_scorecard.png")


# ─── Cell 10: Why Prices Can't Be Normal — Visual Proof ──────────────────────
x_norm   = np.linspace(-5, 5, 400)
x_prices = np.linspace(-0.5, 8, 400)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle("Why Stock PRICES Cannot Follow a Normal Distribution",
             fontsize=13, color="#eeeeff")

# Left: normal extends to -inf
ax = axes[0]
ax.set_title("Normal Distribution\n(extends to −∞  →  impossible for prices)", color=RED)
ax.plot(x_norm, stats.norm.pdf(x_norm), color=BLUE, lw=2.5)
ax.fill_between(x_norm[x_norm < 0], stats.norm.pdf(x_norm[x_norm < 0]),
                color=RED, alpha=0.35, label="Negative prices (impossible!)")
ax.fill_between(x_norm[x_norm >= 0], stats.norm.pdf(x_norm[x_norm >= 0]),
                color=BLUE, alpha=0.25)
ax.axvline(0, color=YELLOW, lw=2, linestyle="--", label="Zero price boundary")
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.set_xlabel("Price"); ax.set_yticks([])
ax.grid(True)

# Right: lognormal is bounded at zero
ax = axes[1]
ax.set_title("Lognormal Distribution\n(bounded at 0  →  fits prices better)", color=GREEN)
for s, col in [(0.4, CYAN), (0.8, GREEN), (1.2, PURPLE)]:
    ax.plot(x_prices, stats.lognorm.pdf(x_prices, s=s),
            lw=2, color=col, label=f"σ = {s}")
ax.axvline(0, color=YELLOW, lw=2, linestyle="--", label="Zero boundary")
ax.fill_between(x_prices, stats.lognorm.pdf(x_prices, s=0.8), alpha=0.15, color=GREEN)
ax.legend(fontsize=9, facecolor="#1a1a2e", edgecolor="#333355")
ax.set_xlabel("Price"); ax.set_yticks([])
ax.grid(True)

plt.tight_layout()
plt.savefig("fin07_prices_vs_lognormal.png", dpi=150, bbox_inches="tight",
            facecolor="#0d0d1a")
plt.show()
print("Saved → fin07_prices_vs_lognormal.png")


# ─── Cell 11: Practical Workflow Summary ─────────────────────────────────────
fig, ax = plt.subplots(figsize=(14, 7))
fig.patch.set_facecolor("#0d0d1a")
ax.set_facecolor("#0d0d1a")
ax.set_xlim(0, 10); ax.set_ylim(0, 10)
ax.axis("off")
ax.set_title("Practical Workflow for Analyzing Financial Data", fontsize=14,
             color="#eeeeff", pad=12)

BOX  = dict(boxstyle="round,pad=0.55", facecolor="#1a1a2e", edgecolor="#444466", lw=1.5)
STEP = dict(boxstyle="round,pad=0.55", facecolor="#12122a", edgecolor="#7eb8f7",  lw=2)
WARN = dict(boxstyle="round,pad=0.55", facecolor="#2a1010", edgecolor="#f77e7e",  lw=2)
OK   = dict(boxstyle="round,pad=0.55", facecolor="#0a2a0a", edgecolor="#7ef7c0",  lw=2)

steps = [
    (5.0, 9.0,  "Step 1 — Plot Histogram",                              STEP, BLUE),
    (5.0, 7.5,  "Step 2 — Does it look normal?",                        BOX,  YELLOW),
    (2.0, 6.0,  "YES → Overlay normal curve\ncheck visually",           OK,   GREEN),
    (8.0, 6.0,  "NO  → Try lognormal / fat-tail\ndistribution",         WARN, ORANGE),
    (5.0, 4.6,  "Step 3 — Run formal tests\n(Q-Q plot, Shapiro, KS …)", STEP, BLUE),
    (2.5, 3.2,  "Tests PASS\n→ Use Normal Distribution\nassumptions",   OK,   GREEN),
    (7.5, 3.2,  "Tests FAIL\n→ Use alternative distribution\nor non-parametric methods", WARN, RED),
    (5.0, 1.5,  "⚠  Never ASSUME a distribution\nAlways TEST first",   WARN, YELLOW),
]

for x, y, text, style, color in steps:
    ax.text(x, y, text, ha="center", va="center", fontsize=9,
            color=color, bbox=style, zorder=3, linespacing=1.4)

arrows = [
    (5.0, 8.7,  5.0, 7.85),
    (5.0, 7.15, 2.0, 6.4),
    (5.0, 7.15, 8.0, 6.4),
    (2.0, 5.6,  3.8, 4.95),
    (8.0, 5.6,  6.2, 4.95),
    (5.0, 4.25, 2.5, 3.65),
    (5.0, 4.25, 7.5, 3.65),
    (2.5, 2.75, 4.2, 1.75),
    (7.5, 2.75, 5.8, 1.75),
]
for x1, y1, x2, y2 in arrows:
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle="->", color="#666688", lw=1.5))

plt.tight_layout()
plt.savefig("fin08_workflow.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → fin08_workflow.png")


# ─── Cell 12: Full Summary Dashboard ─────────────────────────────────────────
summary_data = {
    "Asset":         ["Disney (annual)", "Apple (annual)", "S&P 500 (annual)", "S&P 500 (daily)"],
    "Mean Return":   [f"{disney_returns.mean():.1%}", f"{apple_returns.mean():.1%}",
                      f"{sp500_returns.mean():.1%}",  f"{daily_returns.mean():.3%}"],
    "Std Dev":       [f"{disney_returns.std():.1%}",  f"{apple_returns.std():.1%}",
                      f"{sp500_returns.std():.1%}",   f"{daily_returns.std():.3%}"],
    "Skewness":      [f"{stats.skew(disney_returns):.2f}", f"{stats.skew(apple_returns):.2f}",
                      f"{stats.skew(sp500_returns):.2f}",  f"{stats.skew(daily_returns):.2f}"],
    "Kurtosis":      [f"{stats.kurtosis(disney_returns, fisher=False):.2f}",
                      f"{stats.kurtosis(apple_returns,  fisher=False):.2f}",
                      f"{stats.kurtosis(sp500_returns,  fisher=False):.2f}",
                      f"{stats.kurtosis(daily_returns,  fisher=False):.2f}"],
    "~Normal?":      ["Yes ✓", "No ✗", "Yes ✓", "No ✗"],
}

fig, ax = plt.subplots(figsize=(14, 4))
fig.patch.set_facecolor("#0d0d1a")
ax.set_facecolor("#0d0d1a")
ax.axis("off")
ax.set_title("Summary — Normality of Financial Return Series",
             fontsize=14, color="#eeeeff", pad=16)

df_summary = pd.DataFrame(summary_data)
tbl = ax.table(
    cellText=df_summary.values,
    colLabels=df_summary.columns,
    cellLoc="center",
    loc="center",
)
tbl.auto_set_font_size(False)
tbl.set_fontsize(10)
tbl.scale(1.3, 2.2)

for (row, col), cell in tbl.get_celld().items():
    cell.set_edgecolor("#333355")
    if row == 0:
        cell.set_facecolor("#1a1a3a")
        cell.set_text_props(color=BLUE, fontweight="bold")
    else:
        cell.set_facecolor("#0d0d1a" if row % 2 == 0 else "#12122a")
        text = cell.get_text().get_text()
        if "Yes" in text:
            cell.set_text_props(color=GREEN)
        elif "No" in text:
            cell.set_text_props(color=RED)
        else:
            cell.set_text_props(color="#ccccee")

plt.tight_layout()
plt.savefig("fin09_summary_table.png", dpi=150, bbox_inches="tight", facecolor="#0d0d1a")
plt.show()
print("Saved → fin09_summary_table.png")


print("\n" + "="*58)
print("  ALL FINANCE DISTRIBUTION VISUALIZATIONS COMPLETE")
print("="*58)
print("  fin01 — Histograms + normal overlay (Disney/Apple/S&P)")
print("  fin02 — Q-Q plots for all three")
print("  fin03 — Daily vs annual returns (time horizon effect)")
print("  fin04 — Stock price vs log price distribution")
print("  fin05 — Fat tails & Black Swan zone")
print("  fin06 — Normality test scorecard")
print("  fin07 — Why prices can't be normal → lognormal")
print("  fin08 — Practical workflow flowchart")
print("  fin09 — Summary table")
print("="*58)
```
