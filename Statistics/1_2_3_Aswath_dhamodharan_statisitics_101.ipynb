# Statistics — Lecture 1

## Introduction to Statistics: Understanding Data

---

# 1. Opening Thought: What is Statistics?

Let me begin this course with a deceptively simple question:

**What exactly is statistics?**

At its core, **statistics is the science of data**.

It is the discipline that allows us to:

* Collect data
* Organize it
* Analyze it
* Visualize it
* And most importantly, **convert data into meaningful information**

In a world overflowing with raw data, statistics acts as the **tool that extracts insight from chaos**.

Without statistics, data is just noise.

With statistics, data becomes **knowledge**.

---

# 2. What is Data?

Before we talk about statistics, we must understand **data**.

**Data simply means information.**

This information can appear in many forms.

### Two main types of data

### 1. Quantitative Data (Numerical)

This type of data is expressed using numbers.

Examples:

* Age
* Height
* Income
* Temperature
* Stock price

Example:

| Person | Height |
| ------ | ------ |
| A      | 5.7 ft |
| B      | 6.1 ft |

Here the data is **measurable**.

---

### 2. Qualitative Data (Categorical)

This type describes **qualities or characteristics**, not numbers.

Examples:

* Gender
* Color
* Pain level
* Customer satisfaction
* Political preference

Example:

Pain is a **qualitative feeling**, but doctors often ask patients to rate pain **from 0 to 10**.

This is an example of **converting qualitative data into quantitative data**.

---

# 3. Continuous vs Discrete Data

Another important classification of data is:

### 1. Discrete Data

Discrete data can take **specific separated values**.

Example:

Basketball score.

Possible values:

* 80 points
* 82 points
* 83 points

But **83.5 points is impossible**.

Therefore the data moves in **steps**.

Examples of discrete data:

* Number of students
* Number of cars
* Number of goals

---

### 2. Continuous Data

Continuous data can take **any value within a range**.

Example:

Height.

Possible values:

* 5.72 ft
* 5.721 ft
* 5.7213 ft

There are **infinite possible values**.

Examples:

* Weight
* Temperature
* Distance
* Time

---

# 4. Why Statistics is More Important Today

Human beings have **always used data for decision making**.

But something has changed dramatically in the last few decades.

### 200 Years Ago

Data problems were:

* Hard to collect
* Hard to store
* Hard to access

Even if you knew statistics, **getting the data was extremely difficult**.

---

### Today

Technology has changed everything.

Now we can:

* Collect massive data
* Store it cheaply
* Access it instantly

Examples:

* Smartphone tracking
* Online transactions
* Social media activity
* GPS location

This leads to a new problem.

---

# 5. The Problem of Data Overload

Today we **do not suffer from lack of data**.

Instead, we suffer from:

**Too much data.**

When people are overwhelmed with information, something interesting happens.

They stop analyzing data carefully and fall back on:

### Rules of Thumb

Simple shortcuts developed from past experience.

Example:

> "Stocks always recover."

Even if new data contradicts it.

---

### Mental Shortcuts (Heuristics)

Humans often repeat actions that worked previously.

Even when the situation is completely different.

This is why **statistics is essential**.

It forces us to rely on **evidence rather than intuition**.

---

# 6. Statistics Influences Almost Everything

Statistics affects:

### Public Policy

Examples:

* Fiscal policy
* Healthcare policy
* Education decisions

### Business

Examples:

* Pricing decisions
* Risk management
* Marketing analytics

### Personal Life

Examples:

* Insurance decisions
* Medical treatment
* Investment decisions

However, there is a major danger.

---

# 7. The Misuse of Statistics

Statistics is often perceived as **objective and unbiased** because it uses numbers.

But that belief is dangerous.

Statistics **can be manipulated**.

People can manipulate data through:

### 1. Selective Sampling

Choosing only the data that supports their agenda.

### 2. Misleading Measurement

Measuring the wrong variable.

### 3. Cherry Picking

Ignoring data that contradicts their conclusions.

---

# 8. Bad Data vs Good Data

In economics there is a concept called **Gresham's Law**.

It states:

> **Bad money drives out good money.**

If fake coins circulate with real gold coins, eventually only fake coins remain in circulation.

Something similar happens with data.

In the age of social media:

**Bad data spreads faster than good data.**

Why?

Because:

* It is sensational
* It confirms biases
* It spreads quickly online

This makes statistical literacy extremely important.

Every individual must learn to ask:

* Is the data reliable?
* Is the sample biased?
* Is the conclusion justified?

---

# 9. Big Data

One of the most popular buzzwords today is **Big Data**.

Big Data does **not only mean large amounts of data**.

It also refers to **new types of data that previously could not be collected**.

Example:

Location tracking.

Your smartphone apps constantly collect:

* Location
* Search history
* Shopping habits
* Social interactions

Companies use this information to make decisions.

Examples:

* Targeted advertising
* Product recommendations
* Pricing strategies

---

# 10. Data Analytics

Another modern term is **Data Analytics**.

But in reality:

**Statistics has always been data analytics.**

The difference today is:

* Data sets are extremely large
* Data includes both qualitative and quantitative information

Therefore modern analytics focuses on:

* Large datasets
* Complex models
* Machine learning techniques

---

# 11. The Statistical Process

Statistics involves a sequence of steps.

We can divide the process into several stages.

---

# Step 1 — Data Collection

The first stage is **collecting data**.

Historically this was done manually.

Example:

* Surveys
* Paper forms
* Interviews

Today data collection is mostly digital.

Examples:

* Online forms
* Sensors
* Databases
* Transaction logs

---

# Step 2 — Sampling

In many cases, studying the entire population is impossible.

Example:

Suppose we want to study **all businesses in the world**.

There are millions of them.

So instead we study a **sample**.

Examples:

* Only publicly traded companies
* Only companies in the US
* Only large companies

Sampling is extremely important because **bad samples produce bad conclusions**.

---

# 12. Two Major Problems in Sampling

There are two major dangers when sampling data.

---

## 1. Bias

Bias occurs when the sample **systematically favors certain outcomes**.

Example:

Surveying only urban voters to predict a national election.

This produces **misleading results**.

---

## 2. Noise (Sampling Error)

Even with a fair sample, results are **never perfectly accurate**.

There is always **random variation**.

This uncertainty is called **sampling error**.

Understanding this error is critical for interpreting results.

---

# 13. Descriptive Statistics

Once data is collected, the next step is **summarizing it**.

This is called **descriptive statistics**.

Instead of showing all raw data points, we summarize them using numbers.

Three important types of summary measures are:

---

## 1. Measures of Location

These represent the **typical value**.

Example:

Average (mean).

If we have 100 observations, the mean gives a single number summarizing them.

---

## 2. Measures of Dispersion

These describe **variation in the data**.

Examples:

* Range
* Variance
* Standard deviation

They tell us how spread out the values are.

---

## 3. Measures of Skewness

Skewness tells us **whether data is symmetric or tilted**.

Example:

Income distributions are usually **right-skewed**.

Few people earn extremely high incomes.

---

# 14. Visualizing Data: Histograms

Another way to understand data is through **visualization**.

One of the most common tools is the **histogram**.

A histogram shows:

**How many observations fall into different ranges.**

Example:

Suppose we measure heights of 100 people.

We count how many fall into ranges like:

* 5.0 – 5.2 ft
* 5.2 – 5.4 ft
* 5.4 – 5.6 ft

This gives us a visual representation of the data distribution.

---

# 15. Statistical Distributions

Sometimes data follows known patterns called **statistical distributions**.

The most famous is the:

**Normal Distribution (Bell Curve).**

Distributions are useful because they have known mathematical properties.

This allows us to make predictions.

Distributions can be:

### Symmetric

Both sides look similar.

Example:

Normal distribution.

### Asymmetric (Skewed)

One side has a longer tail.

Example:

Income distribution.

---

# 16. Relationships Between Variables

Often we study **multiple variables together**.

Example:

* Interest rates
* Stock prices

We may ask:

Do they move together?

Possible relationships:

* Positive relationship
* Negative relationship
* No relationship

---

# 17. Correlation vs Causation

Finding a relationship does **not mean causation**.

Two variables might move together for many reasons.

Example:

Suppose we observe:

Stock prices increase when interest rates decrease.

We must ask:

* Do interest rates influence stock prices?
* Do stock prices influence interest rates?
* Or is another factor influencing both?

This distinction is extremely important.

---

# 18. From Description to Prediction

Statistical analysis often progresses through three stages:

### 1. Description

Document how variables move together.

### 2. Causation

Identify whether one variable causes another.

### 3. Prediction

Use the relationship to forecast future values.

Example:

Predicting stock prices using interest rates.

---

# 19. Probability

Probability measures **the likelihood of an event occurring**.

Events can be:

### Discrete

Two outcomes.

Example:

* Company goes bankrupt
* Company survives

---

### Continuous

Many possible outcomes.

Example:

Profit could be any value between 0 and 10 billion.

Probability allows us to estimate:

* Risk
* Uncertainty
* Future outcomes

---

# 20. Statistical Tools Using Probability

Statistics uses several probability-based tools.

Examples include:

### 1. Logit / Probit Models

Used to estimate probability of events like:

* Bankruptcy
* Acquisition
* Loan default

---

### 2. Decision Trees

Used when events happen in **stages**.

Example:

Drug approval process.

* Phase 1
* Phase 2
* Phase 3

Each stage has a probability of success.

---

### 3. Simulation Models

Instead of using fixed numbers, we use **distributions** for variables.

Example:

In company valuation we may treat:

* Growth rate
* Risk
* Cash flow

as probability distributions.

This produces a **range of possible outcomes** rather than a single estimate.

---

# Final Summary

Statistics is the discipline that transforms **raw data into actionable information**.

The full statistical process includes:

1. Data Collection
2. Sampling
3. Descriptive Statistics
4. Data Visualization
5. Statistical Distributions
6. Relationship Analysis
7. Probability Modeling
8. Decision Making

Understanding statistics allows us to:

* interpret data correctly
* detect misleading information
* make better decisions in business, finance, and everyday life.

# Statistics Lecture Notes

## Session 2 — Populations and Samples

---

# 1. Introduction

Welcome everyone.

In this session we are going to discuss a topic that many people consider boring, but which is actually **one of the most important foundations of statistics**:

**Populations and Samples.**

Statistics is essentially about **drawing conclusions about a large group using a smaller group of observations.**

Before we do any statistical analysis, we must understand:

* What is the **population**
* What is the **sample**
* How sampling works
* What problems sampling can introduce

---

# 2. Population

### Definition

A **population** is the **complete set of all possible observations** that we are interested in studying.

In other words:

> Population = Entire universe of objects or events we want to analyze.

---

### Example 1: Business Behavior in a Crisis

Suppose we want to study:

**How businesses behave during a crisis like COVID-19.**

The **population** would include:

* Every business in the world
* Small businesses
* Large corporations
* Public companies
* Private companies
* Businesses in all countries

Clearly, collecting data for **every business on Earth** is almost impossible.

---

# 3. Sample

Because studying the entire population is usually impossible, we use:

### Sample

A **sample** is a **subset of the population used to represent the whole population.**

Example:

Population = All businesses in the world
Sample = 46,000 publicly traded companies globally

Even though 46,000 sounds large, it is still **only a subset of all businesses**.

---

# 4. Why Do We Use Sampling?

Sampling is used mainly for **three reasons**.

---

## 1. Practicality

Sometimes the population is **too large or inaccessible**.

Example:

The **U.S. Census** is supposed to count every person in the country.

But in reality:

* Population > 300 million
* Some people are difficult to reach
* Budget and time are limited

So the government uses **large samples** to estimate population characteristics.

---

## 2. Cost

Collecting data on the entire population can be **very expensive**.

Example:

Collecting financial data for **every company in the world**.

Even today, despite technology, collecting full population data can still be expensive.

---

## 3. Time

Sometimes we need **results quickly**.

Example: **Election polls**

Polls usually sample:

* 800 people
* 1000 people
* 2000 people

From these small samples, they try to **predict the behavior of millions of voters**.

---

### Key Idea

Because of these reasons:

> Sampling is the rule in statistics, not the exception.

---

# 5. Types of Sampling

Sampling can broadly be divided into two types.

---

# 1. Probability Sampling

In **probability sampling**, observations are selected **randomly**.

Every member of the population has a **known probability of being selected**.

Example:

Selecting **500 companies randomly** from 9000 publicly traded companies.

---

# 2. Non-Probability Sampling

In **non-probability sampling**, the researcher **selects observations based on specific criteria**.

Example:

Selecting the **largest 500 companies**.

This is exactly how the **S&P 500** index is created.

This sample is **not random**.

---

# 6. Types of Random Sampling

There are **three major types** of random sampling.

---

# 1. Simple Random Sampling

Every observation has an **equal chance of being selected**.

Example:

Out of 9000 companies:

Randomly pick **500 companies**.

Problem:

The sample might contain:

* Too many companies from one sector
* Too few from another

---

# 2. Stratified Random Sampling

To fix that problem, we use **stratified sampling**.

Steps:

1. Divide population into **groups (strata)**
   Example: Industries

2. Sample **from each group**

Example:

9000 companies → divide into industries:

* Tech
* Utilities
* Healthcare
* Finance

Then randomly select companies **from each industry**.

This ensures **better representation of the population**.

---

# 3. Cluster Sampling

Cluster sampling works differently.

Steps:

1. Divide population into **clusters**
2. Randomly choose some clusters
3. Study **all observations within selected clusters**

Example:

Divide companies by **first letter of company name**.

Clusters:

A, B, C, ... Z

Randomly pick 5 letters:

A, E, G, N, T

Then study **all companies starting with those letters**.

---

### Problem with Cluster Sampling

Clusters may not represent the population well.

Example:

Some letters may have very few companies.

Therefore cluster sampling is **rarely used in finance**.

---

# 7. Sampling Bias

The biggest danger in sampling is **bias**.

### Bias

Bias occurs when the sample **systematically misrepresents the population**.

If bias exists:

> Your conclusions about the population can be completely wrong.

Bias is extremely difficult to eliminate.

---

# 8. Sources of Sampling Bias

There are **four major sources of bias**.

---

# 1. Exclusion Bias

Some members of the population **never appear in the sampling frame**.

Example:

Suppose you sample people using a **phone directory**.

Who gets excluded?

People who **do not have phones**.

Therefore the sample **does not represent the population**.

---

# 2. Self-Selection Bias

Occurs when **people choose whether to participate**.

Example:

Suppose you want to measure **virus infection rates**.

You open clinics and allow people to **come voluntarily for testing**.

Result:

People who feel sick are **more likely to show up**.

Therefore the infection rate in the sample will be **higher than in the population**.

---

# 3. Non-Response Bias

Occurs when **people refuse to respond**.

Example: Election polls

Pollsters call voters.

But some voters **refuse to answer**.

If one political group refuses to respond more often, the sample becomes **biased**.

---

# 4. Survivorship Bias

This is very common in finance.

Example:

Researchers ask:

**Do hedge funds beat the market?**

Early studies looked only at **hedge funds that currently exist**.

But they ignored hedge funds that **failed and shut down**.

Since failed funds disappear from the data, the results become biased.

This is called **survivorship bias**.

---

# 9. What Can You Do If Bias Exists?

There are three possibilities.

---

## 1. Ignore the Bias (Worst Case)

If researchers ignore bias, they may make **completely wrong conclusions**.

This is the most dangerous situation.

---

## 2. Adjust the Data

Sometimes researchers can **correct for bias**.

Example:

Pollsters may adjust results to account for **non-respondents**.

---

## 3. Narrow the Conclusion

If bias cannot be fixed, researchers should **limit their conclusions**.

Example:

Instead of saying:

"All U.S. companies are paying lower dividends."

Say:

"Large U.S. companies (S&P 500) are paying lower dividends."

---

# 10. Sampling Noise (Sampling Error)

Even when there is **no bias**, samples still contain **random error**.

This is called **sampling noise**.

---

### Example: Coin Toss

Suppose you toss a **fair coin**.

True probability:

* Heads = 50%
* Tails = 50%

Now perform **100 tosses**.

Possible results:

* 52 heads, 48 tails
* 46 heads, 54 tails

Rarely exactly **50-50**.

This difference is **sampling noise**.

---

# 11. How to Reduce Sampling Error

The solution is simple:

> Increase sample size.

Example:

| Sample Size | Accuracy         |
| ----------- | ---------------- |
| 100         | moderate error   |
| 1,000       | smaller error    |
| 10,000      | very small error |

As sample size increases:

**Sample results approach population truth.**

---

# 12. IID Assumption

In statistics we often assume samples are **IID**.

IID means:

### Independent and Identically Distributed

---

# Independence

Two events are independent if **one event does not influence another**.

Example:

Coin tosses

Result of previous toss **does not affect next toss**.

---

# Identical Distribution

All observations come from the **same probability distribution**.

Example:

Coin toss

Each toss has the same probability:

Heads = 50%
Tails = 50%

---

### Problem in Finance

Financial data often **violates IID assumptions**.

Example:

Stock returns:

* Economic conditions change
* Markets evolve
* Companies change

Therefore returns may **not be independent or identically distributed**.

---

# 13. Law of Large Numbers

This is one of the most important laws in statistics.

### Law of Large Numbers

As sample size increases:

> Sample average → Population average

Example:

Population mean height = 170 cm

Sample sizes:

| Sample Size | Sample Mean |
| ----------- | ----------- |
| 10          | 168 cm      |
| 100         | 169 cm      |
| 1000        | 170 cm      |

Larger samples produce **more accurate estimates**.

---

# 14. Central Limit Theorem (CLT)

The **Central Limit Theorem** is extremely powerful.

It states:

> Even if the population distribution is NOT normal, the distribution of sample means will become normal as sample size increases.

Why is this important?

Because **normal distributions allow powerful statistical analysis**, such as confidence intervals.

---

# 15. Chebyshev’s Inequality

Even if the distribution is **not normal**, we can still make statements.

Chebyshev's Inequality states:

At least:

| Range                 | Minimum observations |
| --------------------- | -------------------- |
| 2 standard deviations | 75%                  |
| 3 standard deviations | 89%                  |

So even without normal distribution, we know **most observations lie close to the mean**.

---

# 16. Final Summary

Key takeaways from this session:

### 1. Statistics relies heavily on sampling.

### 2. We sample because of:

* Practicality
* Cost
* Time

### 3. Sampling methods include:

* Simple random sampling
* Stratified sampling
* Cluster sampling

### 4. Major risks in sampling:

* Bias
* Sampling error

### 5. Important statistical concepts:

* IID assumption
* Law of Large Numbers
* Central Limit Theorem
* Chebyshev’s Inequality

---

✔ Sampling allows us to **draw conclusions about very large populations** using relatively small data sets.

But we must always be careful about:

* **Bias**
* **Sampling noise**
* **Assumptions**

# Statistics Lecture Notes

## Session — Sampling Issues in Finance & Investing

---

# 1. Introduction

In this session we examine an important practical topic:

> **Sampling problems that arise in finance and investing.**

Unlike textbook statistics where samples may be random, in finance **most samples are systematic and non-random**.

This means when we analyze financial data, we must always ask:

* What sample was used?
* How was it chosen?
* Can we extrapolate the result to the broader market?

Sampling issues appear **everywhere in finance**, including:

* Market indices
* Historical market data
* Investment strategy backtesting
* Event studies

---

# 2. Market Indices as Samples

One of the most common examples of sampling in finance is **stock market indices**.

When we want to understand how the **stock market is performing**, we often look at an **index**.

Examples include:

* Dow Jones Industrial Average
* S&P 500
* NASDAQ
* Total Market Index

---

### Important Idea

> **Every market index is a sample of the stock market.**

Why?

Because no index contains **all publicly traded companies**.

Even large indices represent **only a subset of stocks**.

---

# 3. Indices Are NOT Random Samples

Market indices are **non-random samples**.

Stocks are selected using **specific criteria decided by humans**.

These criteria determine:

1️⃣ Which stocks are included
2️⃣ How many stocks are included
3️⃣ How stocks are weighted

Because of this, indices represent **specific segments of the market**, not the entire market.

---

# 4. Index Size Matters

Indices differ in **number of constituent companies**.

Example in the US market:

| Index         | Number of Stocks  |
| ------------- | ----------------- |
| Dow Jones     | 30                |
| S&P 500       | 500               |
| NYSE Index    | All NYSE stocks   |
| NASDAQ        | All NASDAQ stocks |
| Wilshire 5000 | ~5000 stocks      |

Even the **Wilshire 5000** is still only a **sample**, because there are **more than 5000 public companies**.

---

# 5. Selection Criteria Matters

Different indices use different **stock selection rules**.

Example:

### S&P 500

Stocks are selected primarily based on **market capitalization**.

It includes:

> The 500 largest companies by market value.

---

### Dow Jones Industrial Average

The Dow contains **30 companies** selected by editors.

Selection is more **subjective**.

Companies are chosen based on:

* Size
* Reputation
* Longevity
* Industry representation

This makes the Dow a **non-systematic sample**.

---

# 6. Weighting of Stocks

Indices also differ in **how stocks are weighted**.

Two common approaches:

### 1. Market Cap Weighted

Larger companies have more influence.

Example:

* S&P 500
* NASDAQ

---

### 2. Equal Weighted

Each company has equal importance.

Example:

A stock with $1B market cap counts the same as a $500B company.

---

### Why Weighting Matters

Weighting determines **how well the index represents the market**.

Example:

If large tech companies dominate the index, the index will behave like **tech sector performance**.

---

# 7. Example: Market Divergence in 2020

In 2020, three major indices behaved very differently:

| Index     | Behavior             |
| --------- | -------------------- |
| Dow Jones | Moderate recovery    |
| S&P 500   | Strong recovery      |
| NASDAQ    | Very strong recovery |

All fell sharply in **Feb–Mar 2020**.

But afterward:

NASDAQ outperformed significantly.

---

### Why?

Because NASDAQ is **dominated by technology companies**, which performed extremely well during the pandemic.

This illustrates an important lesson:

> Different samples (indices) capture different parts of the market.

---

# 8. Passive Investing and Index Funds

Originally, indices were used only to **measure market performance**.

But around **40–50 years ago**, a new investment idea emerged:

### Passive Investing

Instead of trying to beat the market:

> Investors simply try to match the index.

---

### Major Passive Investment Vehicles

1️⃣ **Index Funds**
2️⃣ **Exchange Traded Funds (ETFs)**

Both attempt to **replicate the performance of an index**.

---

# 9. Growth of Passive Investing

Passive investing has grown dramatically.

Many investors now prefer:

* Lower fees
* Simpler strategies
* Market matching returns

As a result, passive funds have grown **at the expense of active funds** such as:

* Mutual funds
* Hedge funds

---

# 10. Sampling Inside ETFs and Index Funds

Even ETFs and index funds sometimes rely on **sampling**.

Example:

An ETF tracking the **entire stock market**.

The US market contains roughly:

> 9000 publicly traded companies.

But a fund may hold only **a sample of those stocks**.

Why?

Because buying all stocks is sometimes **impractical or expensive**.

---

### Example: Energy ETFs

Different energy ETFs claim to track **energy sector performance**.

But they may contain:

| ETF    | Number of Stocks |
| ------ | ---------------- |
| Fund A | 25 stocks        |
| Fund B | 40 stocks        |
| Fund C | 171 stocks       |

Because each fund samples differently:

> Their returns may differ significantly.

---

# 11. Time Sampling in Finance

Another type of sampling occurs across **time periods**.

Researchers often select **specific historical periods** for analysis.

This is called:

> **Time sampling**

---

### Reasons for Time Sampling

1️⃣ Data may not exist earlier
2️⃣ Data may be unreliable
3️⃣ Researchers want to study specific market conditions

---

### Example

Much US stock market research begins in **1926**.

Why?

Because earlier data is **difficult to obtain reliably**.

---

### Structural Break Example

Some researchers believe:

> The financial crisis of 2008 changed markets permanently.

Therefore they analyze only:

2008 – present

instead of longer historical periods.

---

# 12. Example: PE Ratio Analysis

A common way to judge market valuation is the **Price-Earnings (P/E) ratio**.

Standard formula:

[
PE = \frac{Price}{Earnings}
]

---

### Alternative PE Measures

Researchers sometimes modify earnings:

1️⃣ **Trailing PE**

Uses last 12 months earnings.

---

2️⃣ **Normalized PE**

Uses average earnings over 5–10 years.

Purpose:

Remove economic cycle effects.

---

3️⃣ **CAPE Ratio (Shiller PE)**

Uses:

* 10-year inflation-adjusted earnings.

This smooths long-term earnings fluctuations.

---

### Key Observation

When comparing PE ratios historically:

> Current PE ratios are higher than historical averages.

But that does **not automatically mean markets are overpriced**.

Why?

Because:

* Interest rates change
* Economic structure changes
* Risk levels change

---

# 13. Dangers of Using Historical Data

Investors frequently **extrapolate historical results into the future**.

But this can be dangerous because:

* Markets evolve
* Conditions change
* Relationships disappear

---

# 14. Example: Small Cap Premium

In 1981, researcher **Rolf Banz** discovered something surprising.

Small cap stocks earned **higher returns than expected**.

This became known as:

> **Small Cap Premium**

---

### Historical Evidence

From **1927–1980**:

Small cap stocks outperformed large caps by about:

> **6.56% annually**

This led to:

* Investment strategies favoring small caps
* Analysts adding small cap premiums to discount rates

---

# 15. But Did It Continue?

If we extend the data to **2020**:

Small cap premium drops to about:

> **3.5%**

If we look only at **1980–2020**:

Small caps actually **underperformed slightly**.

---

### Lesson

Long time periods can **hide structural changes**.

Just because something worked historically **does not mean it will continue**.

---

# 16. Backtesting Investment Strategies

Investors constantly test strategies such as:

* Buy low PE stocks
* Buy high dividend stocks
* Buy low price-to-book stocks

These strategies are tested using **historical data**.

This is called:

> **Backtesting**

---

# 17. Key Questions When Evaluating Backtests

When you see research claiming a strategy works, ask:

---

### 1. What Was the Universe of Stocks?

Did researchers examine:

* All stocks?
* Only NYSE stocks?
* Only S&P 500 companies?

Using a restricted universe introduces **sampling bias**.

---

### 2. How Were Metrics Calculated?

Example: PE ratios.

Researchers must explain:

* Which earnings measure was used
* How negative earnings were treated

Negative earnings companies are often **excluded**, which creates bias.

Better metric:

> **Earnings / Price ratio**

because it can include negative earnings.

---

### 3. Survivorship Bias

Did the study include companies that **failed or went bankrupt**?

Correct method:

Start with all companies **existing at the start of the period**.

Incorrect method:

Start with companies **still alive at the end**.

---

### 4. Timing Bias

Results can depend heavily on **starting year**.

Example:

Value stocks performed best in the **1970s**.

A study starting in **1970** will find strong value effects.

A study starting in **1980** may find weaker results.

---

# 18. Event Studies

Another area where sampling is critical is **event studies**.

---

### Event Study Definition

An event study examines **stock price reactions around specific events**, such as:

* Acquisitions
* Earnings announcements
* Bankruptcy filings

---

### Example: Acquisition Study

Researchers examine returns around **acquisition announcements**.

Typical result:

Stock returns increase:

* Slightly before the announcement
* On the announcement date
* Then stabilize afterward

---

### Important Questions

Before trusting such results, ask:

1️⃣ How many events were studied?
2️⃣ Over what time period?
3️⃣ Were events randomly selected?
4️⃣ Is the announcement date accurate?

---

### Information Leakage Problem

Sometimes markets know about an acquisition **before the official announcement**.

If so:

Returns may rise **before the event date**.

This complicates statistical interpretation.

---

# 19. Key Takeaway

Sampling issues are **everywhere in finance**.

Most financial samples are:

* Non-random
* Systematically constructed
* Subject to bias

Therefore whenever you see research findings, always ask:

* How was the sample constructed?
* Is there survivorship bias?
* Is there timing bias?
* Can the results be generalized?

---

# 20. Final Insight

If research is based on **S&P 500 companies**, you can safely extrapolate to:

> **Large US companies**

But not necessarily to:

* Small companies
* International stocks
* Private companies

---

# Final Summary

Sampling in finance appears in:

1️⃣ Market indices
2️⃣ Index funds and ETFs
3️⃣ Historical market data
4️⃣ Investment strategy research
5️⃣ Event studies

Because financial samples are rarely random:

> Careful interpretation of results is essential.
