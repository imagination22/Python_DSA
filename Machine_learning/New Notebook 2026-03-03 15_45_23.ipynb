A **model** is something that allows you to make predictions.

Example:
You see dark clouds.
Your brain predicts → “It will probably rain.”

So remember this:

> 🧠 Intelligence = Ability to build models that predict reality.

There are two ways:

# 2️⃣ What is Machine Learning?

> Machine Learning = Getting computers to do things **without explicitly programming every step**.

* Human writes rules

# 3️⃣ Supervised Machine Learning

Machine Learning has two main phases:

## Phase 1: Training

We:

* Collect data (inputs + outputs)
* Feed it into a learning algorithm
* Produce a model

## Phase 2: Inference

We:

* Give the model new data
* It makes predictions

---

## 📌 Example: Linear Regression

Suppose we want to predict tomorrow’s temperature.

We use:

[
y = mx + b
]

Where:

* x = today’s temperature
* y = tomorrow’s temperature
* m, b = parameters

During inference:
We plug in x → get prediction y.

---

## But How Do We Find m and b?

This happens during **training**.

We:

1. Compare prediction vs real value
2. Measure difference using a **Loss Function**
3. Find parameter values that minimize loss

Mathematically:

[
\theta = (X^TX)^{-1}X^Ty
]

But don’t get lost in math.

### The Big Idea:

> We use data + math to align model predictions with reality.

That’s Machine Learning.

# 6️⃣ Anatomy of a Neural Network

## The Basic Unit: Neuron

A neuron does 3 things:

1. Multiply inputs by weights
2. Add bias
3. Apply activation function

Mathematically:

[
z = g(\sum w_i x_i + b)
]

Key difference from linear regression?

👉 Activation function = non-linearity.

Without non-linearity → network becomes just linear model.

---

## Common Activation Functions

* ReLU
* Sigmoid
* Tanh
* Softmax (outputs probability distribution)

---

## Network Architecture

Neurons → Layers → Networks

Common architectures:

* Feedforward Network
* Recurrent Neural Network (RNN)
* Convolutional Neural Network (CNN)
* Transformer (used in LLMs)

Modern LLMs use Transformers.

# 7️⃣ Training Neural Networks

Unlike linear regression, we cannot solve neural networks in closed form.

Loss landscape looks messy.

We use:

## Gradient Descent

Update rule:

[
\theta_{new} = \theta_{old} - \gamma \nabla L
]

Where:

* Gradient = direction
* Learning rate (γ) = step size

---

## Types of Gradient Descent

1. Full Batch (slow, unstable)
2. Stochastic (noisy)
3. Mini-batch (most common)
4. Adam Optimizer (most widely used today)

Adam = Adaptive Momentum Estimation.

It improves stability and speed.

---

* Model Parameters → learned from data
* Hyperparameters → set by us

Examples:

* Learning rate
* Batch size
* Epochs
* Dropout rate

Dropout helps prevent overfitting.

Instead of minimizing loss:

👉 Maximize reward.

Update rule becomes:

[
\theta_{new} = \theta_{old} + \gamma \nabla J
]

Notice:
Plus sign (gradient ascent).

## 🔧 ML vs Normal Code

### Normal Code
- You write **explicit rules**.  
- Example:  
  ```python
  if age > 18:
      print("Adult")
  else:
      print("Child")
  ```
- The program only does what you tell it. It **cannot improve** unless you change the code.

### Machine Learning (ML)
- You don’t write rules. Instead, you give the computer **examples** (data).  
- The computer finds patterns and creates its own rules (a **model**).  
- Example:  
  - Input: house size, bedrooms, location.  
  - Output: house price.  
  - You give 1000 examples → ML learns the relationship.  
- ML code **adapts** as it sees more data. That’s what makes it “learn.”

---

## 🧠 What Is a Model?
- A **model** is just a mathematical formula inside code.  
- In ML, the model’s **parameters** (like weights) are adjusted automatically during training.  
- Example: Linear regression model:  
  \[
  \text{price} = m \cdot \text{size} + c
  \]  
  Training finds the best \(m\) and \(c\).

---

## 📊 Classification vs Regression
- **Classification** = predict categories (spam vs not spam, cat vs dog).  
- **Regression** = predict numbers (house price, temperature).  
- Both are ML tasks.

---

## 🧱 Deep Learning (DL)
- A subset of ML using **neural networks**.  
- Learns features automatically (no manual feature engineering).  
- Handles complex data like images, speech, text.  
- Needs **big data + GPUs**.  

---

## 🕹️ Reinforcement Learning (RL)
- Learns by **trial and error**.  
- Gets **rewards/punishments** for actions.  
- Example: A robot learns to walk by trying moves and adjusting.  
- Different from ML/DL, which learn from datasets.

---

## 📚 Large Language Models (LLMs)
- A type of **deep learning model** trained on massive text data.  
- Learns grammar, facts, reasoning, and patterns in language.  
- Examples: GPT, Claude, Gemini.  
- Often fine-tuned with **RLHF (Reinforcement Learning from Human Feedback)** to make responses more helpful and safe.

---

## 🔤 Tokens in LLMs
- LLMs don’t read text as whole sentences — they break it into **tokens** (chunks of text).  
- A token can be a word, part of a word, or even punctuation.  
- Example:  
  - Sentence: “Machine learning is fun.”  
  - Tokens: [“Machine”, “learning”, “is”, “fun”, “.”]  
- The model predicts the **next token** step by step, which is how it generates text.

---

## ✅ Summary Table

| Aspect              | Normal Code | ML | DL | RL | LLM |
|---------------------|-------------|----|----|----|-----|
| Rules               | Hardcoded   | Learned from data | Learned automatically | Learned by trial/error | Learned from massive text |
| Data size           | Small       | Small/medium | Large | Experiences | Huge text datasets |
| Adaptability        | None        | Improves with data | Improves with data | Improves with rewards | Improves with fine-tuning |
| Example             | Age check   | House price prediction | Image recognition | Robot walking | ChatGPT |

---

👉 So:  
- **ML vs normal code** = ML learns rules from data, normal code follows fixed rules.  
- **DL** = ML with neural networks, good for complex data.  
- **RL** = learns from experience, not just data.  
- **LLM** = a deep learning model for language, trained on tokens, often fine-tuned with RLHF.

A mathematical procedure that:

* Minimizes loss (or maximizes reward)

* Makes predictions

# 🧠 PART 1 — MACHINE LEARNING (Traditional ML Algorithms)

These are mostly used for **tabular/structured data**.

---

# 1️⃣ Linear Regression

### 🎯 Use When:

* Target is continuous
* Relationship is roughly linear
* You want interpretability

Example:

* House price prediction
* Revenue forecasting

### ⚙ How It Works:

We assume:

[
y = w^Tx + b
]

Goal:
Minimize Mean Squared Error.

Closed-form solution:

[
\theta = (X^TX)^{-1}X^Ty
]

Or use gradient descent.

### 🚨 Not Good When:

* Non-linear relationships
* Complex patterns
* Images/text

---

# 2️⃣ Logistic Regression

### 🎯 Use When:

* Binary classification
* You want probabilities
* Data is linearly separable-ish

### ⚙ How It Works:

[
p = \sigma(w^Tx + b)
]

Where sigmoid squashes output between 0 and 1.

Loss = Cross Entropy.

### 🚨 Not Good When:

* Complex decision boundary
* High-dimensional unstructured data

---

# 3️⃣ Decision Tree

### 🎯 Use When:

* Interpretability is important
* Non-linear relationships
* Mixed data types

### ⚙ How It Works:

Algorithm:

1. Try all possible splits
2. Choose split that maximizes information gain
3. Repeat recursively

Metrics:

* Gini
* Entropy

### 🚨 Weakness:

* Overfitting
* High variance

---

# 4️⃣ Random Forest

### 🎯 Use When:

* Tabular data
* Want strong baseline
* Avoid overfitting

### ⚙ How It Works:

* Build many trees
* Each tree sees random subset of data + features
* Average predictions

Reduces variance.

---

# 5️⃣ Gradient Boosting (XGBoost, LightGBM)

### 🎯 Use When:

* Tabular data
* Kaggle competitions
* Maximum performance needed

### ⚙ How It Works:

Sequential training:
Each new tree fixes previous tree’s errors.

Additive model:

[
F_{new} = F_{old} + h(x)
]

Very powerful.

---

# 🧠 PART 2 — DEEP LEARNING ALGORITHMS

Important:

Neural network is a **model**.
Training method is the **algorithm**.

---

# 1️⃣ Gradient Descent

### 🎯 Use When:

Small dataset

### ⚙ How It Works:

[
\theta = \theta - \gamma \nabla L
]

Uses full dataset each step.

Slow + unstable.

Rare in practice.

---

# 2️⃣ Stochastic Gradient Descent (SGD)

### 🎯 Use When:

Large datasets

### ⚙ How It Works:

* Pick 1 data point
* Compute gradient
* Update

Noisy but faster.

---

# 3️⃣ Mini-Batch Gradient Descent

### 🎯 Use When:

Almost always

### ⚙ How It Works:

* Pick small batch (32, 64, 128)
* Compute gradient
* Update

Best trade-off.

---

# 4️⃣ Adam (Adaptive Momentum Estimation)

Most used today.

### 🎯 Use When:

* Training deep networks
* Transformers
* Complex models

Used in systems like
ChatGPT

### ⚙ How It Works:

Combines:

* Momentum (smooth updates)
* Adaptive learning rate per parameter

Stable + fast.

---

# 5️⃣ RMSProp

Similar to Adam but simpler.

Good for:

* RNN training

---

# 🧠 PART 3 — REINFORCEMENT LEARNING ALGORITHMS

Here the objective is:

Maximize Reward, not minimize loss.

---

# 1️⃣ Q-Learning (Value-Based)

### 🎯 Use When:

* Discrete action space
* Small environments

### ⚙ How It Works:

Learn Q(s,a):

Expected reward of action a in state s.

Update rule:

[
Q(s,a) = Q(s,a) + \alpha [r + \gamma \max Q(s',a') - Q(s,a)]
]

Good for:

* Games
* Simple robotics

---

# 2️⃣ Policy Gradient (REINFORCE)

### 🎯 Use When:

* Continuous action space
* Complex policies

### ⚙ How It Works:

Directly optimize:

[
J(\theta) = E[\log \pi_\theta(a|s) R]
]

Update:

[
\theta = \theta + \gamma \nabla J
]

High variance.

---

# 3️⃣ PPO (Proximal Policy Optimization)

Most practical RL algorithm today.

Used in RLHF training for modern systems.

### 🎯 Use When:

* Large neural policies
* LLM alignment
* Complex environments

### ⚙ How It Works:

Adds clipping to prevent large policy updates.

Improves stability.

---

# 4️⃣ Actor-Critic

Hybrid method.

Actor:

* Chooses action

Critic:

* Evaluates action

More stable than pure policy gradient.

---

# 🔥 RL Algorithm Selection Rule

| ------------------ | --------------- |
| Small discrete     | Q-Learning      |
| Continuous control | Policy Gradient |
| Large-scale RL     | PPO             |
| LLM Alignment      | PPO + variants  |
